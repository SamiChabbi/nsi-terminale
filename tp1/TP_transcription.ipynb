> ### <u>Cours — Gérer une erreur avec try / except</u>
>
> **Cours :**
>
> Quand une instruction provoque une erreur (une « exception »), le programme s'arrête immédiatement en affichant un message d'erreur. Par exemple, ouvrir un fichier qui n'existe pas encore :
>
> ```python
> fichier = open("scores.csv", "r", encoding="utf-8")
> ```
>
> provoque une erreur FileNotFoundError si le fichier n'existe pas, et arrête le programme.
> Le mot-clé try (« essayer ») permet de tenter d'exécuter des instructions, et except (« sauf si ») permet de définir ce qu'il faut faire si une erreur précise se produit, sans arrêter le programme :
>
> ```python
> try:
>     fichier = open("scores.csv", "r", encoding="utf-8")
>     contenu = fichier.readlines()
>     fichier.close()
> except FileNotFoundError:
>     contenu = []    # le fichier n'existe pas encore
> ```
>
> Si le fichier n'existe pas, Python « attrape » l'erreur FileNotFoundError, exécute le bloc except (contenu = []) au lieu de s'arrêter, et le programme continue normalement.

> ### <u>Cours — Le tri à bulle</u>
>
> **Cours :**
>
> Le tri à bulle est une méthode simple pour trier une liste :
>
> - on parcourt la liste en comparant chaque élément à son voisin de droite ;
> - si les deux éléments ne sont pas dans le bon ordre, on les échange ;
> - on répète ce parcours : à chaque parcours complet, le plus grand élément restant « remonte » à sa place, comme une bulle qui remonte à la surface ;
> - on effectue autant de parcours que d'éléments dans la liste (au pire).
>
> Exemple sur la liste [5, 1, 4, 2] :
>
> | Étape | Liste avant | Comparaison | Action | Liste après |
> |:---|:---|:---|:---|:---|
> | i=0, j=0 | [5, 1, 4, 2] | 5 > 1 | on échange | [1, 5, 4, 2] |
> | i=0, j=1 | [1, 5, 4, 2] | 5 > 4 | on échange | [1, 4, 5, 2] |
> | i=0, j=2 | [1, 4, 5, 2] | 5 > 2 | on échange | [1, 4, 2, 5] |
> | i=1, j=0 | [1, 4, 2, 5] | 1 > 4 ? non | rien | [1, 4, 2, 5] |
> | i=1, j=1 | [1, 4, 2, 5] | 4 > 2 | on échange | [1, 2, 4, 5] |
> | i=2, j=0 | [1, 2, 4, 5] | 1 > 2 ? non | rien | [1, 2, 4, 5] |
>
> Ce qui se traduit en Python par deux boucles imbriquées :
>
> ```python
> n = len(liste)
> for i in range(n):
>     for j in range(n - 1 - i):
>         if liste[j] > liste[j + 1]:
>             temporaire = liste[j]
>             liste[j] = liste[j + 1]
>             liste[j + 1] = temporaire
> ```
