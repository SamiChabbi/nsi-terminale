# Leçon 02D : Les Piles

## 1) Définition

Le TDA **Pile** est une structure de données fondée sur le principe <u>LIFO</u> pour « <u>L</u>ast In, <u>F</u>irst Out », soit en français, « dernier arrivé, premier sorti ».
Il s'agit d'une structure linéaire dans laquelle on aura toujours un accès direct au dernier élément entré.

*[Schémas : un empilement avec « in » / « out », puis une suite d'états de la pile au cours du temps t : Empiler « Push » 1, 2, 3, 4 puis Dépiler « Pop » 4, 3, 2.]*

```
Empiler 1 → [1]
Empiler 2 → [1, 2]
Empiler 3 → [1, 2, 3]
Empiler 4 → [1, 2, 3, 4]
Dépiler   → [1, 2, 3]   (4 sort)
Dépiler   → [1, 2]      (3 sort)
Dépiler   → [1]         (2 sort)
```

<u>**Exemple**</u> Une pile d'assiettes.
On ajoute des assiettes sur la pile, et on récupère la dernière assiette ajoutée.

## 2) Primitives

```python
P = cree_pile() # retourne une instance nommée P de type Pile
empile(P,e)     # l'element e est empile au sommet de P
depile(P)       # retourne et supprime l'element situe au sommet de P
est_vide(P)     # retourne Vrai si la Pile est pleine, Faux sinon
```

> *Remarque : la feuille écrit `cree_pile()` ici et `creer_pile()` dans les exercices : c'est la même primitive. Et `est_vide` renvoie Vrai quand la pile est **vide** (« pleine » est une coquille).*

## 3) Applications

- ✓ Dans un navigateur web, une pile sert à mémoriser les pages Web visitées.
  L'adresse de chaque nouvelle page visitée est empilée
  L'option « Afficher la page précédente » consiste à dépiler la dernière adresse.
- ✓ De la même façon, un éditeur empile toutes les instructions réalisées.
  L'option « Annuler la frappe » (en anglais « <u>Undo</u> ») consiste à dépiler la dernière commande.
- ✓ On verra plus tard, que pour évaluer des expressions mathématiques, on utilise aussi une pile pour gérer les priorités opératoires.
- ✓ On verra aussi les fonctions récursives (une fonction qui fait appel à elle-même).
  Le compilateur utilise également une pile pour gérer ces appels successifs.
