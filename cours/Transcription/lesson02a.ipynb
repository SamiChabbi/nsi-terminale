# Leçon 02A : Structures linéaires — TDA -- TDC

## Partie A : Généralités

Une structure de données est une organisation logique des données qui permet de simplifier et d'accélérer leur traitement.

### a) Types de données natifs

Les types de données natifs sont ceux disponibles dans tous les langages de programmation courants. Par exemple : les entiers, les flottants et les booléens

### b) Type de données abstrait (TDA)

Lorsque l'on écrit un programme, on manipule des données sur lesquelles on va réaliser un certain nombre d'opérations. Un algorithme doit être autant que possible indépendant d'une implémentation dans un langage particulier. C'est pour cela que l'on va raisonner sur un TDA.

<u>Exemple</u>

Une université gère ses étudiants à l'aide d'une base de données.
Chaque enregistrement est alors constitué d'informations concernant par exemple le nom, la date de naissance et le cursus d'un élève.

- ☐ Ces enregistrements sont les données.
- ☐ Les opérations : Lire et modifier des informations, créer ou supprimer un étudiant

> **<u>A retenir</u>**
>
> On appelle type de données abstrait (TDA), la formulation mathématique des données et des opérations que l'on doit leur appliquer. Les 4 primitives (opérartions) élémentaires, sont :
>
> - ☐ **`Create`** : créer une nouvelle donnée
> - ☐ **`Read`** : lire une donnée
> - ☐ **`Update`** : modifier une donnée
> - ☐ **`Delete`** : supprimer une donnée

### c) Type de données concret (TDC)

Pour écrire un programme dans un langage particulier, il faut implémenter les <u>TDA</u>. On parlera alors de type de données concret, noté <u>TDC</u>. Dans un même langage, il peut exister différentes implémentations d'un même <u>TDA</u>.

### d) Les structures linéaires

Une **structure de données linéaire** est une structure dans laquelle on peut ranger ses éléments les uns derrière les autres, comme des wagons d'un train ou des personnes dans une file.
Cette définition s'oppose aux structures **non linéaires** (arbres, graphes), où un élément peut avoir plusieurs successeurs ou prédécesseurs. Notions que nous aborderont plus tard dans l'année.

Nous allons étudier différentes familles de structures linéaires :

- ☐ Les listes (vues en première)
- ☐ Les dictionnaires (vus en première)
- ☐ Les Files
  Ces structures correspondent aux files d'attente, le premier arrivé est le premier servi.
- ☐ Les Piles
  Ces structures correspondent aux piles d'assiettes, on a accès à la dernière assiette posée.
- ☐ Les listes chainées
  Ces structures enchainent des maillons qui se suivent. On doit parcourir toute la chaine por accéder à un maillon.
