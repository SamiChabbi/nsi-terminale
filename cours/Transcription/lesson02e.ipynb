# Leçon 02E : Structures linéaires — Les Files

## 1) Définition

La TDA **File** est une structure de données fondée sur le principe <u>FIFO</u> pour « <u>F</u>irst In, <u>F</u>irst Out », soit en français « premier arrivé, premier sorti ». Il s'agit d'une structure linéaire dans laquelle on aura toujours un accès direct au premier élément entré.

*[Illustration : une file de personnes — « Premier arrivé, Premier servi » à l'avant, « Dernier arrivé, Dernier servi » à l'arrière.]*

<u>**Exemple :**</u> Une file d'attente
Une file clients qui attendent à la caisse est la meilleure image pour décrire cette structure.
Le client qui passe en premier est le premier arrivé.

<u>**Primitives**</u>

```python
F = creer_file() # retourne une instance nommée F de type File
enfile(F,e)      # l'element e est ajoute a la queue de la File
defile(F)        # supprime et retourne l'element situe en tete de la File
est_vide(F)      # retourne Vrai si la File est pleine, Faux sinon
```

> *Remarque : `est_vide` renvoie Vrai quand la file est **vide** (« pleine » est une coquille de la feuille).*

## 2) Applications

- ✓ Files d'accessibilité à des ressources partagées, par exemple une imprimante.
- ✓ Messages en attente dans un commutateur de réseau comme Transpac
- ✓ Travaux en attente d'exécution dans un système de traitement par lots.
- ✓ On verra plus tard lors de l'étude des arbres et des graphes, un algorithme de parcours en largeur qui utilise une File.

## 3) Un TDC File en Python avec une classe utilisant une liste

````python
class File:
    def __init__(self): self.valeurs = []
    def enfile(self, valeur): self.valeurs.append(valeur)
    def defile(self):
        if self.valeurs:  return self.valeurs.pop(0)
    def est_vide(self):  return self.valeurs == []
    def longueur(self):  return len(self.valeurs)
    def __str__(self):
        ch = ""
        for x in self.valeurs:
            ch = ch + " <- " +  str(x)
        return  ch
````

````python
# Essai de la classe (cette cellule n'est pas sur la feuille)
F = File()
F.enfile(1)
F.enfile(2)
F.enfile(3)
print(F)
print(F.defile())
print(F, "| longueur :", F.longueur(), "| vide ?", F.est_vide())
````
