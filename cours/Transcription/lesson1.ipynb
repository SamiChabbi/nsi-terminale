# Leçon 01 : POO (programmation orientée objet) - Les classes

## 1) Introduction

Les classes sont des structures adaptées à la manipulation de données composées de plusieurs éléments et de natures différentes. Tout au long de l'année, nous allons utiliser des classes pour implémenter les structures telles que les piles, les files, les arbres ....

## 2) Création d'une classe

Une classe définit une structure de données qui vient s'ajouter aux structures de base du langage. La structure ainsi définie peut regrouper plusieurs composantes de natures variées.
Chacune de ces composantes est appelée **attribut**. On la définit avec la syntaxe suivante :

```python
class Nom_classe :
    def __init__(self,parametres):
        self.attributs = parametres
```

Nous détaillerons le constructeur `__init__` plus tard. Cette méthode possède un premier paramètre `self` obligatoire puis des paramètres optionnels.

### Exemple

````python
class Eleve :
    def __init__(self,n,a):
        self.nom,self.age = n, a
````

Elle permet de manipuler des données sur des élèves. Elle a deux attributs **`nom`** et **`age`**.

## 3) Création d'un objet

Une fois la classe définie, on peut créer une <u>instance</u>, on parle aussi d'objet, c'est-à-dire un représentant de cette classe avec la syntaxe suivante :

```python
instance = Nom_classe(parametres)
```

### Exemple

On peut créer deux <u>instances</u> **`e1`** et **`e2`** de la classe **`Eleve`** :

````python
e1, e2 = Eleve('Bob',13), Eleve('Léa',15)
````

Tout s'écrit comme si le nom de la classe était également celui d'une fonction prenant deux paramètres avec lesquels elle construit l'instance.

## 4) Manipulation des attributs

On accède à l'attribut d'une instance **`inst`** de la classe avec la syntaxe suivante

```python
inst.attribut
```

Les attributs sont mutables en Python, c'est-à-dire qu'on peut les modifier.
En python, les types natifs chaines de caractères, listes et dictionnaires sont des classes.

### Exemple

> *Remarque : sur la feuille, la 2ᵉ ligne est écrite `print(e1.age, e2.moyenne)`. La classe `Eleve` n'a pas d'attribut `moyenne` (le code planterait) ; l'affichage attendu `13 15` montre qu'il faut lire `e2.age`. C'est cette version qui est utilisée ci-dessous.*

````python
e1, e2 = Eleve('Bob',13), Eleve('Léa',15)
print(e1.age, e2.age)
e1.age = e1.age + 1
print(e1.age, e2.age)
````

affiche dans la console :

```
13 15
14 15
```
