## 1) La classe `list` (native)

### Créer et accéder aux éléments

| Syntaxe | Rôle |
|:---|:---|
| `Liste = []` | Crée une liste vide |
| `Liste[i]` | Renvoie l'élément d'indice i (le premier élément a pour indice 0) |
| `Liste[-1]` | Renvoie le dernier élément d'une liste. |
| `Liste[-2]` | Renvoie l'avant dernier élément d'une liste. |
| `Liste[p:q]` | Renvoie la liste des éléments d'indices p à q - 1 |
| `Liste[:q]` | Renvoie la liste des éléments d'indices 0 à q - 1 |
| `Liste[p:]` | Renvoie la liste de tous les éléments à partir de l'indice p |

````python
# Illustration des lignes du tableau « Créer et accéder aux éléments »
Liste = []
Liste = [10, 20, 30, 40, 50]
print(Liste[1])
print(Liste[-1])
print(Liste[-2])
print(Liste[1:3])
print(Liste[:3])
print(Liste[2:])
````

### Méthodes

| Syntaxe | Rôle |
|:---|:---|
| `print(Liste)` | Affiche la liste |
| `Liste1 + Liste2` | Concatène les deux listes. Exemple : [1,2] + [4,5] → [1,2,4,5]. |
| `Liste.append(e)` | Ajoute e à la fin de la liste |
| `Liste.insert(i,elem)` | Ajoute e dans la liste au rang i de la liste, le reste étant décalé. |
| `Liste.remove(elem)` | Supprime la première occurrence de l'élément e de la liste |
| `Liste.pop()` | Supprime le dernier élément de la liste |
| `Liste.pop(i)` | Supprime le l'élément d'indice i de la liste |
| `Liste.sort()` | Ordonne tous les éléments de la liste |
| `Liste.reverse()` | Inverse l'ordre des éléments de la liste |
| `Liste.index(elem)` | Renvoie l'indice du premier élément de la liste ayant la valeur e |
| `Liste.count(elem)` | Renvoie le nombre d'éléments de la liste ayant la valeur e |

````python
# Illustration des lignes du tableau « Méthodes »
Liste = [3, 1, 2]
print(Liste)
print([1,2] + [4,5])
Liste.append(5)
print(Liste)
Liste.insert(1, 7)
print(Liste)
Liste.remove(7)
print(Liste)
Liste.pop()
print(Liste)
Liste.pop(0)
print(Liste)
Liste = [4, 2, 9, 2]
Liste.sort()
print(Liste)
Liste.reverse()
print(Liste)
print(Liste.index(2))
print(Liste.count(2))
````

### Fonctions particulières

| Syntaxe | Rôle |
|:---|:---|
| `len(Liste)` | Renvoi le nombre d'éléments de la liste |
| `min(Liste)` | Renvoie le plus petit élément de la liste |
| `max(Liste)` | Renvoie le plus grand élément de la liste |
| `choice(Liste)` | Choisit au **hasard** un élément de la liste Liste. (module random) |
| `shuffle(Liste)` | **Mélange** les éléments d'une liste. (module random) |

````python
# Illustration des lignes du tableau « Fonctions particulières »
from random import choice, shuffle
Liste = [4, 8, 1, 6]
print(len(Liste))
print(min(Liste))
print(max(Liste))
print(choice(Liste))   # résultat aléatoire
shuffle(Liste)         # résultat aléatoire
print(Liste)
````
