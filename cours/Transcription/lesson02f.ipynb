# Leçon 02F : Les listes chaînées

La classe Maillon a deux attributs **`valeur (float)`** et **`suivant (Maillon)`**.
La liste chainée est alors une suite de maillons dont chacun, excepté le dernier pointe vers le suivant. On considèrera que la chaine est vide si la valeur du premier maillon est None.

*[Schéma : → [ 3.6 | • ] → [ 5.4 | • ] → [ 6.0 | • ] → None — chaque maillon contient une valeur (case verte) et un lien (case orange) vers le maillon suivant ; le dernier lien pointe vers None.]*

````python
class Maillon:
    def __init__(self, v = None):
        self.valeur = v
        self.suivant = None
    def __str__(self):
        affichage = ""
        curseur = self
        while not curseur == None :
            affichage =  affichage + str(curseur.valeur) + " -> "
            curseur = curseur.suivant
        return affichage + "None"
````

Bob souhaite utiliser la structure de données « liste doublement chaînée ».
Une telle liste est composée de maillons contenants chacun trois informations :
■ une valeur ■ un prédécesseur ■ un successeur qui sont tous deux des maillons.
Cette structure se prête bien au problème des bonbons : dans un collier, un bonbon est précédé et suivi par d'autres bonbons.
Le successeur du dernier bonbon est le premier et le prédécesseur du premier, le dernier.

*[Schéma : cinq maillons 0, 1, 2, 3, 4 côte à côte. Chaque maillon a une case `pred` à gauche, la valeur au centre et une case `succ` à droite. Les flèches `succ` vont de 0 vers 1, 1 vers 2, …, 3 vers 4, et une flèche en pointillés relie le `succ` de 4 au bonbon 0. Les flèches `pred` vont dans l'autre sens (1 vers 0, …, 4 vers 3), et une flèche en pointillés relie le `pred` de 0 au bonbon 4.]*

````python
class Bonbon:
    def __init__(self, valeur):
        self.pred = None # prédécesseur de ce bonbon
        self.valeur = valeur
        self.succ = None # successeur de ce bonbon
````
