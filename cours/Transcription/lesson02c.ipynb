## 1) La classe `dic` (native)

### Créer et accéder aux éléments

| Syntaxe | Rôle |
|:---|:---|
| `Dico = {}` | Crée une liste vide |
| `Dico[c]` | Renvoie l'élément de clé c (les clés sont des int ou str) |

### Méthodes et fonctions

| Syntaxe | Rôle |
|:---|:---|
| `print(Dico)` | Affiche le dictionnaire |
| `Dico[c] = e` | Modifie l'élément de clé c s'il existe et sinon l'ajoute. |
| `Dico.get(c)` | Renvoie l'élément de clé c s'il existe et None sinon. |
| `Dico.pop(c)` | Supprime l'élément de clé c |
| `Dico.keys()` | Renvoie la liste des clés. |
| `Dico.values()` | Renvoie la liste des valeurs. |
| `Dico.items()` | Renvoie la liste des paires (cle, valeur). |
| `len(Dico)` | Renvoi le nombre d'éléments du dictionnaire |

````python
# Illustration des lignes des tableaux
Dico = {}
Dico["a"] = 1
Dico["b"] = 2
print(Dico)
print(Dico["a"])
Dico["a"] = 10
print(Dico.get("a"))
print(Dico.get("z"))
Dico.pop("b")
print(Dico.keys())
print(Dico.values())
print(Dico.items())
print(len(Dico))
````

> *La cellule de code ci-dessus n'est pas sur la feuille : elle exécute simplement chaque ligne des tableaux.*
