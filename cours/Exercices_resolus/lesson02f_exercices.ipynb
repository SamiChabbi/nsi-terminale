# Rappel : la classe `Maillon` (donnée dans le cours)

````python
class Maillon:
    def __init__(self, v = None):
        self.valeur = v
        self.suivant = None
    def __str__(self):
        affichage = ""
        curseur = self
        while not curseur == None :
            affichage =  affichage + str(curseur.valeur) + " -> "
            curseur = curseur.suivant
        return affichage + "None"
````

#### 💡 EXPLICATION PÉDAGOGIQUE

**Ce qu'il faut comprendre avant de commencer :**
- Un **maillon** est un objet qui contient **deux attributs** : une `valeur` et un lien `suivant` vers le maillon d'après (ou `None` s'il n'y en a pas).
- Une **liste chaînée**, c'est le **premier maillon**. À partir de lui, on retrouve tous les autres en suivant les liens `suivant`, un par un. On ne peut **pas** sauter directement au 5e maillon comme avec `L[4]` dans une liste Python.
- **Convention de la feuille :** la chaîne est **vide** quand la `valeur` du premier maillon vaut `None`. Ainsi, `Maillon()` crée une chaîne vide.
- **Le schéma de parcours à retenir** (dans `__str__`) sert dans presque tous les exercices :

```python
curseur = self                 # on part du premier maillon
while not curseur == None:     # tant qu'on n'est pas sorti de la chaîne
    ...                        # on traite curseur.valeur
    curseur = curseur.suivant  # on passe au maillon suivant
```

**Erreur fréquente :** oublier `curseur = curseur.suivant`, ce qui donne une **boucle infinie**.

## Question 2

#### 📝 ÉNONCÉ

> Écrire une méthode **`ajoute(self,v)`** qui ajoute un maillon **`v`** à la fin de la chaine.

#### ✅ RÉSOLUTION

Méthode à écrire **dans** la classe `Maillon` :

````python
class Maillon:
    def __init__(self, v = None):
        self.valeur = v
        self.suivant = None
    def __str__(self):
        affichage = ""
        curseur = self
        while not curseur == None :
            affichage =  affichage + str(curseur.valeur) + " -> "
            curseur = curseur.suivant
        return affichage + "None"
    def ajoute(self, v):
        if self.valeur == None:          # chaîne vide : on remplit le premier maillon
            self.valeur = v
        else:
            curseur = self
            while not curseur.suivant == None:   # on avance jusqu'au dernier maillon
                curseur = curseur.suivant
            curseur.suivant = Maillon(v)         # on accroche le nouveau maillon
````

#### 💡 EXPLICATION PÉDAGOGIQUE

**Idée :** pour ajouter à la fin, il faut **trouver le dernier maillon**, c'est-à-dire celui dont `suivant` vaut `None`. On l'atteint avec le schéma de parcours, mais on s'arrête **sur** le dernier maillon (`while not curseur.suivant == None`), pas après lui.

**Les deux cas :**
1. **Chaîne vide** (`self.valeur == None`) : il n'y a rien à parcourir, on met simplement `v` dans le premier maillon.
2. **Sinon** : on avance jusqu'au dernier maillon, puis on lui accroche `Maillon(v)`.

**Erreurs fréquentes :**
- Écrire `while not curseur == None` : on sort de la chaîne, `curseur` vaut `None`, et `curseur.suivant = ...` provoque une erreur (`AttributeError`).
- Écrire `curseur.suivant = v` au lieu de `Maillon(v)` : on accroche un **nombre**, pas un **maillon**. La chaîne est alors cassée.
- Le mot « maillon `v` » de l'énoncé veut dire « un maillon de valeur `v` ».

> En s'inspirant de la méthode **`str`** de la classe **`Maillon`**, écrire une méthode **`parcours(self)`** qui permet d'afficher à l'aide d'une boucle tous les maillons de la chaine.

#### ✅ RÉSOLUTION

````python
class Maillon:
    def __init__(self, v = None):
        self.valeur = v
        self.suivant = None
    def __str__(self):
        affichage = ""
        curseur = self
        while not curseur == None :
            affichage =  affichage + str(curseur.valeur) + " -> "
            curseur = curseur.suivant
        return affichage + "None"
    def ajoute(self, v):
        if self.valeur == None:          # chaîne vide : on remplit le premier maillon
            self.valeur = v
        else:
            curseur = self
            while not curseur.suivant == None:   # on avance jusqu'au dernier maillon
                curseur = curseur.suivant
            curseur.suivant = Maillon(v)         # on accroche le nouveau maillon
    def parcours(self):
        curseur = self
        while not curseur == None:
            print(curseur.valeur)
            curseur = curseur.suivant
````

````python
chaine = Maillon()
for i in range(1, 6):
    chaine.ajoute(i)
chaine.parcours()
````

#### 💡 EXPLICATION PÉDAGOGIQUE

C'est **exactement** la boucle de `__str__`. La seule différence : au lieu de **construire une chaîne de caractères** (`affichage = affichage + ...`), on **affiche** directement chaque valeur avec `print`.

**Différence `print` / `return` à bien faire comprendre :**
- `__str__` **renvoie** un texte (`return`), et c'est `print(chaine)` qui l'affiche.
- `parcours` **affiche** lui-même et ne renvoie rien. On l'appelle donc avec `chaine.parcours()`, **pas** avec `print(chaine.parcours())`, sinon un `None` en trop s'affiche à la fin.

# EXERCICE 3
## Question 1 et Question 2

#### 📝 ÉNONCÉ

> 1) Écrire une méthode **`supprime_tete(self)`** qui supprime le premier maillon de la chaine.
> 2) Écrire une méthode **`supprime_queue(self)`** qui supprime le dernier maillon de la chaine.

#### ✅ RÉSOLUTION

````python
class Maillon:
    def __init__(self, v = None):
        self.valeur = v
        self.suivant = None
    def __str__(self):
        affichage = ""
        curseur = self
        while not curseur == None :
            affichage =  affichage + str(curseur.valeur) + " -> "
            curseur = curseur.suivant
        return affichage + "None"
    def ajoute(self, v):
        if self.valeur == None:          # chaîne vide : on remplit le premier maillon
            self.valeur = v
        else:
            curseur = self
            while not curseur.suivant == None:   # on avance jusqu'au dernier maillon
                curseur = curseur.suivant
            curseur.suivant = Maillon(v)         # on accroche le nouveau maillon
    def parcours(self):
        curseur = self
        while not curseur == None:
            print(curseur.valeur)
            curseur = curseur.suivant
    def supprime_tete(self):
        if self.suivant == None:         # un seul maillon : la chaîne devient vide
            self.valeur = None
        else:                            # on recopie le 2e maillon dans le 1er
            self.valeur = self.suivant.valeur
            self.suivant = self.suivant.suivant

    def supprime_queue(self):
        if self.suivant == None:         # un seul maillon : la chaîne devient vide
            self.valeur = None
        else:
            curseur = self
            while not curseur.suivant.suivant == None:   # on s'arrête sur l'avant-dernier
                curseur = curseur.suivant
            curseur.suivant = None                       # il devient le dernier
````

````python
chaine = Maillon()
for i in range(1, 6):
    chaine.ajoute(i)
print("Départ           :", chaine)
chaine.supprime_tete()
print("supprime_tete()  :", chaine)
chaine.supprime_queue()
print("supprime_queue() :", chaine)

# Cas limite : un seul maillon
petite = Maillon(7)
petite.supprime_tete()
print("Un seul maillon, après supprime_tete() :", petite, "-> chaîne vide")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

**`supprime_tete` : le piège principal**
On aimerait écrire `self = self.suivant`, mais **ça ne marche pas** : cela change seulement la variable locale `self` dans la méthode. La variable `chaine` du programme désigne toujours le même premier maillon.
👉 L'astuce : on **garde** le premier maillon, mais on y **recopie** le contenu du deuxième (sa valeur et son lien `suivant`). L'ancien deuxième maillon n'est plus relié à rien : il a disparu de la chaîne.

**`supprime_queue` : s'arrêter sur l'avant-dernier**
Pour supprimer le dernier maillon, il faut modifier le `suivant` de l'**avant-dernier** (le mettre à `None`). On avance donc tant que `curseur.suivant.suivant` n'est pas `None` : on regarde **deux maillons plus loin**.

**Cas particulier à ne pas oublier (dans les deux méthodes) :** s'il n'y a **qu'un seul maillon**, on ne peut ni recopier le suivant ni trouver un avant-dernier. On applique alors la convention du cours : on met `valeur` à `None`, et la chaîne est vide.

**Comparaison à faire avec l'élève :**

| Méthode | Nombre d'étapes |
|:---|:---|
| `supprime_tete` | toujours le même (pas de boucle) : **rapide** |
| `supprime_queue` | il faut parcourir toute la chaîne : **plus long si la chaîne est longue** |

# EXERCICE 4 — Implémenter une FILE

#### 📝 ÉNONCÉ

> Avec la classe **`Maillon`**, implémenter une File.
> ```python
> class FileLC:
>     def __init__(self):
>         self.tete = Maillon()
>     def est_vide(self):   ________________________
>     def enfile(self,v):   ________________________
>     def defile(self):     ________________________
>     def __str__(self):
>         return self.tete.__str__()
> ```

#### ✅ RÉSOLUTION

````python
class FileLC:
    def __init__(self):
        self.tete = Maillon()
    def est_vide(self):
        return self.tete.valeur == None
    def enfile(self,v):
        self.tete.ajoute(v)              # on ajoute en queue (à la fin)
    def defile(self):
        v = self.tete.valeur             # le premier arrivé est en tête
        self.tete.supprime_tete()
        return v
    def __str__(self):
        return self.tete.__str__()
````

````python
F = FileLC()
print("Vide ?", F.est_vide())
for x in [21, 22, 23]:
    F.enfile(x)
print("F =", F)
print("defile() renvoie", F.defile())
print("F =", F)
print("Vide ?", F.est_vide())
````

#### 💡 EXPLICATION PÉDAGOGIQUE

**Rappel (leçon 02E) :** une file est **FIFO**, « premier arrivé, premier sorti ». On **ajoute à la fin** et on **retire au début**.

**On réutilise ce qu'on a déjà écrit :**

| Primitive de la file | Méthode de `Maillon` utilisée |
|:---|:---|
| `est_vide` | convention du cours : la valeur du premier maillon vaut `None` |
| `enfile(v)` | `ajoute(v)` (exercice 1) : ajout **en fin** de chaîne |
| `defile()` | on lit la valeur du premier maillon, puis `supprime_tete()` (exercice 3) |

**Erreur fréquente dans `defile` :** supprimer le maillon **avant** d'avoir lu sa valeur. Il faut **d'abord** mémoriser `v`, **puis** supprimer, **puis** renvoyer `v`.

**Lien avec le TDA (leçon 02A) :** les primitives sont les mêmes que pour la classe `File` de la leçon 02E (écrite avec une liste Python). Seule l'**implémentation** change. Celui qui utilise la file ne voit pas la différence.

# EXERCICE 5 — Implémenter une PILE

#### 📝 ÉNONCÉ

> Avec la classe **`Maillon`**, implémenter une File.
> ```python
> class PileLC:
>     def __init__(self):
>         self.tete = Maillon()
>     def est_vide(self):   ________________________
>     def empile(self,v):   ________________________
>     def depile(self):     ________________________
>     def __str__(self):
>         return self.tete.__str__()
> ```

> ✏️ **Coquille dans la feuille :** le titre et la classe `PileLC` montrent qu'il faut implémenter une **Pile** (et non une File).

#### ✅ RÉSOLUTION

````python
class PileLC:
    def __init__(self):
        self.tete = Maillon()
    def est_vide(self):
        return self.tete.valeur == None
    def empile(self,v):
        nouveau = Maillon(v)             # le nouveau maillon devient le sommet
        if not self.est_vide():
            nouveau.suivant = self.tete  # il pointe vers l'ancien sommet
        self.tete = nouveau
    def depile(self):
        v = self.tete.valeur             # le sommet est en tête
        self.tete.supprime_tete()
        return v
    def __str__(self):
        return self.tete.__str__()
````

````python
P = PileLC()
print("Vide ?", P.est_vide())
for x in [1, 2, 3]:
    P.empile(x)
print("P =", P, "  (le sommet est à gauche)")
print("depile() renvoie", P.depile())
print("P =", P)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

**Rappel (leçon 02D) :** une pile est **LIFO**, « dernier arrivé, premier sorti ». On ajoute et on retire **du même côté** : le **sommet**.

**Quel côté choisir pour le sommet ?** C'est la vraie question de l'exercice.
- **Sommet = tête de chaîne (solution choisie)** : `empile` et `depile` ne parcourent pas la chaîne, ils sont **rapides** quelle que soit la taille de la pile.
- **Sommet = fin de chaîne** : on pourrait écrire `empile` avec `ajoute` et `depile` avec `supprime_queue`. C'est **correct aussi**, mais chaque opération parcourt toute la chaîne (voir le tableau de l'exercice 3).

**Pourquoi le `if not self.est_vide()` dans `empile` ?** Si la pile est vide, la tête est le maillon « vide » de valeur `None`. Sans ce test, on garderait ce maillon vide au fond de la pile, et l'affichage serait `1 -> None -> None`.

**Ici on peut écrire `self.tete = nouveau`** (contrairement à `self = ...` dans `supprime_tete`) : `tete` est un **attribut** de l'objet pile, on a donc le droit de le changer.

**Comparaison File / Pile avec maillons :**

| | File (`FileLC`) | Pile (`PileLC`) |
|:---|:---|:---|
| Ajout | en **fin** de chaîne | en **tête** de chaîne |
| Retrait | en **tête** de chaîne | en **tête** de chaîne |

# EXERCICE 9 — Partie C : le collier de bonbons (liste doublement chaînée)

> ℹ️ Cet exercice est la **suite de l'exercice 13 de la leçon 02E** (parties A et B), qui ne figure pas sur les photos reçues. La partie C se résout sans elles. Pour la question 6, la règle du jeu (« décalage de 3 bonbons ») est **déduite du code à compléter**.

#### 📝 ÉNONCÉ

> Bob souhaite utiliser la structure de données « liste doublement chaînée ».
> Une telle liste est composée de maillons contenants chacun trois informations : une valeur, un prédécesseur, un successeur qui sont tous deux des maillons.
> Le successeur du dernier bonbon est le premier et le prédécesseur du premier, le dernier.
> ```python
> class Bonbon:
>     def __init__(self, valeur):
>         self.pred = None # prédécesseur de ce bonbon
>         self.valeur = valeur
>         self.succ = None # successeur de ce bonbon
> ```

````python
class Bonbon:
    def __init__(self, valeur):
        self.pred = None # prédécesseur de ce bonbon
        self.valeur = valeur
        self.succ = None # successeur de ce bonbon
````

#### 💡 EXPLICATION PÉDAGOGIQUE

**Différence avec la classe `Maillon` :**
- Un `Maillon` connaît seulement **le suivant** : on ne peut avancer que dans un sens.
- Un `Bonbon` connaît **le suivant (`succ`) et le précédent (`pred`)** : on peut aller dans les **deux sens**. C'est une liste **doublement** chaînée.
- De plus, le collier est **refermé** : il n'y a plus de `None` au bout. Après le dernier bonbon, on revient au premier.

## Question 1

#### 📝 ÉNONCÉ

> Comment s'appelle les variables **`pred`**, **`valeur`** et **`succ`** dans le vocabulaire POO.

#### ✅ RÉSOLUTION

Ce sont les **attributs** de la classe `Bonbon`.

#### 💡 EXPLICATION PÉDAGOGIQUE

Vocabulaire de la leçon 01 : les variables `self.xxx` définies dans `__init__` sont les **attributs**. Les fonctions définies dans la classe sont les **méthodes**. `Bonbon(0)` crée une **instance** (un objet) de la classe.

## Question 2

#### 📝 ÉNONCÉ

> ```python
> zero, un, deux = Bonbon(0), Bonbon(1), Bonbon(2)
> zero.succ = un
> un.pred = zero
> un.succ = deux
> deux.pred = un
> deux.succ = zero
> zero.pred = deux
> a , b = zero.succ.valeur, un.succ.succ.pred.valeur
> ```
> Déterminer les valeurs des variables a et b après l'exécution de ces instructions.

#### ✅ RÉSOLUTION

- `a = zero.succ.valeur` : le successeur de `zero` est `un`, donc **`a = 1`**.
- `b = un.succ.succ.pred.valeur` : `un.succ` est `deux`, `deux.succ` est `zero`, `zero.pred` est `deux`, donc **`b = 2`**.

````python
zero, un, deux = Bonbon(0), Bonbon(1), Bonbon(2)
zero.succ = un
un.pred = zero
un.succ = deux
deux.pred = un
deux.succ = zero
zero.pred = deux
a , b = zero.succ.valeur, un.succ.succ.pred.valeur
print("a =", a, "  b =", b)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

**Méthode :** on lit l'expression **de gauche à droite**, un point à la fois, en notant à chaque étape sur quel bonbon on se trouve :

| Étape | On est sur |
|:---|:---|
| `un` | 1 |
| `.succ` | 2 |
| `.succ` | 0 (le collier est refermé) |
| `.pred` | 2 |
| `.valeur` | **2** |

Conseil : faire **dessiner** le collier (trois cases en cercle, flèches `succ` dans un sens et `pred` dans l'autre) avant de suivre les flèches.

## Question 3

#### 📝 ÉNONCÉ

> Compléter le code de la fonction **`creer_collier`** prend en paramètre un entier n strictement positif représentant la taille d'un collier et renvoie un objet de type Bonbon représentant le premier bonbon (de valeur 0) du collier. On prendra soin de faire se succéder et précéder les différents bonbons ainsi que de « refermer » le collier en liant le dernier bonbon au premier.

#### ✅ RÉSOLUTION

````python
def creer_collier(n):
    premier = Bonbon(0)
    actuel = premier
    for i in range(1, n):
        nouveau = Bonbon(i)
        actuel.succ = nouveau
        nouveau.pred = actuel
        actuel = nouveau
    actuel.succ = premier
    premier.pred = actuel
    return premier
````

*Vérification (non demandée) : on fait le tour du collier dans les deux sens.*

````python
def tour(premier, sens):
    """Renvoie la liste des valeurs en faisant un tour complet (sens = "succ" ou "pred")."""
    valeurs = [premier.valeur]
    b = getattr(premier, sens)
    while b != premier:
        valeurs.append(b.valeur)
        b = getattr(b, sens)
    return valeurs

c = creer_collier(5)
print("Tour par succ :", tour(c, "succ"))
print("Tour par pred :", tour(c, "pred"))

c1 = creer_collier(1)
print("n = 1 : succ et pred du bonbon pointent sur lui-même ?", c1.succ == c1 and c1.pred == c1)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

**Ce que fait chaque ligne à compléter :**

| Ligne | Rôle |
|:---|:---|
| `nouveau = Bonbon(i)` | le bonbon numéro `i` (valeurs 1, 2, …, n-1, car 0 est déjà créé) |
| `actuel.succ = nouveau` | lien **vers l'avant** : l'actuel est suivi par le nouveau |
| `nouveau.pred = actuel` | lien **vers l'arrière** : le nouveau est précédé par l'actuel |
| `actuel.succ = premier` | **après la boucle** : le dernier bonbon est suivi par le premier |
| `premier.pred = actuel` | **après la boucle** : le premier est précédé par le dernier |

**Points clés :**
- Dans une liste **doublement** chaînée, chaque lien se fait **dans les deux sens** : deux affectations à chaque fois.
- Les deux dernières lignes sont **en dehors** de la boucle (une seule fois, à la fin). À la sortie de la boucle, `actuel` désigne le **dernier** bonbon.
- **Cas `n = 1` :** la boucle ne tourne pas, et `actuel` est encore `premier`. Les deux dernières lignes font alors `premier.succ = premier` et `premier.pred = premier`. C'est exactement le schéma du bonbon 3 tout seul donné dans l'énoncé.

        ┌──────────────── succ ────────────────┐
        ▼                                      │
 bonbon → [ 1 ] ──succ──▶ [ 2 ] ──succ──▶ [ 3 ]
          [ 1 ] ◀──pred── [ 2 ] ◀──pred── [ 3 ]
          │                                ▲
          └──────────────── pred ──────────┘
```

C'est-à-dire : `1.succ = 2`, `2.succ = 3`, `3.succ = 1` et `1.pred = 3`, `3.pred = 2`, `2.pred = 1`.

````python
premier = creer_collier(4)
premier.pred.succ = premier.succ
premier.succ.pred = premier.pred
bonbon = premier.succ
print("Tour par succ à partir de bonbon :", tour(bonbon, "succ"))
print("Tour par pred à partir de bonbon :", tour(bonbon, "pred"))
````

#### 💡 EXPLICATION PÉDAGOGIQUE

**Raisonnement pas à pas** (au départ : 0 ⇄ 1 ⇄ 2 ⇄ 3 ⇄ 0) :
1. `premier.pred` est le bonbon **3** et `premier.succ` est le bonbon **1**.
2. `premier.pred.succ = premier.succ` donne `3.succ = 1` : le 3 « saute » le 0 vers l'avant.
3. `premier.succ.pred = premier.pred` donne `1.pred = 3` : le 1 « saute » le 0 vers l'arrière.
4. `bonbon = premier.succ` : `bonbon` est le **1**.

👉 Ces deux lignes sont **la technique pour retirer un maillon** d'une liste doublement chaînée : on relie directement son voisin de gauche et son voisin de droite. Elles reviennent à la question 6.

Remarque : le bonbon 0 pointe encore vers 3 et vers 1, mais **plus personne ne pointe vers lui**. En partant de `bonbon`, on ne le rencontre plus jamais : il ne fait plus partie du collier.

````python
bonbon = creer_collier(1)
print(bonbon.valeur == bonbon.succ.valeur)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Quand il ne reste qu'**un** bonbon, son successeur est **lui-même** (comme le bonbon 3 tout seul de l'énoncé). Donc `bonbon.succ.valeur` est sa propre valeur, et l'égalité est vraie.
- **Les trois autres expressions sont fausses ou n'ont pas de sens :** `valeur`, `pred` et `succ` employés **seuls** ne sont pas des variables du programme. Ce sont des **attributs**, qui s'écrivent toujours après un objet et un point (`bonbon.succ`). Python afficherait une erreur `NameError`.

## Question 6

#### 📝 ÉNONCÉ

> Compléter le code de la fonction **`dernier_chaine`** qui prend en paramètre le nombre de bonbons n et renvoie la valeur du dernier bonbon restant.
> ```python
> def dernier_chaine(n):
>     bonbon = creer_collier(n)
>     while __________ != __________:
>         bonbon.pred.succ = __________
>         __________ = bonbon.pred
>         bonbon = __________ # décalage de 3 bonbons
>     return bonbon.valeur
> ```

#### ✅ RÉSOLUTION

````python
def dernier_chaine(n):
    bonbon = creer_collier(n)
    while bonbon != bonbon.succ:
        bonbon.pred.succ = bonbon.succ
        bonbon.succ.pred = bonbon.pred
        bonbon = bonbon.succ.succ.succ # décalage de 3 bonbons
    return bonbon.valeur

for n in range(1, 11):
    print("n =", n, "-> dernier bonbon :", dernier_chaine(n))
````

#### 💡 EXPLICATION PÉDAGOGIQUE

**On assemble les questions précédentes :**

| Ligne | D'où elle vient |
|:---|:---|
| `while bonbon != bonbon.succ:` | question 5 : il reste **un seul** bonbon quand il est son propre successeur. On continue **tant que ce n'est pas le cas**. |
| `bonbon.pred.succ = bonbon.succ` | question 4 : le voisin de gauche saute le bonbon… |
| `bonbon.succ.pred = bonbon.pred` | … et le voisin de droite aussi : le bonbon est **retiré** du collier. |
| `bonbon = bonbon.succ.succ.succ` | on avance de **3** bonbons pour trouver le prochain à retirer. |

**Pourquoi peut-on encore écrire `bonbon.succ` après avoir retiré `bonbon` ?** Le bonbon retiré n'est plus dans le collier, mais **ses propres attributs n'ont pas changé** : il « se souvient » de son ancien voisin de droite. On part de là pour avancer.

**Variante acceptée pour le `while` :** `bonbon.succ != bonbon` (même test, écrit dans l'autre sens).

**Erreur fréquente :** écrire `while bonbon.valeur != bonbon.succ.valeur`. Cela marche ici, car toutes les valeurs sont différentes, mais il vaut mieux comparer les **objets** eux-mêmes : on veut savoir si c'est **le même bonbon**.

*Vérification (non demandée) : on refait le même jeu avec la **file** `FileLC` de l'exercice 4, comme dans l'exercice 13 de la leçon 02E sur les Files.*

On retire le bonbon en tête de file, puis on fait passer 2 bonbons de la tête à la queue (`enfile(defile())`). Le prochain bonbon retiré est ainsi le 3e après le bonbon retiré.

````python
def dernier_file(n):
    F = FileLC()
    for i in range(n):
        F.enfile(i)
    reste = n
    while reste > 1:
        F.defile()              # on retire un bonbon
        reste = reste - 1
        for k in range(2):      # on fait tourner le collier de 2 bonbons
            F.enfile(F.defile())
    return F.defile()

print("Mêmes résultats avec la file :",
      all(dernier_file(n) == dernier_chaine(n) for n in range(1, 30)))
````

#### 💡 EXPLICATION PÉDAGOGIQUE

Les deux méthodes donnent le **même résultat**. Le collier (liste doublement chaînée) et la file font le **même travail**. C'est le lien entre la partie C et l'exercice 13 de la leçon 02E.
