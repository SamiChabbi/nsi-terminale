# Leçon 02E — Exercices résolus
*Structures linéaires — Les Files*

Notions utilisées : principe **FIFO** (premier arrivé, premier sorti), les primitives `creer_file`, `enfile`, `defile`, `est_vide`, et le TDC `File` de la partie 3.

Pour vérifier les réponses, on reprend la classe `File` de la leçon et on écrit les primitives avec elle. L'affichage `print(F)` montre la file de la **tête (à gauche)** à la **queue (à droite)**.

````python
# Le TDC File de la leçon (partie 3)
class File:
    def __init__(self): self.valeurs = []
    def enfile(self, valeur): self.valeurs.append(valeur)
    def defile(self):
        if self.valeurs:  return self.valeurs.pop(0)
    def est_vide(self):  return self.valeurs == []
    def longueur(self):  return len(self.valeurs)
    def __str__(self):
        ch = ""
        for x in self.valeurs:
            ch = ch + " <- " +  str(x)
        return  ch

# Les primitives de la leçon, écrites avec cette classe
def creer_file():
    return File()

def enfile(F, e):
    F.enfile(e)

def defile(F):
    return F.defile()

def est_vide(F):
    return F.est_vide()
````

## Exercice 1

> ```python
> F = creer_file()
> enfile(F,21)
> enfile(F,22)
> enfile(F,23)
> N = defile(F)
> enfile(F,24)
> enfile(F,25)
> N = defile(F)
> ```
> 1) Que contient la variable **`N`** ?  2) Que contient la file **`F`** ?

### Déroulement à la main (tête à gauche, queue à droite)

| Instruction | File F | N |
|:---|:---|:---|
| `F = creer_file()` | `[]` | |
| `enfile(F,21)` | `[21]` | |
| `enfile(F,22)` | `[21, 22]` | |
| `enfile(F,23)` | `[21, 22, 23]` | |
| `N = defile(F)` | `[22, 23]` | 21 (premier entré) |
| `enfile(F,24)` | `[22, 23, 24]` | 21 |
| `enfile(F,25)` | `[22, 23, 24, 25]` | 21 |
| `N = defile(F)` | `[23, 24, 25]` | **22** |

**Réponses :**
1) `N` contient **22**. Le 2ᵉ `defile` écrase la valeur 21.
2) `F` contient **23, 24, 25**, avec **23 en tête** (prochain à sortir).

````python
# Vérification
F = creer_file()
enfile(F,21)
enfile(F,22)
enfile(F,23)
N = defile(F)
enfile(F,24)
enfile(F,25)
N = defile(F)
print("N =", N)
print("F =", F)
````

## Exercice 2

> Écrire une suite d'instructions qui permet de «retourner» une file F c'est-à-dire de l'écrire dans l'autre sens (le premier devient le dernier). On pourra utiliser une Pile.

### Idée
- Une **file** rend les éléments **dans l'ordre d'arrivée** (FIFO).
- Une **pile** les rend **dans l'ordre inverse** (LIFO, leçon 02D).

Donc on **défile tout dans une pile**, puis on **dépile tout dans la file** : l'ordre est inversé.

On utilise les primitives de pile de la leçon 02D. On les renomme `est_vide_pile` pour ne pas les confondre avec `est_vide` de la file.

````python
# Primitives de la Pile (leçon 02D), avec une liste
def creer_pile():
    return []
def empile(P, e):
    P.append(e)
def depile(P):
    return P.pop()
def est_vide_pile(P):
    return len(P) == 0

# Une file de départ
F = creer_file()
for x in [1, 2, 3, 4, 5]:
    enfile(F, x)
print("Avant   :", F)

# Étape 1 : on vide la file dans la pile (1 est empilé en premier, 5 finit au sommet)
P = creer_pile()
while not est_vide(F):
    empile(P, defile(F))

# Étape 2 : on vide la pile dans la file (5 sort en premier et devient la tête)
while not est_vide_pile(P):
    enfile(F, depile(P))

print("Après   :", F)
````

Résultat attendu :
```
Avant   :  <- 1 <- 2 <- 3 <- 4 <- 5
Après   :  <- 5 <- 4 <- 3 <- 2 <- 1
```
Le premier élément (1) est bien devenu le dernier. On n'a utilisé **que les primitives**, sans toucher à l'intérieur de la liste : c'est tout l'intérêt de raisonner sur des TDA.
