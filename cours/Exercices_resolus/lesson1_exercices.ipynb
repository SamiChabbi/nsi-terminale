Notions utilisées : définir une classe avec `class` et `__init__`, créer des instances, lire et modifier un attribut avec `inst.attribut`.

## Exercice 1

> 1) Créer une class **`Personne`** avec comme attributs **`nom`**, **`prenom`**, **`age`**.
>    Le constructeur prendra en paramètres deux chaines de caractères et un entier.
> 2) Créer trois instances de la class **`Personne`** et afficher leurs attributs.
> 3) Ajouter un an à chacune des instances.
>    Afficher les attributs pour vérifier.

### Question 1

````python
# Définition de la classe Personne (même modèle que la classe Eleve du cours)
class Personne :
    # Le constructeur reçoit self (obligatoire) puis 2 chaînes et 1 entier
    def __init__(self, n, p, a):
        self.nom = n        # attribut nom (chaîne de caractères)
        self.prenom = p     # attribut prenom (chaîne de caractères)
        self.age = a        # attribut age (entier)
````

`__init__` est le **constructeur** : il s'exécute à chaque création d'une instance. `self` représente l'instance en cours de création, et `self.nom = n` enregistre la valeur `n` dans l'attribut `nom` de cette instance.

### Question 2

````python
# Création de trois instances : on appelle la classe comme une fonction
p1 = Personne('Dupont', 'Alice', 16)
p2 = Personne('Martin', 'Hugo', 17)
p3 = Personne('Durand', 'Inès', 15)

# Affichage des attributs avec la syntaxe inst.attribut
print(p1.nom, p1.prenom, p1.age)
print(p2.nom, p2.prenom, p2.age)
print(p3.nom, p3.prenom, p3.age)
````

Résultat attendu :
```
Dupont Alice 16
Martin Hugo 17
Durand Inès 15
```
Chaque instance a **ses propres** valeurs d'attributs, même si elles viennent toutes de la même classe.

### Question 3

````python
# Les attributs sont mutables : on peut les modifier (comme e1.age = e1.age + 1 dans le cours)
p1.age = p1.age + 1
p2.age = p2.age + 1
p3.age = p3.age + 1

# Vérification
print(p1.nom, p1.prenom, p1.age)
print(p2.nom, p2.prenom, p2.age)
print(p3.nom, p3.prenom, p3.age)
````

Résultat attendu :
```
Dupont Alice 17
Martin Hugo 18
Durand Inès 16
```
L'âge de chaque personne a augmenté de 1 : modifier `p1.age` ne change que l'instance `p1`, pas les autres.
