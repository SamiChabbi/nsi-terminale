```
       5

```

Arbre des appels (`s(i)` = `somme_liste(nombres, i)`), dans l'ordre où Python les fait :

| n° | appel | ce qu'il fait | renvoie |
|:---|:---|:---|:---|
| 1 | `s(0)` | nœud 5, appelle `s(1)` puis `s(2)` | 5 + 3 + 12 = **20** |
| 2 | `s(1)` | nœud 3, appelle `s(-1)` puis `s(-1)` | 3 + 0 + 0 = 3 |
| 3 | `s(-1)` | vide | 0 |
| 4 | `s(-1)` | vide | 0 |
| 5 | `s(2)` | nœud 10, appelle `s(3)` puis `s(-1)` | 10 + 2 + 0 = 12 |
| 6 | `s(3)` | nœud 2, appelle `s(-1)` puis `s(-1)` | 2 + 0 + 0 = 2 |
| 7 | `s(-1)` | vide | 0 |
| 8 | `s(-1)` | vide | 0 |
| 9 | `s(-1)` | vide (fils droit de 10) | 0 |

Valeurs de `i` : 0, 1, -1, -1, 2, 3, -1, -1, -1. **9 appels en tout.** On vérifie en comptant les appels :

````python
nb_appels = 0

def somme_liste(arbre, i=0):
    global nb_appels
    nb_appels = nb_appels + 1
    print("appel avec i =", i)
    if i == -1:
        return 0
    noeud = arbre[i]
    return noeud['v'] + somme_liste(arbre, noeud['g']) + somme_liste(arbre, noeud['d'])

nombres = [{'v':5, 'g':1, 'd':2}, {'v':3, 'g':-1, 'd':-1},
           {'v':10, 'g':3, 'd':-1}, {'v':2, 'g':-1, 'd':-1}]
print("somme =", somme_liste(nombres))
print("nombre d'appels =", nb_appels)
````

### 💡 EXPLICATION PÉDAGOGIQUE

- Le compteur `global` et les `print` servent **seulement** à vérifier le déroulé ; ils ne font pas partie de la fonction à apprendre.
- Python termine **complètement** l'appel de gauche (`s(1)` et tout ce qu'il déclenche) avant de commencer celui de droite (`s(2)`).
- Il y a 4 nœuds, donc 4 appels « utiles », et 5 appels sur le vide (`-1`). Règle : `n` nœuds → `n + 1` sous-arbres vides → `2n + 1` appels.
- Une feuille fait quand même **deux** appels (sur `-1`) : c'est normal, c'est le cas de base qui s'en occupe.

## Exercice 03 — Taille

### 📝 ÉNONCÉ

Écris une fonction **récursive** `taille_liste(arbre, i=0)` qui renvoie la **taille** (nombre de nœuds) du sous-arbre dont la racine est à l'indice `i`.

### ✅ RÉSOLUTION

On copie le modèle de `somme_liste` en se posant les deux questions :
1. taille d'un arbre vide : **0** ;
2. combinaison : **1** (le nœud lui-même) + taille à gauche + taille à droite.

````python
def taille_liste(arbre, i=0):
    if i == -1:                       # sous-arbre vide : aucun nœud
        return 0
    noeud = arbre[i]
    return 1 + taille_liste(arbre, noeud['g']) + taille_liste(arbre, noeud['d'])

# Tests
assert taille_liste(arbre1) == 10
assert taille_liste(arbre2) == 9
assert taille_liste(arbre_ex01) == 10
assert taille_liste(arbre1, 1) == 4       # sous-arbre de B : B, C, D, E
assert taille_liste(arbre1, -1) == 0      # arbre vide
print("✔ Tous les tests passent")
````

### 💡 EXPLICATION PÉDAGOGIQUE

- C'est **exactement** `somme_liste` où l'on remplace `noeud['v']` par `1` : chaque nœud compte pour 1, quelle que soit sa valeur.
- Le cas de base est l'arbre **vide**. Pas besoin de cas spécial pour les feuilles : une feuille renvoie `1 + 0 + 0 = 1`.
- `len(arbre)` donnerait aussi 10 pour l'arbre entier, mais pas la taille d'un **sous-arbre** (`taille_liste(arbre1, 1)` vaut 4) : c'est pour ça qu'on suit la forme de l'arbre.

## Exercice 04 — Hauteur

### 📝 ÉNONCÉ

Écris une fonction **récursive** `hauteur_liste(arbre, i=0)` qui renvoie la **hauteur** du sous-arbre dont la racine est à l'indice `i` (arbre vide → 0, une feuille → 1).

### ✅ RÉSOLUTION

1. hauteur d'un arbre vide : **0** ;
2. combinaison : le nœud ajoute **un niveau** au-dessus du plus haut de ses deux sous-arbres : `1 + max(hauteur gauche, hauteur droite)`.

````python
def hauteur_liste(arbre, i=0):
    if i == -1:                       # sous-arbre vide : 0 niveau
        return 0
    noeud = arbre[i]
    hg = hauteur_liste(arbre, noeud['g'])
    hd = hauteur_liste(arbre, noeud['d'])
    if hg > hd:
        return 1 + hg
    else:
        return 1 + hd

# Tests
assert hauteur_liste(arbre1) == 4
assert hauteur_liste(arbre2) == 4
assert hauteur_liste(arbre_ex01) == 5
assert hauteur_liste(arbre1, 4) == 1      # D est une feuille
assert hauteur_liste(arbre1, -1) == 0     # arbre vide
print("✔ Tous les tests passent")
````

### 💡 EXPLICATION PÉDAGOGIQUE

- On ne **somme** pas les deux hauteurs (ce serait la taille) : on garde **la plus grande**, car la hauteur est le plus **long** chemin, et ce chemin passe d'un seul côté.
- On peut écrire `return 1 + max(hg, hd)` : c'est la même chose que le `if`.
- On range les deux résultats dans `hg` et `hd` pour ne faire **chaque appel qu'une fois**. Écrire `if hauteur_liste(...) > hauteur_liste(...)` puis rappeler la fonction dans le `return` referait les mêmes calculs plusieurs fois.
- Convention : feuille → 1. Avec la convention « feuille → 0 », le cas de base renverrait `-1` pour l'arbre vide. Toujours lire l'énoncé.

## Exercice 05 — Parcours en largeur

### 📝 ÉNONCÉ

Écris une fonction `largeur_liste(arbre)` qui **renvoie la liste** des valeurs des nœuds dans l'ordre d'un parcours **en largeur**.

Cette fonction n'est **pas récursive** : elle utilise une **file** (classe `File` fournie).

Avant de coder : quel résultat attends-tu pour l'arbre 1 de l'exercice 02 ?

### ✅ RÉSOLUTION

**Résultat attendu pour l'arbre 1** (niveau par niveau, de gauche à droite) : `['A', 'B', 'F', 'C', 'D', 'G', 'H', 'E', 'I', 'J']`.

**Algorithme** : on range dans la file les **indices** des nœuds qui restent à visiter.
1. On enfile 0 (la racine).
2. Tant que la file n'est pas vide : on défile un indice, on ajoute la valeur de ce nœud au résultat, puis on enfile l'indice du fils gauche, puis celui du fils droit, s'ils ne valent pas `-1`.

⚠️ Liberté prise : la feuille dit « classe `File` fournie » mais on ne voit pas son code sur les photos. On reprend la classe `File` de la leçon 02E (`enfile`, `defile`, `est_vide`). Si la classe fournie a d'autres noms de méthodes (par exemple `enfiler`, `defiler`), il suffit de remplacer les noms : le raisonnement ne change pas.

````python
def largeur_liste(arbre):
    resultat = []
    F = File()
    F.enfile(0)                           # on commence par la racine
    while not F.est_vide():
        i = F.defile()                    # le plus ancien nœud en attente
        noeud = arbre[i]
        resultat.append(noeud['v'])
        if noeud['g'] != -1:              # on met le fils gauche en attente
            F.enfile(noeud['g'])
        if noeud['d'] != -1:              # puis le fils droit
            F.enfile(noeud['d'])
    return resultat

print(largeur_liste(arbre1))

# Tests
assert largeur_liste(arbre1) == ['A', 'B', 'F', 'C', 'D', 'G', 'H', 'E', 'I', 'J']
assert largeur_liste(arbre2) == [1, 2, 3, 4, 5, 6, 7, 8, 9]
assert largeur_liste(arbre_ex01) == ['X', 'W', 'C', 'H', 'Z', 'F', 'B', 'U', 'P', 'T']
print("✔ Tous les tests passent")
````

### 💡 EXPLICATION PÉDAGOGIQUE

Tableau de la file pour l'arbre 1 (tête à gauche) :

| Défilé | Résultat | On enfile | File après |
|:---|:---|:---|:---|
| | | A | A |
| A | A | B, F | B F |
| B | A B | C, D | F C D |
| F | A B F | G, H | C D G H |
| C | A B F C | E | D G H E |
| D | A B F C D | — | G H E |
| G | … G | I | H E I |
| H | … H | J | E I J |
| E | … E | — | I J |
| I | … I | — | J |
| J | … J | — | (vide) : fin |

- La file contient des **indices** (numéros de cases), pas des valeurs : c'est avec l'indice qu'on retrouve `'g'` et `'d'`.
- On enfile **gauche avant droite** : c'est ce qui donne l'ordre « de gauche à droite » dans chaque niveau.
- On teste `!= -1` **avant** d'enfiler, sinon `arbre[-1]` donnerait la **dernière case** de la liste (Python accepte les indices négatifs) et le parcours serait faux sans message d'erreur.
- Pourquoi pas de récursivité : une fonction récursive finit tout le sous-arbre gauche avant de toucher au droit ; la file, elle, mélange les deux côtés niveau par niveau.

- postfixe (gauche, droite, nœud) : E C D B I G J H F A

**Code** : même modèle que `feuilles_liste` ; vide → `[]`, puis on concatène. Seule la **place** de `[noeud['v']]` change.

````python
def prefixe_liste(arbre, i=0):
    if i == -1:
        return []
    noeud = arbre[i]
    return [noeud['v']] + prefixe_liste(arbre, noeud['g']) + prefixe_liste(arbre, noeud['d'])

def infixe_liste(arbre, i=0):
    if i == -1:
        return []
    noeud = arbre[i]
    return infixe_liste(arbre, noeud['g']) + [noeud['v']] + infixe_liste(arbre, noeud['d'])

def postfixe_liste(arbre, i=0):
    if i == -1:
        return []
    noeud = arbre[i]
    return postfixe_liste(arbre, noeud['g']) + postfixe_liste(arbre, noeud['d']) + [noeud['v']]

print(prefixe_liste(arbre1))
print(infixe_liste(arbre1))
print(postfixe_liste(arbre1))

# Tests
assert prefixe_liste(arbre1) == ['A', 'B', 'C', 'E', 'D', 'F', 'G', 'I', 'H', 'J']
assert infixe_liste(arbre1) == ['C', 'E', 'B', 'D', 'A', 'I', 'G', 'F', 'H', 'J']
assert postfixe_liste(arbre1) == ['E', 'C', 'D', 'B', 'I', 'G', 'J', 'H', 'F', 'A']
assert prefixe_liste(arbre2) == [1, 2, 4, 5, 7, 8, 3, 6, 9]
assert infixe_liste(arbre2) == [4, 2, 7, 5, 8, 1, 3, 9, 6]
assert postfixe_liste(arbre2) == [4, 7, 8, 5, 2, 9, 6, 3, 1]
assert prefixe_liste(arbre1, -1) == []
print("✔ Tous les tests passent")
````

### 💡 EXPLICATION PÉDAGOGIQUE

- **Pré**fixe : la valeur **avant** les deux appels. **In**fixe : **entre** les deux. **Post**fixe : **après** les deux. Le reste du code est identique.
- On écrit `[noeud['v']]` avec des crochets : on ne peut concaténer qu'une **liste** avec une liste. `noeud['v'] + liste` provoquerait une erreur.
- Contrôle rapide sur le cahier : la racine A est **première** en préfixe, **dernière** en postfixe, et en infixe elle sépare {C, E, B, D} (sous-arbre gauche) de {I, G, F, H, J} (sous-arbre droit).
- Piège de l'infixe : C n'a pas de fils gauche, donc C s'écrit **avant** son fils droit E (C E, pas E C). Pour I et G, c'est l'inverse : I est à gauche de G, donc I G.
