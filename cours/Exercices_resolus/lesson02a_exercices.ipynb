*Structures linéaires — TDA -- TDC*

Notions utilisées : un **TDA**, ce sont des **données** plus des **opérations** (primitives), avec les 4 primitives élémentaires **Create / Read / Update / Delete**.

## Exercice 1

> Imaginer un <u>TDA</u> **Joueurs** pour un jeu vidéo.
> Décrire les attributs et les primitives

### Réponse

Comme dans l'exemple de l'université (un enregistrement par étudiant), le TDA **Joueurs** gère un ensemble d'enregistrements, un par joueur.

#### Les données (attributs d'un joueur)

| Attribut | Nature | Exemple |
|:---|:---|:---|
| pseudo | chaîne de caractères | `"Zelda"` |
| niveau | entier | `12` |
| points de vie | entier | `100` |
| score | entier | `4500` |
| date d'inscription | date | `15/09/2026` |

#### Les primitives (les 4 opérations élémentaires)

| Primitive | Type | Rôle |
|:---|:---|:---|
| `creer_joueur(pseudo)` | **Create** | ajoute un nouveau joueur (niveau 1, 100 points de vie, score 0) |
| `lire_joueur(pseudo)` | **Read** | renvoie les informations du joueur |
| `modifier_joueur(pseudo, attribut, valeur)` | **Update** | change une information (ex. augmenter le score, passer au niveau suivant) |
| `supprimer_joueur(pseudo)` | **Delete** | retire le joueur (compte supprimé) |

#### Pourquoi c'est un TDA ?
On a décrit **quelles données** et **quelles opérations**, sans dire **comment** les programmer. C'est ce qui rend la description **indépendante du langage**, comme le demande le cours. La traduire en Python, ce serait écrire un **TDC** (partie c).

### Pour aller plus loin : un TDC possible

Le cours dit qu'un même TDA peut avoir plusieurs implémentations (TDC). En voici une en Python, qui utilise une classe (leçon 01) pour un joueur et un dictionnaire (vu en première) pour l'ensemble des joueurs.

```python
# Un joueur = une instance de classe (leçon 01)
class Joueur :
    def __init__(self, pseudo):
        self.pseudo = pseudo
        self.niveau = 1
        self.vie = 100
        self.score = 0

joueurs = {}                       # l'ensemble des joueurs, rangés par pseudo

# Create
joueurs["Zelda"] = Joueur("Zelda")
# Read
print(joueurs["Zelda"].pseudo, joueurs["Zelda"].niveau, joueurs["Zelda"].score)
# Update
joueurs["Zelda"].score = joueurs["Zelda"].score + 500
print(joueurs["Zelda"].score)
# Delete
joueurs.pop("Zelda")
print(len(joueurs))
```
