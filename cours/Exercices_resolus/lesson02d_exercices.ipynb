*Les Piles*

Notions utilisées : principe **LIFO** (dernier arrivé, premier sorti) et les primitives `creer_pile`, `empile`, `depile`, `est_vide`.

Les exercices se font **à la main**. Pour vérifier les réponses en Python, on programme d'abord les 4 primitives avec une liste (méthodes `append` et `pop()` de la leçon 02B).

````python
# Implémentation des primitives de la leçon avec une liste Python (leçon 02B) :
# le sommet de la pile est la fin de la liste.
def creer_pile():
    return []                 # une pile vide

def empile(P, e):
    P.append(e)               # on ajoute e au sommet (fin de liste)

def depile(P):
    return P.pop()            # on retire et renvoie le sommet (dernier élément)

def est_vide(P):
    return len(P) == 0        # Vrai si la pile est vide
````

## Exercice 1

> ```python
> P = creer_pile()
> empile(P,3)
> empile(P,5)
> n = depile(P)
> empile(P,7)
> empile(P,9)
> ```
> A la fin de cette séquence, quelles valeurs contient la variable **`n`** ? La pile **`P`** ?

### Déroulement à la main (sommet à droite)

| Instruction | Pile P | n |
|:---|:---|:---|
| `P = creer_pile()` | `[]` | |
| `empile(P,3)` | `[3]` | |
| `empile(P,5)` | `[3, 5]` | |
| `n = depile(P)` | `[3]` | **5** (dernier entré) |
| `empile(P,7)` | `[3, 7]` | 5 |
| `empile(P,9)` | `[3, 7, 9]` | 5 |

**Réponse :** `n` vaut **5** et la pile contient **3, 7, 9**, avec **9 au sommet**.

````python
# Vérification
P = creer_pile()
empile(P,3)
empile(P,5)
n = depile(P)
empile(P,7)
empile(P,9)
print("n =", n)
print("P =", P, "(sommet à droite)")
````

## Exercice 2

> ```python
> P = creer_pile()
> for i in range(5) :
>     empile(P,i**2)
> ```

### Déroulement à la main

`range(5)` donne `i = 0, 1, 2, 3, 4`, et `i**2` vaut alors `0, 1, 4, 9, 16`. On empile dans cet ordre.

**Réponse :** la pile contient **0, 1, 4, 9, 16**, avec **16 au sommet** (c'est le dernier empilé, donc le premier qui sortira).

````python
# Vérification
P = creer_pile()
for i in range(5) :
    empile(P,i**2)
print("P =", P, "(sommet à droite)")

# Si on dépile tout, les valeurs sortent dans l'ordre inverse (LIFO)
while not est_vide(P):
    print(depile(P), end=" ")
````
