## Exercice 1

> On souhaite simuler une course de chevaux.
> On modélise les chevaux au départ de la course par une liste **`Chevaux`** de numéros compris entre 1 et 16. Un tiercé, c'est le tirage de trois numéros différents compris entre 1 et 16.
> Ecrire un code qui permet créer une liste **`Tierce`** qui contient trois éléments <u>distincts</u> de la liste **`Chevaux`** choisis aléatoirement.

### Méthode 1 : `shuffle` puis une tranche

````python
from random import choice, shuffle

# Construction de la liste des chevaux numérotés de 1 à 16
Chevaux = []
for numero in range(1, 17):
    Chevaux.append(numero)
print(Chevaux)

# On mélange la liste, puis on garde les 3 premiers éléments (indices 0 à 2)
shuffle(Chevaux)
Tierce = Chevaux[:3]
print(Tierce)
````

Les éléments de `Chevaux` sont tous différents, donc les 3 premiers après mélange sont forcément **distincts**. `Chevaux[:3]` renvoie les éléments d'indices 0 à 2.

### Méthode 2 : `choice` puis `remove`

````python
# On reconstruit la liste des chevaux
Chevaux = []
for numero in range(1, 17):
    Chevaux.append(numero)

Tierce = []
for k in range(3):
    cheval = choice(Chevaux)    # tirage au hasard d'un cheval
    Tierce.append(cheval)       # on l'ajoute au tiercé
    Chevaux.remove(cheval)      # on le retire pour ne pas le tirer une 2e fois
print(Tierce)
````

Chaque cheval tiré est **retiré** de `Chevaux`, donc il ne peut plus être tiré : les trois numéros sont distincts. *(Le résultat change à chaque exécution : c'est du hasard.)*

## Exercice 2

> Les fonctions **`min(Liste)`** et **`max(Liste)`** renvoient respectivement les plus petit et plus grand élément d'une liste.
> Ecrire une fonction **`extrema(Liste)`** qui, sans utiliser les deux fonctions précédentes, renvoie une liste contenant deux tupples **`[(min,imin),(max, imax)]`** ou juste **`[]`** si la liste est vide.
>
> Exemple : `Liste = [11, 19, 14, 19, 19, 17, 10, 16, 11, 18]` → `[(10, 6), (19, 1)]`

````python
def extrema(Liste):
    # Cas de la liste vide
    if len(Liste) == 0:
        return []
    # On suppose au départ que le 1er élément est à la fois le min et le max
    mini, imini = Liste[0], 0
    maxi, imaxi = Liste[0], 0
    # On parcourt les indices suivants
    for i in range(1, len(Liste)):
        if Liste[i] < mini:        # strictement plus petit : nouveau minimum
            mini, imini = Liste[i], i
        if Liste[i] > maxi:        # strictement plus grand : nouveau maximum
            maxi, imaxi = Liste[i], i
    return [(mini, imini), (maxi, imaxi)]

Liste = [11, 19, 14, 19, 19, 17, 10, 16, 11, 18]
print(extrema(Liste))
print(extrema([]))
````

Résultat attendu :
```
[(10, 6), (19, 1)]
[]
```
**Pourquoi l'indice 1 pour le max ?** La valeur 19 apparaît aux indices 1, 3 et 4. On compare avec `>` (strict), donc un 19 déjà trouvé ne remplace pas le premier : on garde **la première occurrence**, comme dans l'exemple de l'énoncé.
