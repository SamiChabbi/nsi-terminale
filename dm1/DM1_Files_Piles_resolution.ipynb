# EXERCICE 1 — Le panier du supermarché (file + POO)

#### 📝 ÉNONCÉ

> *Cet exercice porte sur les structures de données (files et la programmation objet en langage python).*
>
> Un supermarché met en place un système de passage automatique en caisse. Un client scanne les articles à l'aide d'un scanner de code-barres au fur et à mesure qu'il les ajoute dans son panier. Les articles s'enregistrent alors dans une structure de données.
>
> La structure de données utilisée est une file définie par la classe `Panier`, avec les primitives habituelles sur la structure de file. Pour faciliter la lecture, le code de la classe `Panier` n'est pas écrit.
>
> ```python
> class Panier():
>     def __init__(self):
>         """Initialise la file comme une file vide."""
>
>     def est_vide(self):
>         """Renvoie True si la file est vide, False sinon."""
>
>     def enfiler(self, e):
>         """Ajoute l'élément e en dernière position de la file, ne renvoie rien."""
>
>     def defiler(self):
>         """Retire le premier élément de la file et le renvoie."""
> ```
>
> Le panier d'un client sera représenté par une file contenant les articles scannés. Les articles sont représentés par des tuples `(code_barre, designation, prix, horaire_scan)` où
> - `code_barre` est un nombre entier identifiant l'article ;
> - `designation` est une chaine de caractères qui pourra être affichée sur le ticket de caisse ;
> - `prix` est un nombre décimal donnant le prix d'une unité de cet article ;
> - `horaire_scan` est un nombre entier de secondes permettant de connaitre l'heure où l'article a été scanné.

#### 💡 EXPLICATION PÉDAGOGIQUE

**Ce qu'il faut comprendre avant de commencer :**
- Une **file** suit le principe **FIFO**, « premier arrivé, premier sorti » (Leçon 02E). Le premier article scanné est en **tête** de file, le dernier est en **queue**.
- On ne connaît **pas** le code interne de `Panier`. On doit donc utiliser **uniquement** les 4 primitives : `est_vide`, `enfiler`, `defiler` (et le constructeur). C'est le principe du **TDA** (Leçon 02A) : on se sert des opérations sans savoir comment elles sont programmées.
- Un article est un **tuple** de 4 valeurs. On y accède par indice :

| Indice | 0 | 1 | 2 | 3 |
|:---|:---|:---|:---|:---|
| Champ | `code_barre` | `designation` | `prix` | `horaire_scan` |

- Dans le sujet, l'indentation des méthodes de `Panier` est décalée (on dirait qu'elles sont *dans* `__init__`). C'est une erreur de mise en page : ce sont bien 4 méthodes de la classe.

**Pour pouvoir tester nos réponses**, on écrit ci-dessous une version de `Panier` avec une liste, sur le modèle de la classe `File` de la Leçon 02E. **Elle n'est pas demandée dans le DM.**

````python
# ⚙️ Implémentation de test (NON demandée dans le DM)
class Panier():
    def __init__(self):
        """Initialise la file comme une file vide."""
        self.articles = []

    def est_vide(self):
        """Renvoie True si la file est vide, False sinon."""
        return self.articles == []

    def enfiler(self, e):
        """Ajoute l'élément e en dernière position de la file, ne renvoie rien."""
        self.articles.append(e)

    def defiler(self):
        """Retire le premier élément de la file et le renvoie."""
        return self.articles.pop(0)

    def __str__(self):
        """Affichage (pour les tests) : du premier article scanné au dernier."""
        return str(self.articles)
````

## Question 2

#### 📝 ÉNONCÉ

> On souhaite définir une méthode `remplir(panier_temp)` dans la classe `Panier` permettant de remplir la file avec tout le contenu d'un autre panier `panier_temp` qui est un objet de type `Panier`.
> Recopier et compléter le code de la méthode `remplir` en remplaçant chaque …………… par la primitive de file qui convient.
>
> ```python
> def remplir(self, panier_temp):
>     while not panier_temp. …………… :
>         article = panier_temp. ……………
>         self. ……………(article)
> ```

#### ✅ RÉSOLUTION

````python
def remplir(self, panier_temp):
    while not panier_temp.est_vide():
        article = panier_temp.defiler()
        self.enfiler(article)

Panier.remplir = remplir     # (pour les tests : on ajoute la méthode à la classe)

# Test
p_temp = Panier()
p_temp.enfiler((1, "pain", 1.20, 100))
p_temp.enfiler((2, "lait", 0.95, 130))
p = Panier()
p.remplir(p_temp)
print("p      :", p)
print("p_temp :", p_temp)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

**Lecture ligne par ligne :**
1. `while not panier_temp.est_vide()` : **tant que** `panier_temp` contient encore des articles ;
2. `article = panier_temp.defiler()` : on **retire** l'article de tête de `panier_temp` et on le garde dans `article` ;
3. `self.enfiler(article)` : on l'**ajoute** à la fin de notre panier (`self`).

**Pourquoi l'ordre est conservé :** on retire par la tête et on ajoute par la queue. Le 1er article sorti est donc le 1er entré dans `self`. C'est la propriété FIFO.

**À retenir :** après `remplir`, `panier_temp` est **vide**. Les articles ont été *déplacés*, pas copiés. On va s'en servir à la question 3.

⚠️ Dans la classe, on écrirait `def remplir(self, panier_temp):` **indenté dans** `class Panier`. La ligne `Panier.remplir = remplir` ne sert qu'à tester ici ; elle n'est pas à écrire sur la copie.

## Question 3

#### 📝 ÉNONCÉ

> Pour que le client puisse connaitre à tout moment le montant de son panier, on souhaite ajouter une méthode `prix_total()` à la classe `Panier` qui renvoie la somme des prix de tous les articles présents dans le panier.
> Ecrire le code de la méthode `prix_total`. Attention, après l'appel de cette méthode, le panier devra toujours contenir ses articles.

#### ✅ RÉSOLUTION

````python
def prix_total(self):
    total = 0
    panier_temp = Panier()             # panier vide pour garder les articles de côté
    while not self.est_vide():
        article = self.defiler()       # on sort l'article de tête
        total = total + article[2]     # on ajoute son prix (indice 2 du tuple)
        panier_temp.enfiler(article)   # on le met de côté
    self.remplir(panier_temp)          # on remet tous les articles dans le panier
    return total

Panier.prix_total = prix_total         # (pour les tests)

# Test
p = Panier()
p.enfiler((31002, "café noir", 1.50, 50525))
p.enfiler((20001, "pain", 1.20, 50540))
p.enfiler((40005, "lait", 0.95, 50600))
print("Prix total :", p.prix_total())
print("Panier après l'appel :", p)
````

Résultat attendu : `3.65` (à un arrondi près : les nombres décimaux peuvent afficher `3.6500000000000004`), et le panier contient toujours ses 3 articles, **dans le même ordre**.

#### 💡 EXPLICATION PÉDAGOGIQUE

**La difficulté :** pour lire les articles d'une file, on doit les **défiler**, ce qui **vide** le panier. Or l'énoncé exige qu'il soit intact à la fin.

**La méthode classique (à connaître par cœur) :** vider dans une structure temporaire, puis tout remettre.
1. on défile chaque article ;
2. on ajoute son prix `article[2]` au total ;
3. on l'enfile dans `panier_temp` ;
4. à la fin, `self.remplir(panier_temp)` remet tout dans `self`. On **réutilise la question 2** : c'est souvent l'intention du sujet !

**Pourquoi l'ordre est conservé :** une file vidée dans une autre file garde son ordre (FIFO → FIFO). Avec une **pile**, l'ordre serait inversé (comme dans l'exercice 2 de la Leçon 02E).

**Erreurs courantes :**
- ❌ oublier de remettre les articles : le panier est vide après l'appel ;
- ❌ prendre `article[3]` (l'horaire) ou `article[1]` (le nom) au lieu de `article[2]` ;
- ❌ écrire `self.articles` : on ne connaît pas l'intérieur de la classe, on n'utilise que les primitives ;
- ❌ mettre `return total` **dans** la boucle : la fonction s'arrêterait au premier article.

## Question 4

#### 📝 ÉNONCÉ

> Le magasin souhaite connaitre pour chaque client la durée des achats. Cette durée sera obtenue en faisant la différence entre le champ `horaire_scan` du dernier article scanné et le champ `horaire_scan` du premier article scanné dans le panier du client. Un panier vide renverra une durée égale à zéro. On pourra accepter que le panier soit vide après l'appel de cette méthode.
> Ecrire une méthode `duree_courses` de la classe `Panier` qui renvoie cette durée.

#### ✅ RÉSOLUTION

````python
def duree_courses(self):
    if self.est_vide():
        return 0
    premier = self.defiler()          # le premier article scanné est en tête de file
    dernier = premier                 # s'il n'y a qu'un article, il est aussi le dernier
    while not self.est_vide():
        dernier = self.defiler()      # à la fin de la boucle : le dernier article scanné
    return dernier[3] - premier[3]    # indice 3 = horaire_scan

Panier.duree_courses = duree_courses  # (pour les tests)

# Tests
p = Panier()
p.enfiler((31002, "café noir", 1.50, 50525))
p.enfiler((20001, "pain", 1.20, 50540))
p.enfiler((40005, "lait", 0.95, 50600))
print("Durée (3 articles) :", p.duree_courses(), "secondes")

p = Panier()
p.enfiler((31002, "café noir", 1.50, 50525))
print("Durée (1 article)  :", p.duree_courses())

print("Durée (panier vide):", Panier().duree_courses())
````

Résultats attendus : `75` secondes (50600 − 50525), puis `0`, puis `0`.

#### 💡 EXPLICATION PÉDAGOGIQUE

- Dans une file, le **premier** article scanné est en **tête** : c'est le premier `defiler()`.
- Le **dernier** article scanné est en **queue**. On ne peut l'atteindre qu'en défilant **tous** les autres. On garde à chaque tour le dernier article sorti dans la variable `dernier`.
- L'énoncé autorise à vider le panier (« On pourra accepter que le panier soit vide après l'appel »). Pas besoin de tout remettre, contrairement à la question 3.

**Les 3 cas à vérifier (bon réflexe de test) :**

| Cas | Ce qui se passe | Résultat |
|:---|:---|:---|
| panier vide | `return 0` immédiat | 0 |
| 1 article | `dernier = premier`, la boucle ne tourne pas | 0 |
| plusieurs articles | la boucle va jusqu'au dernier | dernier − premier |

**Erreurs courantes :**
- ❌ oublier le cas du panier vide : `defiler()` sur une file vide provoque une **erreur** ;
- ❌ oublier `dernier = premier` : avec un seul article, la variable `dernier` n'existe pas (`NameError`) ;
- ❌ faire `premier - dernier` : on obtiendrait une durée **négative**.

# EXERCICE 2 — Défi Tubes (pile + POO + algorithmique)

#### 📝 ÉNONCÉ

> *Cet exercice porte sur la structure de pile, la programmation objet et l'algorithmique.*
>
> Défi Tubes est un jeu à un joueur. Le joueur dispose de 4 tubes. Chaque tube peut contenir de 0 à 3 phases. Chaque phase possède une couleur. Il y a 3 couleurs possibles. On peut s'imaginer ces phases comme des palets de couleur dans le tube. Pour modéliser les couleurs, on utilisera les entiers 1, 2 et 3. Lorsqu'un tube contient 0 phase, on dit que le tube est vide. Lorsqu'il en a 3, on dit qu'il est plein. Lorsqu'un tube n'est pas vide, sa dernière couleur est la couleur de sa phase supérieure.
>
> Le jeu Défi Tube consiste à verser successivement la dernière couleur des tubes dans les autres tubes avec les contraintes suivantes :
> - on ne peut rien verser dans un tube plein ;
> - pour verser un tube 1 dans un tube 2, il faut que la dernière couleur du tube 1 soit la même que celle du tube 2 ou que le tube 2 soit vide. Dans ces deux cas, on retire la dernière couleur du tube 1 pour qu'elle devienne la dernière couleur du tube 2. On réitère cela tant que la dernière couleur du tube 1 est la même et que le tube 2 n'est pas plein.
>
> Le jeu se termine lorsque 3 des 4 tubes sont pleins et que leurs 3 phases sont de même couleur.
>
> Les figures 2 à 5 (le haut du tube est en haut) :
>

## Question 2

#### 📝 ÉNONCÉ

> Expliquer ce qu'est la structure de pile en précisant ce que sont les méthodes `empiler` et `depiler`.

#### ✅ RÉSOLUTION

Une **pile** est une structure de données linéaire fondée sur le principe **LIFO** (*Last In, First Out*, « dernier arrivé, premier sorti ») : on n'a accès qu'à l'élément situé au **sommet**, c'est-à-dire le dernier ajouté.

- **`empiler(e)`** ajoute l'élément `e` au **sommet** de la pile.
- **`depiler()`** retire l'élément situé au **sommet** de la pile et le **renvoie**.

#### 💡 EXPLICATION PÉDAGOGIQUE

- C'est la définition de la **Leçon 02D**, à savoir réciter. L'image à donner : une **pile d'assiettes**, ou ici un **tube** dans lequel on ne peut prendre que le palet du haut.
- Les mots-clés attendus par le correcteur : **LIFO**, **sommet**, « dernier arrivé, premier sorti », et le fait que `depiler` **renvoie** l'élément (pas seulement le supprime).

## Question 3

#### 📝 ÉNONCÉ

> Expliquer les lignes 11 et 12 du code de la classe `tube`.

#### ✅ RÉSOLUTION

- **Ligne 11** `self.contenu[self.taille] = couleur` : on place la couleur dans la **première case libre** du tube. Comme les indices commencent à 0, la case d'indice `self.taille` est juste au-dessus de la dernière couleur.
- **Ligne 12** `self.taille = self.taille + 1` : le tube contient **un élément de plus**, on augmente donc la taille de 1.

(Ces deux lignes ne sont exécutées que si le tube n'est pas plein, grâce au test `if self.taille < 3` de la ligne 10.)

#### 💡 EXPLICATION PÉDAGOGIQUE

Exemple avec le tube de la figure 6 (`contenu = [1, 3, 0]`, `taille = 2`) et `empiler(2)` :

| Étape | `contenu` | `taille` |
|:---|:---|:---|
| avant | `[1, 3, 0]` | 2 |
| ligne 11 : `contenu[2] = 2` | `[1, 3, 2]` | 2 |
| ligne 12 | `[1, 3, 2]` | 3 |

⚠️ **L'ordre des deux lignes compte.** Si on augmentait `taille` d'abord, on écrirait dans `contenu[3]`, qui n'existe pas (`IndexError`).

````python
class tube:
    def __init__(self):
        self.taille = 0
        self.contenu = [0, 0, 0]

    def est_vide(self):
        return self.taille == 0

    def empiler(self, couleur):
        if self.taille < 3:
            self.contenu[self.taille] = couleur
            self.taille = self.taille + 1

# Vérification de l'exemple ci-dessus
t = tube()
t.taille = 2
t.contenu = [1, 3, 0]
t.empiler(2)
print(t.contenu, t.taille)
````

## Question 4

#### 📝 ÉNONCÉ

> Recopier et compléter le code de la méthode `depiler` précédente. Lorsque le tube est vide, la méthode `depiler` doit renvoyer -1.

#### ✅ RÉSOLUTION

````python
def depiler(self):
    if self.taille > 0:
        self.taille = self.taille - 1
        couleur = self.contenu[self.taille]
        self.contenu[self.taille] = 0
        return couleur
    else:
        return -1

tube.depiler = depiler        # (pour les tests)

# Tests
t = tube()
t.taille = 2
t.contenu = [1, 3, 0]
print(t.depiler(), t.contenu, t.taille)   # 3 [1, 0, 0] 1
print(t.depiler(), t.contenu, t.taille)   # 1 [0, 0, 0] 0
print(t.depiler())                        # -1 (tube vide)
````

Les trois trous à compléter sont donc :
- ligne 17 : `couleur = self.contenu[self.taille]`
- ligne 19 : `return couleur`
- ligne 21 : `return -1`

#### 💡 EXPLICATION PÉDAGOGIQUE

- La ligne 16 a **déjà** diminué `taille` de 1. `self.taille` est donc maintenant l'indice du **sommet**, la case à vider. C'est l'opération **inverse** de `empiler` : on recule d'abord, puis on lit.
- Ligne 17 : on **lit** la couleur **avant** de mettre la case à 0 (ligne 18), sinon on la perdrait.
- Ligne 19 : une méthode `depiler` doit **renvoyer** l'élément retiré.

⚠️ **Erreur fréquente :** écrire `self.contenu[self.taille - 1]` à la ligne 17. On oublie que `taille` a déjà été diminuée ligne 16, et on lit la mauvaise case.

## Question 6

#### 📝 ÉNONCÉ

> Écrire une méthode `est_homogene` de la classe `tube` qui renvoie `True` si le tube est plein et si son contenu est composé de trois fois la même couleur, et qui renvoie `False` sinon.

#### ✅ RÉSOLUTION

````python
def est_homogene(self):
    return (self.est_plein()
            and self.contenu[0] == self.contenu[1]
            and self.contenu[1] == self.contenu[2])

tube.est_homogene = est_homogene    # (pour les tests)

t = tube(); t.contenu = [2, 2, 2]; t.taille = 3
print(t.est_homogene())             # True
t = tube(); t.contenu = [2, 1, 1]; t.taille = 3
print(t.est_homogene())             # False
t = tube(); t.contenu = [2, 2, 0]; t.taille = 2
print(t.est_homogene())             # False (pas plein)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Deux conditions **à la fois**, d'où le `and` : le tube est **plein**, et les **3 cases sont égales**.
- On **réutilise** `est_plein()` de la question 5. Sur une copie, réutiliser les méthodes déjà écrites est apprécié.
- Pour vérifier que 3 valeurs sont égales, il suffit de **deux** comparaisons : case 0 = case 1, **et** case 1 = case 2.
- Pourquoi tester « plein » ? Sans ce test, un tube **vide** `[0, 0, 0]` aurait 3 cases égales et serait considéré comme homogène !

## Question 7

#### 📝 ÉNONCÉ

> Écrire une méthode `derniere_couleur` de la classe `tube` qui renvoie le numéro de la dernière couleur du tube. Si le tube est vide, la méthode renverra la valeur -1.

#### ✅ RÉSOLUTION

````python
def derniere_couleur(self):
    if self.est_vide():
        return -1
    return self.contenu[self.taille - 1]

tube.derniere_couleur = derniere_couleur    # (pour les tests)

t = tube(); t.contenu = [1, 3, 0]; t.taille = 2
print(t.derniere_couleur())                 # 3
print(tube().derniere_couleur())            # -1
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- La dernière couleur est au **sommet**, à l'indice `taille - 1` (voir le schéma de la partie A).
- On **lit** la couleur sans la retirer. C'est la différence avec `depiler`, qui retire en plus.
- ❌ `self.contenu[-1]` est **faux** ici : la liste fait toujours 3 cases, donc `contenu[-1]` est toujours la case du haut, même quand elle vaut 0.
- ❌ Oublier le cas vide : `contenu[0 - 1]` donnerait `contenu[-1]`, c'est-à-dire 0, au lieu de -1.

## Question 8

#### 📝 ÉNONCÉ

> Le code incomplet d'une méthode `verser` de la classe `tube` est donné ci-dessous :
>
> ```python
> 1 def verser(self, other):
> 2     while ...
> 3         couleur = self.depiler()
> 4         other.empiler(couleur)
> ```
>
> Recopier et compléter le code de cette méthode `verser` afin de verser l'instance `self` de la classe `tube` dans l'instance `other`. On veillera à vérifier toutes les conditions nécessaires au bon déroulement de cette opération.

#### ✅ RÉSOLUTION

````python
def verser(self, other):
    while (not self.est_vide()
           and not other.est_plein()
           and (other.est_vide() or self.derniere_couleur() == other.derniere_couleur())):
        couleur = self.depiler()
        other.empiler(couleur)

tube.verser = verser    # (pour les tests)
````

Sur la copie, on peut écrire la condition sur une seule ligne :
```python
while not self.est_vide() and not other.est_plein() and (other.est_vide() or self.derniere_couleur() == other.derniere_couleur()):
```

#### 💡 EXPLICATION PÉDAGOGIQUE

**On traduit chaque règle du jeu en condition :**

| Règle de l'énoncé | Condition Python |
|:---|:---|
| il faut quelque chose à verser | `not self.est_vide()` |
| « on ne peut rien verser dans un tube plein » | `not other.est_plein()` |
| « même dernière couleur **ou** tube 2 vide » | `other.est_vide() or self.derniere_couleur() == other.derniere_couleur()` |

- Le `while` (et pas un `if`) traduit « **On réitère** cela tant que… » : on verse palet par palet tant que c'est possible.
- Les **parenthèses autour du `or` sont indispensables**. Sans elles, Python évalue d'abord les `and`, et la condition change de sens.
- Pourquoi la boucle s'arrête quand la couleur change : après chaque versement, le sommet de `other` est la couleur qu'on vient de verser. Si le nouveau sommet de `self` est différent, la 3ᵉ condition devient fausse.

**Erreur courante :** oublier `not self.est_vide()`. Sur un tube vide, `depiler()` renverrait -1, et on empilerait la « couleur » -1 dans `other` !

````python
# Vérification sur les exemples du sujet
t1 = tube(); t1.contenu = [1, 3, 0]; t1.taille = 2
t2 = tube(); t2.contenu = [3, 3, 0]; t2.taille = 2
t1.verser(t2)
print("1 dans 2 :", t1.contenu, t2.contenu)     # [1, 0, 0] [3, 3, 3]  (figure 3)

t4 = tube(); t4.contenu = [2, 1, 1]; t4.taille = 3
t4.verser(t1)
print("4 dans 1 :", t4.contenu, t1.contenu)     # [2, 0, 0] [1, 1, 1]  (figure 4 : deux palets versés)

t3 = tube(); t3.contenu = [2, 2, 0]; t3.taille = 2
t3.verser(t2)
print("3 dans 2 :", t3.contenu, t2.contenu)     # rien ne bouge : tube 2 plein
````

## Question 10

#### 📝 ÉNONCÉ

> Écrire une fonction `gagne` qui prend comme argument un état et qui renvoie `True` si la partie est terminée et `False` sinon.

#### ✅ RÉSOLUTION

````python
def gagne(etat):
    nb_homogenes = 0
    for t in etat:
        if t.est_homogene():
            nb_homogenes = nb_homogenes + 1
    return nb_homogenes >= 3

# Tests : on rejoue la partie du sujet (etat est déjà à la figure 3)
print("Figure 3 :", gagne(etat))       # False
etat[3].verser(etat[0])                # tube 4 dans tube 1 -> figure 4
print("Figure 4 :", gagne(etat))       # False
etat[2].verser(etat[3])                # tube 3 dans tube 4 -> figure 5
print("Figure 5 :", gagne(etat))       # True
afficher(etat)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- La partie est terminée quand **3 des 4 tubes** sont pleins et d'une seule couleur, c'est-à-dire **homogènes**. On réutilise `est_homogene()` (question 6).
- C'est un **algorithme de comptage** classique : un compteur à 0, une boucle sur tous les éléments, `+ 1` quand la condition est vraie.
- `gagne` est une **fonction** (pas une méthode) : elle prend `etat` en paramètre, et il n'y a pas de `self`.
- On renvoie `nb_homogenes >= 3` (« au moins 3 »). Dans ce jeu, il y a 3 palets de chaque couleur : on ne peut pas avoir 4 tubes homogènes. `== 3` serait donc accepté aussi.

**Erreur courante :**
```python
for t in etat:
    if not t.est_homogene():
        return False      # ❌ faux : un tube non homogène (le tube vide) est normal !
return True
```
Le 4ᵉ tube n'est **jamais** homogène quand on gagne (il est vide). Il faut **compter**, pas exiger que tous les tubes soient homogènes.
