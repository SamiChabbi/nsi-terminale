#### ✅ RÉSOLUTION

````python
from math import pi


class Angle:

    def __init__(self, v):
        assert 0 <= v <= 180, "La mesure doit être comprise entre 0 et 180 degrés"
        self.v = v

    def mesure_en_degres(self):
        return self.v

    def est_aigu(self):
        return self.v < 90

    def mesure_en_radians(self):
        return self.v * pi / 180

    def mesure_en_degres_minutes_secondes(self):
        total = round(self.v * 3600)          # mesure en secondes d'angle (entier)
        degres = total // 3600
        minutes = (total % 3600) // 60
        secondes = total % 60
        return (degres, minutes, secondes)

    def supplementaire(self):
        return Angle(180 - self.v)

    def __str__(self):
        return str(self.v) + "°"

    def __add__(self, autre):
        return Angle(self.v + autre.v)

    def __sub__(self, autre):
        return Angle(self.v - autre.v)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

L'énoncé **est** l'API : pour chaque méthode, il donne son nom, ses paramètres, ce qu'elle renvoie et ses préconditions. Le travail consiste à écrire le code qui respecte ce contrat.

**Méthode par méthode :**
- **`__init__(v)`** : la précondition `0 ⩽ v ⩽ 180` est vérifiée **dès la création** avec `assert`. Si elle est fausse, Python lève une `AssertionError`. Ainsi, aucun objet `Angle` invalide ne peut exister.
  - ⚠️ **Il faut bien `assert`, et pas `raise ValueError`** : les tests du fichier attrapent une `AssertionError` (`except AssertionError`). Avec une autre erreur, les tests s'arrêtent dès `Angle(-1)`.
- **`mesure_en_degres()`** : renvoie simplement l'attribut. On passe quand même par une méthode : c'est l'API qui le demande, et l'utilisateur n'a pas besoin de connaître le nom de l'attribut (**encapsulation**).
- **`est_aigu()`** : la comparaison `self.v < 90` est **déjà** un booléen. On le renvoie directement, sans `if ... return True else return False`.
- **`mesure_en_radians()`** : 180° = π rad, donc on multiplie par `pi / 180` (`pi` vient du module `math`, comme le suggère le commentaire du fichier).
- **`mesure_en_degres_minutes_secondes()`** : 1° = 60 minutes et 1 minute = 60 secondes. On convertit tout **en secondes entières** (`round(v * 3600)`), puis on découpe avec `//` et `%`, exactement comme pour convertir des secondes en heures, minutes et secondes. La cellule suivante montre pourquoi on arrondit.
- **`supplementaire()`** : renvoie un **nouvel objet** `Angle(180 - v)`, sans modifier l'angle de départ.
- **`__str__()`** : méthode spéciale appelée par `print(a)` et `str(a)`. `str(45)` donne `"45"` et `str(30.5)` donne `"30.5"`, ce qu'attendent les tests.
- **`__add__(autre)`** et **`__sub__(autre)`** : méthodes spéciales appelées par `a + b` et `a - b`. On crée un nouvel `Angle` : si le résultat sort de [0 ; 180], **c'est l'`assert` du constructeur qui lève l'erreur**. C'est le sens du commentaire du fichier « le résultat doit lui-même être un angle géométrique valide ».

*Pourquoi `round` ? La méthode suggérée par les commentaires du fichier (partie entière, puis partie décimale), écrite avec `int`, passe les tests du fichier, mais se trompe sur d'autres angles :*

````python
def dms_naif(v):
    degres = int(v)                      # partie entière : les degrés
    minutes_dec = (v - degres) * 60      # partie décimale convertie en minutes
    minutes = int(minutes_dec)
    secondes = int((minutes_dec - minutes) * 60)
    return (degres, minutes, secondes)


print(dms_naif(30 + 15 / 60 + 20 / 3600))   # (30, 15, 20) : le test du fichier passe...
print(dms_naif(12.7))                       # (12, 41, 59) au lieu de (12, 42, 0) !
print(Angle(12.7).mesure_en_degres_minutes_secondes())
````

#### 💡 EXPLICATION PÉDAGOGIQUE

Les nombres à virgule sont stockés de façon **approchée** : pour 12,7°, la partie décimale 0,7 × 60 donne 41,99999… minutes au lieu de 42. `int` coupe la partie décimale et donne 41 minutes, puis 59 secondes.

Deux solutions : arrondir les secondes à la fin (`round(...)` au lieu de `int(...)`), ou bien, comme dans la résolution, tout convertir en secondes **entières** dès le début avec `round(v * 3600)`.

# Vérification avec les tests du fichier `angle.py`

*Les tests recopiés tels qu'ils sont dans le fichier du cours :*

````python
# ============================================================
# Tests de la classe Angle
# ============================================================

# ------------------------------------------------------------
# 1. Construction et mesure en degrés
# ------------------------------------------------------------

assert Angle(0).mesure_en_degres() == 0
assert Angle(45).mesure_en_degres() == 45
assert Angle(90).mesure_en_degres() == 90
assert Angle(180).mesure_en_degres() == 180

# Les valeurs en dehors de [0, 180] doivent être refusées
for valeur in [-1, -10, 180.1, 200]:
    try:
        Angle(valeur)
        assert False, f"Valeur acceptée à tort : {valeur}"
    except AssertionError:
        pass

# ------------------------------------------------------------
# 2. Est-ce un angle aigu ?
# ------------------------------------------------------------

assert Angle(0).est_aigu() is True
assert Angle(45).est_aigu() is True
assert Angle(89.9).est_aigu() is True

assert Angle(90).est_aigu() is False
assert Angle(120).est_aigu() is False
assert Angle(180).est_aigu() is False

# ------------------------------------------------------------
# 3. Conversion en radians
# ------------------------------------------------------------

import math

assert math.isclose(Angle(0).mesure_en_radians(), 0)

assert math.isclose(Angle(90).mesure_en_radians(), math.pi / 2)

assert math.isclose(Angle(180).mesure_en_radians(), math.pi)

assert math.isclose(Angle(45).mesure_en_radians(), math.pi / 4)

# ------------------------------------------------------------
# 4. Conversion en degrés / minutes / secondes
# ------------------------------------------------------------

assert Angle(0).mesure_en_degres_minutes_secondes() == (0, 0, 0)

assert Angle(30).mesure_en_degres_minutes_secondes() == (30, 0, 0)

assert Angle(30.5).mesure_en_degres_minutes_secondes() == (30, 30, 0)

assert Angle(30.25).mesure_en_degres_minutes_secondes() == (30, 15, 0)

assert Angle(30 + 15 / 60 +
             20 / 3600).mesure_en_degres_minutes_secondes() == (30, 15, 20)

# ------------------------------------------------------------
# 5. Angle supplémentaire
# ------------------------------------------------------------

assert Angle(0).supplementaire().mesure_en_degres() == 180
assert Angle(30).supplementaire().mesure_en_degres() == 150
assert Angle(90).supplementaire().mesure_en_degres() == 90
assert Angle(180).supplementaire().mesure_en_degres() == 0

# ------------------------------------------------------------
# 6. Affichage
# ------------------------------------------------------------

assert str(Angle(0)) == "0°"
assert str(Angle(45)) == "45°"
assert str(Angle(90)) == "90°"
assert str(Angle(30.5)) == "30.5°"
assert str(Angle(180)) == "180°"

# ------------------------------------------------------------
# 7. Addition
# ------------------------------------------------------------

assert (Angle(10) + Angle(20)).mesure_en_degres() == 30
assert (Angle(45) + Angle(45)).mesure_en_degres() == 90
assert (Angle(90) + Angle(90)).mesure_en_degres() == 180

# Dépassement de 180° : le résultat doit être refusé
try:
    Angle(120) + Angle(61)
    assert False, "Une somme supérieure à 180° a été acceptée"
except AssertionError:
    pass

# ------------------------------------------------------------
# 8. Soustraction
# ------------------------------------------------------------

assert (Angle(30) - Angle(20)).mesure_en_degres() == 10
assert (Angle(90) - Angle(30)).mesure_en_degres() == 60
assert (Angle(30) - Angle(30)).mesure_en_degres() == 0
assert (Angle(180) - Angle(90)).mesure_en_degres() == 90

# Résultat négatif : il doit être refusé
try:
    Angle(20) - Angle(30)
    assert False, "Une différence négative a été acceptée"
except AssertionError:
    pass

print("Tous les tests sont réussis !")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

`Tous les tests sont réussis !` : la classe respecte l'API.

**Remarque sur les tests de refus** (valeurs hors de [0 ; 180], somme trop grande, différence négative) :
```python
try:
    Angle(valeur)
    assert False, ...
except AssertionError:
    pass
```
Si la classe accepte la valeur à tort, c'est `assert False` qui lève une `AssertionError`… qui est aussitôt attrapée par le `except`. Ces tests passent donc **même si le constructeur ne vérifie rien**. Pour vérifier soi-même que les refus fonctionnent, la cellule suivante affiche le message d'erreur.

````python
for essai in [-1, 200]:
    try:
        Angle(essai)
        print(essai, ": accepté (c'est une erreur !)")
    except AssertionError as e:
        print(essai, ": refusé ->", e)

try:
    Angle(120) + Angle(61)
    print("120 + 61 : accepté (c'est une erreur !)")
except AssertionError as e:
    print("120 + 61 : refusé ->", e)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

**À retenir :**
- `__str__`, `__add__`, `__sub__`... sont des **méthodes spéciales** : on ne les appelle pas directement, Python les appelle pour nous (`print`, `+`, `-`).
- Une méthode qui « renvoie un Angle » doit renvoyer un **objet** `Angle(...)`, pas un nombre.
- Vérifier les préconditions dans le constructeur (avec `assert`) garantit que **tout objet créé est valide**.
- Les `try / except` ne sont pas à savoir écrire : ils servent seulement à vérifier.
