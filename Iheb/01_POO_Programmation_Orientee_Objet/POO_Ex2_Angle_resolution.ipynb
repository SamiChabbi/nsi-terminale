## 5. `mesure_en_degres_minutes_secondes`

#### 📝 ÉNONCÉ

> - `mesure_en_degres_minutes_secondes() -> tuple` renvoie un triplet (degrés, minutes, secondes) représentant la mesure de l'angle. Exemple : 12,5° correspond à (12,30,0).

#### ✅ RÉSOLUTION

````python
from math import pi


class Angle:

    def __init__(self, v):
        assert 0 <= v <= 180, "La mesure doit être comprise entre 0 et 180 degrés"
        self.v = v

    def mesure_en_degres(self):
        return self.v

    def est_aigu(self):
        return self.v < 90

    def mesure_en_radians(self):
        return self.v * pi / 180

    def mesure_en_degres_minutes_secondes(self):# NOUVEAU
        total = round(self.v * 3600)          # mesure en secondes d'angle (entier)
        degres = total // 3600
        minutes = (total % 3600) // 60
        secondes = total % 60
        return (degres, minutes, secondes)


# 4. Conversion en degrés / minutes / secondes (tests du fichier angle.py)

assert Angle(0).mesure_en_degres_minutes_secondes() == (0, 0, 0)

assert Angle(30).mesure_en_degres_minutes_secondes() == (30, 0, 0)

assert Angle(30.5).mesure_en_degres_minutes_secondes() == (30, 30, 0)

assert Angle(30.25).mesure_en_degres_minutes_secondes() == (30, 15, 0)

assert Angle(30 + 15 / 60 +
             20 / 3600).mesure_en_degres_minutes_secondes() == (30, 15, 20)

print("Tests de la partie 4 réussis ✔")

print(Angle(12.5).mesure_en_degres_minutes_secondes())     # exemple de l'énoncé : (12, 30, 0)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- 1° = 60 minutes et 1 minute = 60 secondes, donc 1° = 3600 secondes.
- On convertit tout **en secondes entières** (`round(v * 3600)`), puis on découpe avec `//` et `%`, exactement comme pour convertir des secondes en heures, minutes et secondes : `// 3600` donne les degrés, le reste `% 3600` divisé par 60 donne les minutes, et `% 60` les secondes.
- La cellule suivante montre pourquoi on arrondit.

*Pourquoi `round` ? La méthode suggérée par les commentaires du fichier (partie entière, puis partie décimale), écrite avec `int`, passe les tests du fichier, mais se trompe sur d'autres angles :*

````python
def dms_naif(v):
    degres = int(v)                      # partie entière : les degrés
    minutes_dec = (v - degres) * 60      # partie décimale convertie en minutes
    minutes = int(minutes_dec)
    secondes = int((minutes_dec - minutes) * 60)
    return (degres, minutes, secondes)


print(dms_naif(30 + 15 / 60 + 20 / 3600))   # (30, 15, 20) : le test du fichier passe...
print(dms_naif(12.7))                       # (12, 41, 59) au lieu de (12, 42, 0) !
print(Angle(12.7).mesure_en_degres_minutes_secondes())
````

#### 💡 EXPLICATION PÉDAGOGIQUE

Les nombres à virgule sont stockés de façon **approchée** : pour 12,7°, la partie décimale 0,7 × 60 donne 41,99999… minutes au lieu de 42. `int` coupe la partie décimale et donne 41 minutes, puis 59 secondes.

Deux solutions : arrondir les secondes à la fin (`round(...)` au lieu de `int(...)`), ou bien, comme dans la résolution, tout convertir en secondes **entières** dès le début avec `round(v * 3600)`.

## 6. `supplementaire`

#### 📝 ÉNONCÉ

> - `supplementaire() -> Angle` : renvoie un nouvel objet Angle représentant le supplémentaire de l'angle (le supplémentaire d'un angle de mesure v a pour mesure 180 − v degrés).

#### ✅ RÉSOLUTION

````python
from math import pi


class Angle:

    def __init__(self, v):
        assert 0 <= v <= 180, "La mesure doit être comprise entre 0 et 180 degrés"
        self.v = v

    def mesure_en_degres(self):
        return self.v

    def est_aigu(self):
        return self.v < 90

    def mesure_en_radians(self):
        return self.v * pi / 180

    def mesure_en_degres_minutes_secondes(self):
        total = round(self.v * 3600)          # mesure en secondes d'angle (entier)
        degres = total // 3600
        minutes = (total % 3600) // 60
        secondes = total % 60
        return (degres, minutes, secondes)

    def supplementaire(self):                 # NOUVEAU
        return Angle(180 - self.v)


# 5. Angle supplémentaire (tests du fichier angle.py)

assert Angle(0).supplementaire().mesure_en_degres() == 180
assert Angle(30).supplementaire().mesure_en_degres() == 150
assert Angle(90).supplementaire().mesure_en_degres() == 90
assert Angle(180).supplementaire().mesure_en_degres() == 0

print("Tests de la partie 5 réussis ✔")

a = Angle(30)
b = a.supplementaire()
print(a.mesure_en_degres(), b.mesure_en_degres())     # a n'a pas changé
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- La méthode doit renvoyer un **nouvel objet** `Angle`, et pas un nombre : `return Angle(180 - self.v)`.
- L'angle de départ n'est **pas modifié** : on ne touche pas à `self.v`.
- Comme `0 ⩽ v ⩽ 180`, la mesure `180 - v` est elle aussi entre 0 et 180 : le nouvel angle est toujours valide.

## 8. `__add__`

#### 📝 ÉNONCÉ

> - `__add__(autre) -> Angle` — Paramètre : autre, un objet de type Angle. Renvoie un nouvel objet Angle représentant la somme des deux angles (précondition : la somme doit être comprise entre 0° et 180°).

#### ✅ RÉSOLUTION

````python
from math import pi


class Angle:

    def __init__(self, v):
        assert 0 <= v <= 180, "La mesure doit être comprise entre 0 et 180 degrés"
        self.v = v

    def mesure_en_degres(self):
        return self.v

    def est_aigu(self):
        return self.v < 90

    def mesure_en_radians(self):
        return self.v * pi / 180

    def mesure_en_degres_minutes_secondes(self):
        total = round(self.v * 3600)          # mesure en secondes d'angle (entier)
        degres = total // 3600
        minutes = (total % 3600) // 60
        secondes = total % 60
        return (degres, minutes, secondes)

    def supplementaire(self):
        return Angle(180 - self.v)

    def __str__(self):
        return str(self.v) + "°"

    def __add__(self, autre):                 # NOUVEAU
        return Angle(self.v + autre.v)


# 7. Addition (tests du fichier angle.py)

assert (Angle(10) + Angle(20)).mesure_en_degres() == 30
assert (Angle(45) + Angle(45)).mesure_en_degres() == 90
assert (Angle(90) + Angle(90)).mesure_en_degres() == 180

# Dépassement de 180° : le résultat doit être refusé
try:
    Angle(120) + Angle(61)
    assert False, "Une somme supérieure à 180° a été acceptée"
except AssertionError:
    pass

print("Tests de la partie 7 réussis ✔")

try:
    Angle(120) + Angle(61)
    print("120 + 61 : accepté (c'est une erreur !)")
except AssertionError as e:
    print("120 + 61 : refusé ->", e)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- `__add__` est appelée par `a + b` : `self` est l'angle de gauche, `autre` celui de droite.
- On renvoie un **nouvel objet** `Angle` dont la mesure est la somme.
- « la somme doit être comprise entre 0° et 180° » : pas besoin de code en plus. Si la somme dépasse 180, c'est l'`assert` **du constructeur** qui lève l'erreur au moment de créer le nouvel angle. C'est le sens du commentaire du fichier « le résultat doit lui-même être un angle géométrique valide ».

## 9. `__sub__`

#### 📝 ÉNONCÉ

> - `__sub__(autre) -> Angle` — Paramètre : autre, un objet de type Angle. Renvoie un nouvel objet Angle représentant la différence entre les deux angles (précondition : la différence doit être comprise entre 0° et 180°).

#### ✅ RÉSOLUTION

````python
from math import pi


class Angle:

    def __init__(self, v):
        assert 0 <= v <= 180, "La mesure doit être comprise entre 0 et 180 degrés"
        self.v = v

    def mesure_en_degres(self):
        return self.v

    def est_aigu(self):
        return self.v < 90

    def mesure_en_radians(self):
        return self.v * pi / 180

    def mesure_en_degres_minutes_secondes(self):
        total = round(self.v * 3600)          # mesure en secondes d'angle (entier)
        degres = total // 3600
        minutes = (total % 3600) // 60
        secondes = total % 60
        return (degres, minutes, secondes)

    def supplementaire(self):
        return Angle(180 - self.v)

    def __str__(self):
        return str(self.v) + "°"

    def __add__(self, autre):
        return Angle(self.v + autre.v)

    def __sub__(self, autre):                 # NOUVEAU
        return Angle(self.v - autre.v)


# 8. Soustraction (tests du fichier angle.py)

assert (Angle(30) - Angle(20)).mesure_en_degres() == 10
assert (Angle(90) - Angle(30)).mesure_en_degres() == 60
assert (Angle(30) - Angle(30)).mesure_en_degres() == 0
assert (Angle(180) - Angle(90)).mesure_en_degres() == 90

# Résultat négatif : il doit être refusé
try:
    Angle(20) - Angle(30)
    assert False, "Une différence négative a été acceptée"
except AssertionError:
    pass

print("Tests de la partie 8 réussis ✔")

try:
    Angle(20) - Angle(30)
    print("20 - 30 : accepté (c'est une erreur !)")
except AssertionError as e:
    print("20 - 30 : refusé ->", e)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Même idée que `__add__`, appelée par `a - b`.
- Une différence négative (par exemple 20 - 30) est refusée par l'`assert` du constructeur.

**Remarque sur les tests de refus** (valeurs hors de [0 ; 180], somme trop grande, différence négative) :
```python
try:
    Angle(valeur)
    assert False, ...
except AssertionError:
    pass
```
Si la classe accepte la valeur à tort, c'est `assert False` qui lève une `AssertionError`… qui est aussitôt attrapée par le `except`. Ces tests passent donc **même si le constructeur ne vérifie rien**. C'est pour cela que, aux étapes 1, 8 et 9, on a affiché nous-mêmes le message de refus.

**À retenir :**
- `__str__`, `__add__`, `__sub__`... sont des **méthodes spéciales** : on ne les appelle pas directement, Python les appelle pour nous (`print`, `+`, `-`).
- Une méthode qui « renvoie un Angle » doit renvoyer un **objet** `Angle(...)`, pas un nombre.
- Vérifier les préconditions dans le constructeur (avec `assert`) garantit que **tout objet créé est valide**.
