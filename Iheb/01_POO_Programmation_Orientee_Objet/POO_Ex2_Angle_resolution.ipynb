# EXERCICE 2

#### 📝 ÉNONCÉ

> Compléter les méthodes de la classe `Angle` définie dans le fichier `angle.py`.
> - `Angle(v)` (Constructeur) — Paramètre : `v`, nombre représentant la mesure de l'angle en degrés (précondition : 0 ⩽ v ⩽ 180). Crée un objet représentant un angle de mesure v. Déclenche une erreur si v n'est pas entre 0 et 180.
> - `mesure_en_degres() -> float` : renvoie la mesure de l'angle en degrés.
> - `est_aigu() -> bool` : Renvoie True si la mesure de l'angle est strictement inférieure à 90°, False sinon.
> - `mesure_en_radians() -> float` : renvoie la mesure de l'angle en radians.
> - `mesure_en_degres_minutes_secondes() -> tuple` renvoie un triplet (degrés, minutes, secondes) représentant la mesure de l'angle. Exemple : 12,5° correspond à (12,30,0).
> - `supplementaire() -> Angle` : renvoie un nouvel objet Angle représentant le supplémentaire de l'angle (le supplémentaire d'un angle de mesure v a pour mesure 180 − v degrés).
> - `__str__() -> str` : renvoie une chaîne de caractères contenant la mesure de l'angle suivie du symbole °.
> - `__add__(autre) -> Angle` — Paramètre : autre, un objet de type Angle. Renvoie un nouvel objet Angle représentant la somme des deux angles (précondition : la somme doit être comprise entre 0° et 180°).
> - `__sub__(autre) -> Angle` — Paramètre : autre, un objet de type Angle. Renvoie un nouvel objet Angle représentant la différence entre les deux angles (précondition : la différence doit être comprise entre 0° et 180°).

#### ✅ RÉSOLUTION

```python
from math import pi


class Angle:

    def __init__(self, v):
        if not (0 <= v <= 180):
            raise ValueError("La mesure d'un angle doit être comprise entre 0 et 180 degrés")
        self.v = v

    def mesure_en_degres(self):
        return self.v

    def est_aigu(self):
        return self.v < 90

    def mesure_en_radians(self):
        return self.v * pi / 180

    def mesure_en_degres_minutes_secondes(self):
        total = round(self.v * 3600)          # mesure en secondes d'angle (entier)
        degres = total // 3600
        minutes = (total % 3600) // 60
        secondes = total % 60
        return (degres, minutes, secondes)

    def supplementaire(self):
        return Angle(180 - self.v)

    def __str__(self):
        return str(self.v) + "°"

    def __add__(self, autre):
        return Angle(self.v + autre.v)

    def __sub__(self, autre):
        return Angle(self.v - autre.v)
```

#### 💡 EXPLICATION PÉDAGOGIQUE

L'énoncé **est** l'API : pour chaque méthode, il donne son nom, ses paramètres, ce qu'elle renvoie et ses préconditions. Le travail consiste à écrire le code qui respecte ce contrat.

**Méthode par méthode :**
- **`__init__(v)`** : la précondition `0 ⩽ v ⩽ 180` est vérifiée **dès la création**. Si elle est fausse, on **lève une erreur** avec `raise ValueError(...)`. Ainsi, aucun objet `Angle` invalide ne peut exister.
- **`mesure_en_degres()`** : renvoie simplement l'attribut. On passe quand même par une méthode : c'est l'API qui le demande, et l'utilisateur n'a pas besoin de connaître le nom de l'attribut (**encapsulation**).
- **`est_aigu()`** : la comparaison `self.v < 90` est **déjà** un booléen. On le renvoie directement, sans `if ... return True else return False`.
- **`mesure_en_radians()`** : 180° = π rad, donc on multiplie par `pi / 180`.
- **`mesure_en_degres_minutes_secondes()`** : 1° = 60 minutes et 1 minute = 60 secondes. L'astuce est de tout convertir **en secondes entières** (`round(v * 3600)`), puis de faire des divisions entières `//` et des restes `%`, exactement comme pour convertir des secondes en heures/minutes/secondes. On évite ainsi les erreurs d'arrondi des flottants (par exemple 12,7 × 60 = 41,99999...).
- **`supplementaire()`** : renvoie un **nouvel objet** `Angle(180 - v)`, sans modifier l'angle de départ.
- **`__str__()`** : méthode spéciale appelée par `print(a)` et `str(a)`.
- **`__add__(autre)`** et **`__sub__(autre)`** : méthodes spéciales appelées par `a + b` et `a - b`. On crée un nouvel `Angle` : si le résultat sort de [0 ; 180], **c'est le constructeur qui lève l'erreur**, donc la précondition est vérifiée sans code supplémentaire.

*Vérification (non demandée) : on teste chaque méthode de l'API, puis les cas d'erreur.*

```python
a = Angle(12.5)
b = Angle(100)

print(a, b)                                       # __str__
print(a.mesure_en_degres(), a.est_aigu(), b.est_aigu())
print(round(b.mesure_en_radians(), 4))
print(a.mesure_en_degres_minutes_secondes())      # attendu : (12, 30, 0)
print(Angle(45.7625).mesure_en_degres_minutes_secondes())   # (45, 45, 45)
print(b.supplementaire())                         # 80°
print(a + b, b - a)                               # __add__ et __sub__

assert a.mesure_en_degres_minutes_secondes() == (12, 30, 0)
assert b.supplementaire().mesure_en_degres() == 80
assert (a + b).mesure_en_degres() == 112.5
```

```python
# Cas d'erreur : les préconditions sont bien contrôlées
for essai in [lambda: Angle(200), lambda: Angle(-5), lambda: b + b, lambda: a - b]:
    try:
        essai()
    except ValueError as e:
        print("Erreur bien détectée :", e)
```

#### 💡 EXPLICATION PÉDAGOGIQUE

**À retenir :**
- `__str__`, `__add__`, `__sub__`... sont des **méthodes spéciales** : on ne les appelle pas directement, Python les appelle pour nous (`print`, `+`, `-`).
- Une méthode qui « renvoie un Angle » doit renvoyer un **objet** `Angle(...)`, pas un nombre.
- Vérifier les préconditions dans le constructeur garantit que **tout objet créé est valide**.
- Les tests d'erreur utilisent `try / except` et des `lambda` : ce n'est pas à savoir écrire, c'est seulement pour vérifier.
