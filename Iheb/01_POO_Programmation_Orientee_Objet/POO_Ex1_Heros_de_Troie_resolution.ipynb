# Le programme de départ

#### 📝 ÉNONCÉ

> Achille et Hector décident de s'affronter en duel devant les murailles de Troie.
> On dispose du programme suivant :
> ```python
> from random import randint
>
> class Heros :
>     def __init__(self, nom, pv, res, dmax):
>         print("Création du héros " + nom)
>         self.nom = nom
>         self.points_de_vie = pv
>         self.resistance = res
>         self.degats_max = dmax
>     def est_en_vie(self):
>         return self.points_de_vie > 0
>     def attaque(self, autre):
>         if self.est_en_vie():
>             degats = randint(1, self.degats_max)
>             autre.subit_attaque(self, degats)
>     def subit_attaque(self, autre, degats):
>         if degats > self.resistance:
>             self.points_de_vie -= degats
>             print(self.nom + " perd " + str(degats) + " points de vie")
>         else:
>             print(autre.nom + " a raté son attaque contre " + self.nom)
>         if not self.est_en_vie():
>             print(autre.nom + " a tué " + self.nom)
>
> achille = Heros("Achille", 100, 10, 30)
> hector = Heros("Hector", 100, 15, 25)
> paris = Heros("Pâris", 80, 8, 20)
>
> # duel
> while achille.est_en_vie() and hector.est_en_vie():
>     achille.attaque(hector)
>     hector.attaque(achille)
> ```

*Le programme recopié, pour pouvoir l'utiliser dans la suite (le duel est lancé plus bas) :*

````python
from random import randint, random, seed, choice

seed(2)   # rend le hasard reproductible


class Heros:

    def __init__(self, nom, pv, res, dmax):
        print("Création du héros " + nom)
        self.nom = nom
        self.points_de_vie = pv
        self.resistance = res
        self.degats_max = dmax

    def est_en_vie(self):
        return self.points_de_vie > 0

    def attaque(self, autre):
        if self.est_en_vie():
            degats = randint(1, self.degats_max)
            autre.subit_attaque(self, degats)

    def subit_attaque(self, autre, degats):
        if degats > self.resistance:
            self.points_de_vie -= degats
            print(self.nom + " perd " + str(degats) + " points de vie")
        else:
            print(autre.nom + " a raté son attaque contre " + self.nom)

        if not self.est_en_vie():
            print(autre.nom + " a tué " + self.nom)


achille = Heros("Achille", 100, 10, 30)
hector = Heros("Hector", 100, 15, 25)
paris = Heros("Pâris", 80, 8, 20)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Indentation :** toutes les méthodes doivent être indentées **dans** la classe, sinon Python ne les rattache pas à `Heros`.
- **`seed(2)`** (ajouté, non demandé) rend le hasard **reproductible** : on obtient le même combat à chaque exécution, ce qui est pratique pour corriger ensemble.
- **Les imports en plus** (`random`, `seed`, `choice`) servent dans les questions suivantes.

# QUESTION 2 — Les capacités spéciales

#### 📝 ÉNONCÉ

> Les héros de la guerre de Troie peuvent disposer d'une capacité particulière. Par exemple :
> - Achille peut entrer dans une furie guerrière et infliger le maximum de dégâts ;
> - Ulysse peut récupérer des points de vie ;
> - Pâris peut effectuer une attaque précise qui ignore la résistance de son adversaire.
>
> (a) Ajouter à la classe Heros un attribut permettant de mémoriser la capacité spéciale du héros.
> (b) Modifier la création des héros afin de pouvoir préciser leur capacité spéciale. Par exemple : `achille = Heros("Achille", 100, 10, 30, "furie")`
> (c) Écrire une méthode `capacite_speciale` qui permet à un héros d'utiliser sa capacité avec un comportement différent selon la capacité du héros :
> - `"furie"` : le héros inflige le maximum de dégâts à son adversaire ;
> - `"soin"` : le héros récupère 20 points de vie, sans dépasser 100 ;
> - `"precision"` : le héros inflige des dégâts sans tenir compte de la résistance de son adversaire.
>
> (d) Modifier le programme afin qu'à chaque tour, un héros ait une chance sur dix d'utiliser sa capacité spéciale plutôt que d'effectuer une attaque normale.
> (e) On souhaite maintenant ajouter une nouvelle capacité appelée `"bouclier"`. Cette capacité permet au héros de réduire de moitié les dégâts reçus lors de la prochaine attaque. Modifier la méthode `capacite_speciale` pour prendre en compte cette nouvelle capacité.

## 2 (a), (b), (c), (e) — la classe complétée

#### ✅ RÉSOLUTION

On **réécrit la classe complète**. Les ajouts sont signalés par un commentaire `# NOUVEAU` avec le numéro de la question.

````python
class Heros:

    def __init__(self, nom, pv, res, dmax, capacite=None):     # NOUVEAU : capacite
        print("Création du héros " + nom)
        self.nom = nom
        self.points_de_vie = pv
        self.resistance = res
        self.degats_max = dmax
        self.capacite = capacite                              # NOUVEAU (2a)
        self.bouclier_actif = False                           # NOUVEAU (2e)

    def est_en_vie(self):
        return self.points_de_vie > 0

    def attaque(self, autre):
        if self.est_en_vie():
            degats = randint(1, self.degats_max)
            autre.subit_attaque(self, degats)

    def subit_attaque(self, autre, degats, ignore_resistance=False):   # NOUVEAU : ignore_resistance
        if self.bouclier_actif:                               # NOUVEAU (2e)
            degats = degats // 2
            self.bouclier_actif = False                       # le bouclier ne sert qu'une fois
            print(self.nom + " se protège avec son bouclier")
        if ignore_resistance or degats > self.resistance:
            self.points_de_vie -= degats
            print(self.nom + " perd " + str(degats) + " points de vie")
        else:
            print(autre.nom + " a raté son attaque contre " + self.nom)

        if not self.est_en_vie():
            print(autre.nom + " a tué " + self.nom)

    def capacite_speciale(self, autre):                       # NOUVEAU (2c et 2e)
        if not self.est_en_vie():
            return
        print(self.nom + " utilise sa capacité spéciale : " + str(self.capacite))
        if self.capacite == "furie":
            autre.subit_attaque(self, self.degats_max)
        elif self.capacite == "soin":
            self.points_de_vie = min(100, self.points_de_vie + 20)
        elif self.capacite == "precision":
            degats = randint(1, self.degats_max)
            autre.subit_attaque(self, degats, True)
        elif self.capacite == "bouclier":
            self.bouclier_actif = True
        else:                         # pas de capacité : attaque normale
            self.attaque(autre)

    def joue_tour(self, autre):                               # NOUVEAU (2d)
        if self.capacite is not None and random() < 0.1:
            self.capacite_speciale(autre)
        else:
            self.attaque(autre)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

**(a) et (b)** On ajoute un paramètre `capacite` au constructeur et un attribut `self.capacite`. On lui donne une **valeur par défaut** `None` : un héros créé sans capacité (`Heros("Ménélas", 95, 11, 24)`) reste valide, et l'ancien code continue de fonctionner.

**(c)** Le comportement dépend de la valeur de l'attribut `self.capacite` : on fait un `if / elif`.

| Capacité | Effet | Comment on le code |
|:---|:---|:---|
| `"furie"` | dégâts maximum | `autre.subit_attaque(self, self.degats_max)` : on ne tire pas au hasard |
| `"soin"` | +20 PV sans dépasser 100 | `min(100, self.points_de_vie + 20)` |
| `"precision"` | ignore la résistance | on ajoute à `subit_attaque` un paramètre `ignore_resistance` |
| `"bouclier"` (question e) | divise par 2 les dégâts de la **prochaine** attaque | attribut booléen `self.bouclier_actif`, consommé dans `subit_attaque` |

**Pourquoi un paramètre avec valeur par défaut pour la précision ?** `ignore_resistance=False` permet de ne rien changer aux appels existants (`autre.subit_attaque(self, degats)`), tout en autorisant l'appel `autre.subit_attaque(self, degats, True)`.

**(e) Pourquoi un attribut pour le bouclier ?** L'effet ne se produit pas tout de suite, mais **lors de la prochaine attaque reçue**. Il faut donc que l'objet **se souvienne** que son bouclier est levé : c'est exactement le rôle d'un attribut (l'**état** de l'objet). Une variable locale disparaîtrait à la fin de la méthode.

⚠️ **Erreur fréquente (e) :** oublier de remettre `bouclier_actif` à `False` après usage. Le héros serait alors protégé pour toujours.

La méthode `joue_tour` correspond à la question (d), expliquée plus bas.

*Tests de chaque capacité (non demandés, mais bon réflexe) :*

````python
seed(0)
achille = Heros("Achille", 100, 10, 30, "furie")
ulysse = Heros("Ulysse", 90, 12, 25, "soin")
paris = Heros("Pâris", 80, 8, 20, "precision")
ajax = Heros("Ajax", 110, 14, 22, "bouclier")
hector = Heros("Hector", 100, 15, 25)

# furie : Hector perd exactement 30 PV (30 > résistance 15)
achille.capacite_speciale(hector)
assert hector.points_de_vie == 70

# soin : on ne dépasse jamais 100
ulysse.points_de_vie = 90
ulysse.capacite_speciale(hector)
assert ulysse.points_de_vie == 100

# precision : même un petit coup touche Ajax malgré sa résistance de 14
pv_avant = ajax.points_de_vie
paris.capacite_speciale(ajax)
assert ajax.points_de_vie < pv_avant

# bouclier : la prochaine attaque est divisée par 2
ajax.capacite_speciale(hector)
assert ajax.bouclier_actif
achille.capacite_speciale(ajax)          # 30 dégâts -> 15 (> 14 donc ça passe)
assert not ajax.bouclier_actif           # le bouclier a été consommé
print("Tous les tests passent ✔")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

Un bon réflexe : **tester chaque cas séparément** avec `assert` avant de lancer une simulation complète.

## 2 (d) — une chance sur dix

#### ✅ RÉSOLUTION

C'est la méthode `joue_tour` de la classe ci-dessus. Le scénario du duel devient :

````python
seed(5)
achille = Heros("Achille", 100, 10, 30, "furie")
hector = Heros("Hector", 100, 15, 25, "bouclier")

tours = 0
while achille.est_en_vie() and hector.est_en_vie():
    achille.joue_tour(hector)
    hector.joue_tour(achille)
    tours += 1

print("---")
print("Combat terminé en", tours, "tours")
````

# QUESTION 3 — Le cheval de Troie

#### 📝 ÉNONCÉ

> Les Grecs ne parviennent pas à prendre la cité. Ulysse propose alors une ruse : construire un immense cheval de bois et cacher des soldats à l'intérieur.
>
> (a) Écrire une classe `ChevalDeTroie` possédant

> - une méthode permettant d'ajouter un héros dans le cheval.
>

> (c) Écrire une méthode `est_complet` indiquant si le cheval contient déjà le nombre maximal de héros.

## 3 (a) et (c)

#### ✅ RÉSOLUTION

La méthode `est_complet` (c) est écrite tout de suite dans la classe, car `ajoute` s'en sert.

````python
class ChevalDeTroie:

    def __init__(self, capacite_max):
        self.capacite_max = capacite_max
        self.heros_caches = []

    def est_complet(self):
        return len(self.heros_caches) >= self.capacite_max

    def ajoute(self, heros):
        if self.est_complet():
            print("Le cheval est plein : " + heros.nom + " ne peut pas entrer")
        else:
            self.heros_caches.append(heros)
            print(heros.nom + " se cache dans le cheval")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Attributs** : `capacite_max` (un entier) et `heros_caches` (une liste, **vide au départ**).
- **Méthode** `ajoute(heros)` : ajoute un objet `Heros` dans la liste. Elle utilise `est_complet` pour **refuser** un héros quand le cheval est plein.
- **`est_complet`** compare la longueur de la liste à la capacité.
- C'est une **composition** : un objet `ChevalDeTroie` **contient** d'autres objets (`Heros`). La liste contient les objets eux-mêmes, pas seulement leurs noms.

⚠️ **Erreur fréquente :** créer la liste en dehors de `__init__`, au niveau de la classe (`heros_caches = []` juste sous `class`). La liste serait alors **partagée par tous les chevaux**. Il faut écrire `self.heros_caches = []` dans le constructeur.

## 3 (b)

#### ✅ RÉSOLUTION

````python
cheval = ChevalDeTroie(5)

achille = Heros("Achille", 100, 10, 30, "furie")
ulysse = Heros("Ulysse", 90, 12, 25, "soin")
menelas = Heros("Ménélas", 95, 11, 24)
diomede = Heros("Diomède", 95, 12, 26)
ajax = Heros("Ajax", 110, 14, 22, "bouclier")

for h in [achille, ulysse, menelas, diomede, ajax]:
    cheval.ajoute(h)

print("Complet ?", cheval.est_complet())
cheval.ajoute(Heros("Nestor", 70, 9, 18))     # 6e héros : refusé
print([h.nom for h in cheval.heros_caches])
````

#### 💡 EXPLICATION PÉDAGOGIQUE

Le 6e héros (Nestor) est refusé : cela vérifie que `est_complet` et `ajoute` fonctionnent ensemble. La dernière ligne affiche les **noms** (`h.nom`) des objets de la liste.

# QUESTION 4 — La bataille finale

#### 📝 ÉNONCÉ

> La bataille finale oppose maintenant deux armées : l'armée grecque et l'armée troyenne. Une armée est constituée de plusieurs héros. On souhaite créer une classe `Armee`.
>
> (a) Écrire le début de la classe `Armee` permettant de mémoriser le nom de l'armée et la liste de ses héros.
> (b) Écrire une méthode `ajoute` permettant d'ajouter un héros à une armée.
> (c) Écrire une méthode `est_vaincue` qui renvoie `True` si tous les héros de l'armée sont morts et `False` sinon.
> (d) Créer les deux armées suivantes : une armée grecque contenant Achille, Ulysse et plusieurs autres héros ; une armée troyenne contenant Hector, Pâris et plusieurs autres héros.
> (e) Écrire un programme permettant de simuler la bataille entre les deux armées. À chaque tour, un héros vivant de l'armée grecque attaque un héros vivant de l'armée troyenne. Les Troyens ripostent ensuite.
> (f) Modifier le programme pour choisir aléatoirement le héros qui attaque.
> (g) Afficher à la fin de la bataille : le nom de l'armée victorieuse ; le nombre de héros encore en vie ; le nombre de tours qu'a duré la bataille.

## 4 (a), (b), (c)

#### ✅ RÉSOLUTION

````python
class Armee:

    def __init__(self, nom):
        self.nom = nom
        self.heros = []

    def ajoute(self, heros):
        self.heros.append(heros)

    def est_vaincue(self):
        for h in self.heros:
            if h.est_en_vie():
                return False      # contre-exemple trouvé : l'armée n'est pas vaincue
        return True

    def heros_vivants(self):
        return [h for h in self.heros if h.est_en_vie()]
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Attributs** : `nom` et `heros` (liste vide au départ) ; `ajoute(heros)` : un simple `append`.
- **`est_vaincue` : algorithme « pour tout ».** Pour vérifier que *tous* les éléments vérifient une propriété, on parcourt la liste et on renvoie `False` **dès qu'on trouve un contre-exemple** (un héros vivant). Si on arrive au bout de la boucle sans en trouver, on renvoie `True`.
- ⚠️ **Erreur fréquente :** mettre le `return True` **dans** la boucle (dans un `else`) : la fonction s'arrêterait dès le premier héros examiné.
- La méthode `heros_vivants()` n'est **pas demandée** : elle renvoie la liste des héros encore en vie et sert pour la suite. Elle utilise une liste en compréhension, qu'on peut aussi écrire avec une boucle `for` et `append`.

## 4 (d)

#### ✅ RÉSOLUTION

On met la création dans une fonction `cree_armees()` pour pouvoir **recréer des armées neuves** avant chaque bataille (les héros d'une bataille terminée sont morts).

## 4 (e), (f), (g)

#### ✅ RÉSOLUTION

````python
import io
from contextlib import redirect_stdout


def choisit(armee, aleatoire):
    vivants = armee.heros_vivants()
    if aleatoire:
        return choice(vivants)          # (f)
    return vivants[0]                    # (e)


def bataille(armee1, armee2, aleatoire=False):
    tours = 0
    while not armee1.est_vaincue() and not armee2.est_vaincue():
        tours += 1
        choisit(armee1, aleatoire).joue_tour(choisit(armee2, aleatoire))
        if not armee2.est_vaincue():
            choisit(armee2, aleatoire).joue_tour(choisit(armee1, aleatoire))

    # (g) bilan
    if armee1.est_vaincue():
        gagnante = armee2
    else:
        gagnante = armee1
    print("Armée victorieuse :", gagnante.nom)
    print("Héros encore en vie :", len(gagnante.heros_vivants()),
          [h.nom for h in gagnante.heros_vivants()])
    print("Durée de la bataille :", tours, "tours")


for aleatoire in [False, True]:
    seed(7)
    with redirect_stdout(io.StringIO()):          # on masque les créations de héros et les coups
        grecs, troyens = cree_armees()
    print("=== Bataille", "avec choix aléatoire" if aleatoire else "sans hasard (premier héros vivant)", "===")
    with redirect_stdout(io.StringIO()) as detail:
        bataille(grecs, troyens, aleatoire)
    print("\n".join(detail.getvalue().splitlines()[-3:]))     # seulement le bilan
    print()
````

#### 💡 EXPLICATION PÉDAGOGIQUE

**Déroulement d'un tour :**
1. un héros grec vivant attaque un héros troyen vivant ;
2. si les Troyens ne sont pas vaincus, un héros troyen vivant riposte contre un héros grec vivant.

La bataille continue **tant qu'aucune armée n'est vaincue**.
- **(e)** sans hasard : on prend le **premier** héros vivant de chaque liste (`vivants[0]`) ;
- **(f)** avec hasard : `choice(liste)` (module `random`) choisit un élément au hasard ;
- **(g)** à la fin, l'armée victorieuse est celle qui **n'est pas** vaincue ; on compte ses héros vivants et les tours.

On écrit **une seule fonction** avec un paramètre `aleatoire`, pour ne pas dupliquer le code.

⚠️ **Piège :** il faut vérifier que les Troyens ne sont pas vaincus **avant** qu'ils ripostent. Sinon `heros_vivants()` renvoie une liste vide et `vivants[0]` (ou `choice`) provoque une erreur.

**Non demandé :** `redirect_stdout` (hors programme) sert seulement à masquer le détail des coups pour n'afficher que le bilan. Sans lui, le programme affiche chaque attaque.
