## Le programme de départ

On recopie le programme de l'énoncé avec une indentation correcte : **toutes les méthodes doivent être indentées dans la classe**, sinon Python ne les rattache pas à `Heros`.

On ajoute `seed(...)` : le hasard devient reproductible, on obtient donc le même combat à chaque exécution (pratique pour corriger ensemble).

```python
from random import randint, random, seed, choice

seed(2)   # rend le hasard reproductible


class Heros:

    def __init__(self, nom, pv, res, dmax):
        print("Création du héros " + nom)
        self.nom = nom
        self.points_de_vie = pv
        self.resistance = res
        self.degats_max = dmax

    def est_en_vie(self):
        return self.points_de_vie > 0

    def attaque(self, autre):
        if self.est_en_vie():
            degats = randint(1, self.degats_max)
            autre.subit_attaque(self, degats)

    def subit_attaque(self, autre, degats):
        if degats > self.resistance:
            self.points_de_vie -= degats
            print(self.nom + " perd " + str(degats) + " points de vie")
        else:
            print(autre.nom + " a raté son attaque contre " + self.nom)

        if not self.est_en_vie():
            print(autre.nom + " a tué " + self.nom)


achille = Heros("Achille", 100, 10, 30)
hector = Heros("Hector", 100, 15, 25)
paris = Heros("Pâris", 80, 8, 20)
```

## Question 1

### (a) Que représentent les quatre valeurs de `Heros("Achille", 100, 10, 30)` ?

On regarde le constructeur `__init__(self, nom, pv, res, dmax)` : les valeurs sont données **dans l'ordre des paramètres** (`self` est fourni automatiquement par Python).

| Valeur | Paramètre | Attribut créé | Signification |
|:---|:---|:---|:---|
| `"Achille"` | `nom` | `self.nom` | le nom du héros |
| `100` | `pv` | `self.points_de_vie` | ses points de vie de départ |
| `10` | `res` | `self.resistance` | sa résistance : une attaque de dégâts **inférieurs ou égaux** à 10 est sans effet |
| `30` | `dmax` | `self.degats_max` | les dégâts maximum qu'il inflige (une attaque fait entre 1 et 30) |

💡 **À retenir :** pour savoir ce que représentent les arguments d'une instanciation, il faut **toujours lire le constructeur**.

```python
# Vérification : on peut lire les attributs d'un objet avec la syntaxe objet.attribut
print(achille.nom, achille.points_de_vie, achille.resistance, achille.degats_max)
```

### (b) Quelles actions un héros peut-il réaliser ?

Les actions sont les **méthodes** de la classe (hors constructeur) :

- `est_en_vie()` : indiquer s'il est encore vivant (renvoie un booléen) ;
- `attaque(autre)` : **attaquer** un autre héros (seulement s'il est lui-même en vie) ;
- `subit_attaque(autre, degats)` : **subir** une attaque, c'est-à-dire perdre des points de vie si les dégâts dépassent sa résistance.

💡 Les deux méthodes `attaque` et `subit_attaque` montrent comment **deux objets communiquent** : l'attaquant (`self`) appelle une méthode **de l'autre objet** (`autre.subit_attaque(self, degats)`) et se passe lui-même en argument, pour que la victime sache qui l'a frappée.

### (d) Le duel s'arrête quand un des héros a **moins de 20** points de vie

Dans l'énoncé, la boucle tourne **tant que** les deux héros sont en vie. On veut maintenant qu'elle tourne **tant que les deux ont au moins 20 PV**.

Condition pour **continuer** : `achille.points_de_vie >= 20 and hector.points_de_vie >= 20`.

⚠️ **Erreur fréquente :** écrire la condition d'**arrêt** (`< 20`) dans le `while`. Un `while` contient la condition pour **continuer**. Pour passer de l'une à l'autre, on prend la négation (loi de De Morgan) :
`non (A < 20 ou H < 20)` ⟺ `A >= 20 et H >= 20`.

Au passage, on compte les tours, ce qui sert à la question (e).

```python
seed(2)
achille = Heros("Achille", 100, 10, 30)
hector = Heros("Hector", 100, 15, 25)

tours = 0
while achille.points_de_vie >= 20 and hector.points_de_vie >= 20:
    achille.attaque(hector)
    hector.attaque(achille)
    tours = tours + 1
```

### (e) Afficher le héros qui a le plus de points de vie **et** le nombre de tours

> *Énoncé : « À la fin du combat, afficher le nom du héros qui possède encore le plus de points de vie et le nombre de tours nécessaires pour terminer le duel. »*

Il y a **deux informations** à afficher :

1. **Le nom du héros qui a le plus de points de vie** : on compare les attributs `points_de_vie` des deux objets. On garde dans une variable **l'objet** gagnant (pas seulement son nom) : on peut ensuite lire tous ses attributs (`meilleur.nom`, `meilleur.points_de_vie`).
2. **Le nombre de tours** : c'est le compteur `tours`, augmenté de 1 à chaque passage dans la boucle (cellule précédente).

⚠️ **Cas à ne pas oublier :** les deux héros peuvent finir avec **le même nombre** de points de vie. Avec un simple `if / else`, on désignerait Hector à tort : on traite donc l'égalité à part avec un `elif`.

⚠️ **Erreur fréquente :** afficher `meilleur` au lieu de `meilleur.nom`. Python affiche alors quelque chose comme `<__main__.Heros object at 0x7f...>` : c'est l'objet, pas son nom.

```python
if achille.points_de_vie > hector.points_de_vie:
    meilleur = achille
elif hector.points_de_vie > achille.points_de_vie:
    meilleur = hector
else:
    meilleur = None          # égalité

if meilleur is None:
    print("Égalité : les deux héros ont", achille.points_de_vie, "points de vie")
else:
    print(meilleur.nom, "possède encore le plus de points de vie :", meilleur.points_de_vie)
print("Nombre de tours nécessaires pour terminer le duel :", tours)
```

💡 **Remarque :** avec la fonction `max` et le paramètre `key`, on peut faire la même chose en une ligne (hors programme mais utile) :
`max([achille, hector], key=lambda h: h.points_de_vie)`.

## Question 2 : les capacités spéciales

On **réécrit la classe complète** : les ajouts sont signalés par un commentaire `# NOUVEAU`.

### (a) Un attribut pour mémoriser la capacité, (b) le préciser à la création

- On ajoute un paramètre `capacite` au constructeur et un attribut `self.capacite`.
- On lui donne une **valeur par défaut** `None` : un héros créé sans capacité (`Heros("Ménélas", 95, 11, 24)`) reste valide. Ainsi, l'ancien code continue de fonctionner.

### (c) La méthode `capacite_speciale(autre)`

Le comportement dépend de la valeur de l'attribut `self.capacite` : on fait un `if / elif`.

| Capacité | Effet | Comment on le code |
|:---|:---|:---|
| `"furie"` | dégâts maximum | `autre.subit_attaque(self, self.degats_max)` : on ne tire pas au hasard |
| `"soin"` | +20 PV sans dépasser 100 | `min(100, self.points_de_vie + 20)` |
| `"precision"` | ignore la résistance | on ajoute à `subit_attaque` un paramètre `ignore_resistance` |
| `"bouclier"` (question e) | divise par 2 les dégâts de la **prochaine** attaque | attribut booléen `self.bouclier_actif`, consommé dans `subit_attaque` |

💡 **Pourquoi un attribut pour le bouclier ?** L'effet ne se produit pas tout de suite, mais **lors de la prochaine attaque reçue**. Il faut donc que l'objet **se souvienne** que son bouclier est levé : c'est exactement le rôle d'un attribut (l'**état** de l'objet). Une variable locale disparaîtrait à la fin de la méthode.

💡 **Pourquoi un paramètre avec valeur par défaut pour la précision ?** `ignore_resistance=False` permet de ne rien changer aux appels existants (`autre.subit_attaque(self, degats)`), tout en autorisant l'appel `autre.subit_attaque(self, degats, True)`.

```python
class Heros:

    def __init__(self, nom, pv, res, dmax, capacite=None):     # NOUVEAU : capacite
        print("Création du héros " + nom)
        self.nom = nom
        self.points_de_vie = pv
        self.resistance = res
        self.degats_max = dmax
        self.capacite = capacite                              # NOUVEAU (2a)
        self.bouclier_actif = False                           # NOUVEAU (2e)

    def est_en_vie(self):
        return self.points_de_vie > 0

    def attaque(self, autre):
        if self.est_en_vie():
            degats = randint(1, self.degats_max)
            autre.subit_attaque(self, degats)

    def subit_attaque(self, autre, degats, ignore_resistance=False):   # NOUVEAU : ignore_resistance
        if self.bouclier_actif:                               # NOUVEAU (2e)
            degats = degats // 2
            self.bouclier_actif = False                       # le bouclier ne sert qu'une fois
            print(self.nom + " se protège avec son bouclier")
        if ignore_resistance or degats > self.resistance:
            self.points_de_vie -= degats
            print(self.nom + " perd " + str(degats) + " points de vie")
        else:
            print(autre.nom + " a raté son attaque contre " + self.nom)

        if not self.est_en_vie():
            print(autre.nom + " a tué " + self.nom)

    def capacite_speciale(self, autre):                       # NOUVEAU (2c et 2e)
        if not self.est_en_vie():
            return
        print(self.nom + " utilise sa capacité spéciale : " + str(self.capacite))
        if self.capacite == "furie":
            autre.subit_attaque(self, self.degats_max)
        elif self.capacite == "soin":
            self.points_de_vie = min(100, self.points_de_vie + 20)
        elif self.capacite == "precision":
            degats = randint(1, self.degats_max)
            autre.subit_attaque(self, degats, True)
        elif self.capacite == "bouclier":
            self.bouclier_actif = True
        else:                         # pas de capacité : attaque normale
            self.attaque(autre)

    def joue_tour(self, autre):                               # NOUVEAU (2d)
        if self.capacite is not None and random() < 0.1:
            self.capacite_speciale(autre)
        else:
            self.attaque(autre)
```

### Tests de chaque capacité

Un bon réflexe : **tester chaque cas séparément** avant de lancer une simulation complète. On vérifie avec `assert`.

```python
seed(0)
achille = Heros("Achille", 100, 10, 30, "furie")
ulysse = Heros("Ulysse", 90, 12, 25, "soin")
paris = Heros("Pâris", 80, 8, 20, "precision")
ajax = Heros("Ajax", 110, 14, 22, "bouclier")
hector = Heros("Hector", 100, 15, 25)

# furie : Hector perd exactement 30 PV (30 > résistance 15)
achille.capacite_speciale(hector)
assert hector.points_de_vie == 70

# soin : on ne dépasse jamais 100
ulysse.points_de_vie = 90
ulysse.capacite_speciale(hector)
assert ulysse.points_de_vie == 100

# precision : même un petit coup touche Ajax malgré sa résistance de 14
pv_avant = ajax.points_de_vie
paris.capacite_speciale(ajax)
assert ajax.points_de_vie < pv_avant

# bouclier : la prochaine attaque est divisée par 2
ajax.capacite_speciale(hector)
assert ajax.bouclier_actif
achille.capacite_speciale(ajax)          # 30 dégâts -> 15 (> 14 donc ça passe)
assert not ajax.bouclier_actif           # le bouclier a été consommé
print("Tous les tests passent ✔")
```

### (d) Une chance sur dix d'utiliser la capacité spéciale

`random()` renvoie un réel entre 0 et 1 au hasard : `random() < 0.1` est vrai **une fois sur dix** en moyenne. (On aurait aussi pu écrire `randint(1, 10) == 1`.)

C'est la méthode `joue_tour` de la classe ci-dessus. Le scénario du duel devient :

```python
seed(5)
achille = Heros("Achille", 100, 10, 30, "furie")
hector = Heros("Hector", 100, 15, 25, "bouclier")

tours = 0
while achille.est_en_vie() and hector.est_en_vie():
    achille.joue_tour(hector)
    hector.joue_tour(achille)
    tours += 1

print("---")
print("Combat terminé en", tours, "tours")
```

## Question 3 : le cheval de Troie

### (a) La classe `ChevalDeTroie`

- **attributs** : `capacite_max` (un entier) et `heros_caches` (une liste, **vide au départ**) ;
- **méthode** `ajoute(heros)` : ajoute un objet `Heros` dans la liste.

💡 C'est une **composition** : un objet `ChevalDeTroie` **contient** d'autres objets (`Heros`). La liste contient les objets eux-mêmes, pas seulement leurs noms.

⚠️ **Erreur fréquente :** créer la liste en dehors de `__init__`, au niveau de la classe (`heros_caches = []` juste sous `class`). La liste serait alors **partagée par tous les chevaux**. Il faut écrire `self.heros_caches = []` dans le constructeur.

### (c) La méthode `est_complet`

Elle compare la longueur de la liste à la capacité. On l'utilise aussi dans `ajoute` pour **refuser** un héros quand le cheval est plein.

```python
class ChevalDeTroie:

    def __init__(self, capacite_max):
        self.capacite_max = capacite_max
        self.heros_caches = []

    def est_complet(self):
        return len(self.heros_caches) >= self.capacite_max

    def ajoute(self, heros):
        if self.est_complet():
            print("Le cheval est plein : " + heros.nom + " ne peut pas entrer")
        else:
            self.heros_caches.append(heros)
            print(heros.nom + " se cache dans le cheval")
```

### (b) Un cheval de 5 places avec Achille, Ulysse et trois autres héros

```python
cheval = ChevalDeTroie(5)

achille = Heros("Achille", 100, 10, 30, "furie")
ulysse = Heros("Ulysse", 90, 12, 25, "soin")
menelas = Heros("Ménélas", 95, 11, 24)
diomede = Heros("Diomède", 95, 12, 26)
ajax = Heros("Ajax", 110, 14, 22, "bouclier")

for h in [achille, ulysse, menelas, diomede, ajax]:
    cheval.ajoute(h)

print("Complet ?", cheval.est_complet())
cheval.ajoute(Heros("Nestor", 70, 9, 18))     # 6e héros : refusé
print([h.nom for h in cheval.heros_caches])
```

## Question 4 : la bataille finale

### (a), (b), (c) La classe `Armee`

- **attributs** : `nom` et `heros` (liste vide au départ) ;
- `ajoute(heros)` : un simple `append` ;
- `est_vaincue()` : renvoie `True` si **tous** les héros sont morts.

💡 **Algorithme « pour tout » :** pour vérifier que *tous* les éléments vérifient une propriété, on parcourt la liste et on renvoie `False` **dès qu'on trouve un contre-exemple** (un héros vivant). Si on arrive au bout de la boucle sans en trouver, on renvoie `True`.

⚠️ **Erreur fréquente :** mettre le `return True` **dans** la boucle (dans un `else`) : la fonction s'arrêterait dès le premier héros examiné.

On ajoute une méthode utile pour la suite : `heros_vivants()`, qui renvoie la liste des héros encore en vie.

```python
class Armee:

    def __init__(self, nom):
        self.nom = nom
        self.heros = []

    def ajoute(self, heros):
        self.heros.append(heros)

    def est_vaincue(self):
        for h in self.heros:
            if h.est_en_vie():
                return False      # contre-exemple trouvé : l'armée n'est pas vaincue
        return True

    def heros_vivants(self):
        return [h for h in self.heros if h.est_en_vie()]
```

### (d) Les deux armées

```python
def cree_armees():
    grecs = Armee("Grecs")
    for h in [Heros("Achille", 100, 10, 30, "furie"), Heros("Ulysse", 90, 12, 25, "soin"),
              Heros("Ajax", 110, 14, 22, "bouclier"), Heros("Ménélas", 95, 11, 24),
              Heros("Diomède", 95, 12, 26)]:
        grecs.ajoute(h)

    troyens = Armee("Troyens")
    for h in [Heros("Hector", 100, 15, 25, "bouclier"), Heros("Pâris", 80, 8, 20, "precision"),
              Heros("Énée", 100, 13, 24, "soin"), Heros("Sarpédon", 95, 12, 23),
              Heros("Polydamas", 85, 10, 21)]:
        troyens.ajoute(h)
    return grecs, troyens


grecs, troyens = cree_armees()
print([h.nom for h in grecs.heros])
print([h.nom for h in troyens.heros])
```

### (e), (f), (g) Simuler la bataille

**Déroulement d'un tour :**
1. un héros grec vivant attaque un héros troyen vivant ;
2. si les Troyens ne sont pas vaincus, un héros troyen vivant riposte contre un héros grec vivant.

La bataille continue **tant qu'aucune armée n'est vaincue**.

- **(e)** sans hasard : on prend le **premier** héros vivant de chaque liste (`vivants[0]`) ;
- **(f)** avec hasard : `choice(liste)` (module `random`) choisit un élément au hasard ;
- **(g)** à la fin, l'armée victorieuse est celle qui **n'est pas** vaincue ; on compte ses héros vivants et les tours.

On écrit **une seule fonction** avec un paramètre `aleatoire`, pour ne pas dupliquer le code.

⚠️ **Piège :** il faut vérifier que les Troyens ne sont pas vaincus **avant** qu'ils ripostent. Sinon `heros_vivants()` renvoie une liste vide et `vivants[0]` (ou `choice`) provoque une erreur.

Pour garder un affichage lisible, on masque le détail des coups (`redirect_stdout`) et on n'affiche que le bilan.

```python
import io
from contextlib import redirect_stdout


def choisit(armee, aleatoire):
    vivants = armee.heros_vivants()
    if aleatoire:
        return choice(vivants)          # (f)
    return vivants[0]                    # (e)


def bataille(armee1, armee2, aleatoire=False):
    tours = 0
    while not armee1.est_vaincue() and not armee2.est_vaincue():
        tours += 1
        choisit(armee1, aleatoire).joue_tour(choisit(armee2, aleatoire))
        if not armee2.est_vaincue():
            choisit(armee2, aleatoire).joue_tour(choisit(armee1, aleatoire))

    # (g) bilan
    if armee1.est_vaincue():
        gagnante = armee2
    else:
        gagnante = armee1
    print("Armée victorieuse :", gagnante.nom)
    print("Héros encore en vie :", len(gagnante.heros_vivants()),
          [h.nom for h in gagnante.heros_vivants()])
    print("Durée de la bataille :", tours, "tours")


for aleatoire in [False, True]:
    seed(7)
    with redirect_stdout(io.StringIO()):          # on masque les créations de héros et les coups
        grecs, troyens = cree_armees()
    print("=== Bataille", "avec choix aléatoire" if aleatoire else "sans hasard (premier héros vivant)", "===")
    with redirect_stdout(io.StringIO()) as detail:
        bataille(grecs, troyens, aleatoire)
    print("\n".join(detail.getvalue().splitlines()[-3:]))     # seulement le bilan
    print()
```
