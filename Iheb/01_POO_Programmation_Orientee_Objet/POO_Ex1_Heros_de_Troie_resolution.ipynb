# Le programme de départ

#### 📝 ÉNONCÉ

> Achille et Hector décident de s'affronter en duel devant les murailles de Troie.
> On dispose du programme suivant :
> ```python
> from random import randint
>
> class Heros :
>     def __init__(self, nom, pv, res, dmax):
>         print("Création du héros " + nom)
>         self.nom = nom
>         self.points_de_vie = pv
>         self.resistance = res
>         self.degats_max = dmax
>     def est_en_vie(self):
>         return self.points_de_vie > 0
>     def attaque(self, autre):
>         if self.est_en_vie():
>             degats = randint(1, self.degats_max)
>             autre.subit_attaque(self, degats)
>     def subit_attaque(self, autre, degats):
>         if degats > self.resistance:
>             self.points_de_vie -= degats
>             print(self.nom + " perd " + str(degats) + " points de vie")
>         else:
>             print(autre.nom + " a raté son attaque contre " + self.nom)
>         if not self.est_en_vie():
>             print(autre.nom + " a tué " + self.nom)
>
> achille = Heros("Achille", 100, 10, 30)
> hector = Heros("Hector", 100, 15, 25)
> paris = Heros("Pâris", 80, 8, 20)
>
> # duel
> while achille.est_en_vie() and hector.est_en_vie():
>     achille.attaque(hector)
>     hector.attaque(achille)
> ```

*Le programme recopié, pour pouvoir l'utiliser dans la suite (le duel est lancé plus bas) :*

````python
from random import randint, random, seed, choice

seed(2)   # rend le hasard reproductible


class Heros:

    def __init__(self, nom, pv, res, dmax):
        print("Création du héros " + nom)
        self.nom = nom
        self.points_de_vie = pv
        self.resistance = res
        self.degats_max = dmax

    def est_en_vie(self):
        return self.points_de_vie > 0

    def attaque(self, autre):
        if self.est_en_vie():
            degats = randint(1, self.degats_max)
            autre.subit_attaque(self, degats)

    def subit_attaque(self, autre, degats):
        if degats > self.resistance:
            self.points_de_vie -= degats
            print(self.nom + " perd " + str(degats) + " points de vie")
        else:
            print(autre.nom + " a raté son attaque contre " + self.nom)

        if not self.est_en_vie():
            print(autre.nom + " a tué " + self.nom)


achille = Heros("Achille", 100, 10, 30)
hector = Heros("Hector", 100, 15, 25)
paris = Heros("Pâris", 80, 8, 20)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Indentation :** toutes les méthodes doivent être indentées **dans** la classe, sinon Python ne les rattache pas à `Heros`.
- **`seed(2)`** (ajouté, non demandé) rend le hasard **reproductible** : on obtient le même combat à chaque exécution, ce qui est pratique pour corriger ensemble.
- **Les imports en plus** (`random`, `seed`, `choice`) servent dans les questions suivantes.

## 2 (a) — un attribut pour la capacité

## 2 (b) — préciser la capacité à la création

#### 📝 ÉNONCÉ

> (b) Modifier la création des héros afin de pouvoir préciser leur capacité spéciale. Par exemple : `achille = Heros("Achille", 100, 10, 30, "furie")`

#### ✅ RÉSOLUTION

On ajoute le paramètre `capacite` au constructeur, et l'attribut reçoit sa valeur. Puis on précise la capacité à la création de chaque héros.

````python
class Heros:

    def __init__(self, nom, pv, res, dmax, capacite=None):     # NOUVEAU (2b) : paramètre capacite
        print("Création du héros " + nom)
        self.nom = nom
        self.points_de_vie = pv
        self.resistance = res
        self.degats_max = dmax
        self.capacite = capacite                              # MODIFIÉ (2b) : la valeur du paramètre

    def est_en_vie(self):
        return self.points_de_vie > 0

    def attaque(self, autre):
        if self.est_en_vie():
            degats = randint(1, self.degats_max)
            autre.subit_attaque(self, degats)

    def subit_attaque(self, autre, degats):
        if degats > self.resistance:
            self.points_de_vie -= degats
            print(self.nom + " perd " + str(degats) + " points de vie")
        else:
            print(autre.nom + " a raté son attaque contre " + self.nom)

        if not self.est_en_vie():
            print(autre.nom + " a tué " + self.nom)


achille = Heros("Achille", 100, 10, 30, "furie")
ulysse = Heros("Ulysse", 90, 12, 25, "soin")
paris = Heros("Pâris", 80, 8, 20, "precision")
hector = Heros("Hector", 100, 15, 25)            # sans capacité spéciale
print(achille.capacite, ulysse.capacite, paris.capacite, hector.capacite)
````

- On donne au paramètre une **valeur par défaut** `None` (`capacite=None`) : un héros créé sans capacité, comme `Heros("Hector", 100, 15, 25)`, reste valide, et tout l'ancien code continue de fonctionner.

## 2 (c) — la méthode `capacite_speciale`

#### 📝 ÉNONCÉ

> (c) Écrire une méthode `capacite_speciale` qui permet à un héros d'utiliser sa capacité avec un comportement différent selon la capacité du héros :
> - `"furie"` : le héros inflige le maximum de dégâts à son adversaire ;
> - `"soin"` : le héros récupère 20 points de vie, sans dépasser 100 ;
> - `"precision"` : le héros inflige des dégâts sans tenir compte de la résistance de son adversaire.

#### ✅ RÉSOLUTION

On écrit la méthode `capacite_speciale(self, autre)` avec un `if / elif` sur `self.capacite`. Pour la précision, on ajoute à `subit_attaque` un paramètre `ignore_resistance`.

````python
class Heros:

    def __init__(self, nom, pv, res, dmax, capacite=None):
        print("Création du héros " + nom)
        self.nom = nom
        self.points_de_vie = pv
        self.resistance = res
        self.degats_max = dmax
        self.capacite = capacite

    def est_en_vie(self):
        return self.points_de_vie > 0

    def attaque(self, autre):
        if self.est_en_vie():
            degats = randint(1, self.degats_max)
            autre.subit_attaque(self, degats)

    def subit_attaque(self, autre, degats, ignore_resistance=False):   # NOUVEAU (2c) : ignore_resistance
        if ignore_resistance or degats > self.resistance:                 # NOUVEAU (2c)
            self.points_de_vie -= degats
            print(self.nom + " perd " + str(degats) + " points de vie")
        else:
            print(autre.nom + " a raté son attaque contre " + self.nom)

        if not self.est_en_vie():
            print(autre.nom + " a tué " + self.nom)

    def capacite_speciale(self, autre):                       # NOUVEAU (2c)
        if not self.est_en_vie():
            return
        print(self.nom + " utilise sa capacité spéciale : " + str(self.capacite))
        if self.capacite == "furie":
            autre.subit_attaque(self, self.degats_max)
        elif self.capacite == "soin":
            self.points_de_vie = min(100, self.points_de_vie + 20)
        elif self.capacite == "precision":
            degats = randint(1, self.degats_max)
            autre.subit_attaque(self, degats, True)
        else:                         # pas de capacité : attaque normale
            self.attaque(autre)


seed(0)
achille = Heros("Achille", 100, 10, 30, "furie")
ulysse = Heros("Ulysse", 90, 12, 25, "soin")
paris = Heros("Pâris", 80, 8, 20, "precision")
hector = Heros("Hector", 100, 15, 25)

# furie : Hector perd exactement 30 PV (30 > résistance 15)
achille.capacite_speciale(hector)
assert hector.points_de_vie == 70

# soin : Ulysse passe de 90 à 100, pas à 110
ulysse.capacite_speciale(hector)
assert ulysse.points_de_vie == 100

# precision : le coup touche Hector même s'il est plus petit que sa résistance
pv_avant = hector.points_de_vie
paris.capacite_speciale(hector)
assert hector.points_de_vie < pv_avant
print("Tous les tests passent ✔")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

Le comportement dépend de la valeur de l'attribut `self.capacite` : on fait un `if / elif`.

| Capacité | Effet | Comment on le code |
|:---|:---|:---|
| `"furie"` | dégâts maximum | `autre.subit_attaque(self, self.degats_max)` : on ne tire pas au hasard |
| `"soin"` | +20 PV sans dépasser 100 | `min(100, self.points_de_vie + 20)` |
| `"precision"` | ignore la résistance | on ajoute à `subit_attaque` un paramètre `ignore_resistance` |

- **Pourquoi un paramètre avec valeur par défaut pour la précision ?** `ignore_resistance=False` permet de ne rien changer aux appels existants (`autre.subit_attaque(self, degats)` dans `attaque`), tout en autorisant l'appel `autre.subit_attaque(self, degats, True)`.
- Le `else` final traite un héros **sans capacité** : il fait une attaque normale.
- Les `assert` testent **chaque capacité séparément** : c'est un bon réflexe avant de lancer un combat complet.

## 2 (d) — une chance sur dix

#### 📝 ÉNONCÉ

> (d) Modifier le programme afin qu'à chaque tour, un héros ait une chance sur dix d'utiliser sa capacité spéciale plutôt que d'effectuer une attaque normale.

#### ✅ RÉSOLUTION

On ajoute une méthode `joue_tour` qui choisit entre la capacité spéciale (une fois sur dix) et l'attaque normale, puis on l'utilise dans le duel.

````python
class Heros:

    def __init__(self, nom, pv, res, dmax, capacite=None):
        print("Création du héros " + nom)
        self.nom = nom
        self.points_de_vie = pv
        self.resistance = res
        self.degats_max = dmax
        self.capacite = capacite

    def est_en_vie(self):
        return self.points_de_vie > 0

    def attaque(self, autre):
        if self.est_en_vie():
            degats = randint(1, self.degats_max)
            autre.subit_attaque(self, degats)

    def subit_attaque(self, autre, degats, ignore_resistance=False):
        if ignore_resistance or degats > self.resistance:
            self.points_de_vie -= degats
            print(self.nom + " perd " + str(degats) + " points de vie")
        else:
            print(autre.nom + " a raté son attaque contre " + self.nom)

        if not self.est_en_vie():
            print(autre.nom + " a tué " + self.nom)

    def capacite_speciale(self, autre):
        if not self.est_en_vie():
            return
        print(self.nom + " utilise sa capacité spéciale : " + str(self.capacite))
        if self.capacite == "furie":
            autre.subit_attaque(self, self.degats_max)
        elif self.capacite == "soin":
            self.points_de_vie = min(100, self.points_de_vie + 20)
        elif self.capacite == "precision":
            degats = randint(1, self.degats_max)
            autre.subit_attaque(self, degats, True)
        else:                         # pas de capacité : attaque normale
            self.attaque(autre)

    def joue_tour(self, autre):                               # NOUVEAU (2d)
        if self.capacite is not None and random() < 0.1:
            self.capacite_speciale(autre)
        else:
            self.attaque(autre)


seed(5)
achille = Heros("Achille", 100, 10, 30, "furie")
hector = Heros("Hector", 100, 15, 25, "precision")

tours = 0
while achille.est_en_vie() and hector.est_en_vie():
    achille.joue_tour(hector)
    hector.joue_tour(achille)
    tours += 1

print("---")
print("Combat terminé en", tours, "tours")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- `random()` renvoie un réel entre 0 et 1 au hasard : `random() < 0.1` est vrai **une fois sur dix** en moyenne. On aurait aussi pu écrire `randint(1, 10) == 1`.
- `self.capacite is not None` : un héros sans capacité fait toujours une attaque normale.
- Dans la boucle du duel, on appelle `joue_tour` au lieu de `attaque` : c'est `joue_tour` qui décide entre l'attaque normale et la capacité spéciale.

## 2 (e) — la capacité bouclier

#### 📝 ÉNONCÉ

> (e) On souhaite maintenant ajouter une nouvelle capacité appelée `"bouclier"`. Cette capacité permet au héros de réduire de moitié les dégâts reçus lors de la prochaine attaque. Modifier la méthode `capacite_speciale` pour prendre en compte cette nouvelle capacité.

#### ✅ RÉSOLUTION

- Dans `capacite_speciale`, le bouclier **lève** une protection : un nouvel attribut `bouclier_actif` passe à `True`.
- Dans `subit_attaque`, si le bouclier est levé, les dégâts sont divisés par 2, puis le bouclier retombe.

````python
class Heros:

    def __init__(self, nom, pv, res, dmax, capacite=None):
        print("Création du héros " + nom)
        self.nom = nom
        self.points_de_vie = pv
        self.resistance = res
        self.degats_max = dmax
        self.capacite = capacite
        self.bouclier_actif = False                           # NOUVEAU (2e)

    def est_en_vie(self):
        return self.points_de_vie > 0

    def attaque(self, autre):
        if self.est_en_vie():
            degats = randint(1, self.degats_max)
            autre.subit_attaque(self, degats)

    def subit_attaque(self, autre, degats, ignore_resistance=False):
        if self.bouclier_actif:                               # NOUVEAU (2e)
            degats = degats // 2                              # NOUVEAU (2e)
            self.bouclier_actif = False                       # NOUVEAU (2e) : le bouclier ne sert qu'une fois
            print(self.nom + " se protège avec son bouclier")  # NOUVEAU (2e)
        if ignore_resistance or degats > self.resistance:
            self.points_de_vie -= degats
            print(self.nom + " perd " + str(degats) + " points de vie")
        else:
            print(autre.nom + " a raté son attaque contre " + self.nom)

        if not self.est_en_vie():
            print(autre.nom + " a tué " + self.nom)

    def capacite_speciale(self, autre):
        if not self.est_en_vie():
            return
        print(self.nom + " utilise sa capacité spéciale : " + str(self.capacite))
        if self.capacite == "furie":
            autre.subit_attaque(self, self.degats_max)
        elif self.capacite == "soin":
            self.points_de_vie = min(100, self.points_de_vie + 20)
        elif self.capacite == "precision":
            degats = randint(1, self.degats_max)
            autre.subit_attaque(self, degats, True)
        elif self.capacite == "bouclier":                     # NOUVEAU (2e)
            self.bouclier_actif = True                        # NOUVEAU (2e)
        else:                         # pas de capacité : attaque normale
            self.attaque(autre)

    def joue_tour(self, autre):
        if self.capacite is not None and random() < 0.1:
            self.capacite_speciale(autre)
        else:
            self.attaque(autre)


seed(0)
achille = Heros("Achille", 100, 10, 30, "furie")
ajax = Heros("Ajax", 110, 14, 22, "bouclier")

ajax.capacite_speciale(achille)
assert ajax.bouclier_actif
achille.capacite_speciale(ajax)          # 30 dégâts -> 15 (> 14 donc ça passe)
assert ajax.points_de_vie == 110 - 15
assert not ajax.bouclier_actif           # le bouclier a été consommé
print("Tous les tests passent ✔")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Pourquoi un attribut pour le bouclier ?** L'effet ne se produit pas tout de suite, mais **lors de la prochaine attaque reçue**. Il faut donc que l'objet **se souvienne** que son bouclier est levé : c'est exactement le rôle d'un attribut (l'**état** de l'objet). Une variable locale disparaîtrait à la fin de la méthode.
- L'énoncé dit « modifier `capacite_speciale` », mais il faut aussi toucher à `subit_attaque` : c'est là que les dégâts sont reçus, donc là qu'on les divise.
- ⚠️ **Erreur fréquente :** oublier de remettre `bouclier_actif` à `False` après usage. Le héros serait alors protégé pour toujours.

# QUESTION 3 — Le cheval de Troie

#### 📝 ÉNONCÉ (début de la question 3)

> Les Grecs ne parviennent pas à prendre la cité. Ulysse propose alors une ruse : construire un immense cheval de bois et cacher des soldats à l'intérieur.

## 3 (a) — la classe `ChevalDeTroie`

#### 📝 ÉNONCÉ

> (a) Écrire une classe `ChevalDeTroie` possédant
> - une capacité maximale ;
> - une liste de héros cachés à l'intérieur ;
> - une méthode permettant d'ajouter un héros dans le cheval.

#### ✅ RÉSOLUTION

````python
class ChevalDeTroie:

    def __init__(self, capacite_max):
        self.capacite_max = capacite_max
        self.heros_caches = []

    def ajoute(self, heros):
        self.heros_caches.append(heros)
        print(heros.nom + " se cache dans le cheval")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Attributs** : `capacite_max` (un entier, donné à la création) et `heros_caches` (une liste, **vide au départ** : le cheval est construit vide).
- **Méthode** `ajoute(heros)` : ajoute un objet `Heros` à la liste avec `append`.
- C'est une **composition** : un objet `ChevalDeTroie` **contient** d'autres objets (`Heros`). La liste contient les objets eux-mêmes, pas seulement leurs noms.

⚠️ **Erreur fréquente :** créer la liste en dehors de `__init__`, au niveau de la classe (`heros_caches = []` juste sous `class`). La liste serait alors **partagée par tous les chevaux**. Il faut écrire `self.heros_caches = []` dans le constructeur.

## 3 (b) — remplir le cheval

#### 📝 ÉNONCÉ

> (b) Créer un cheval pouvant accueillir 5 héros et y cacher Achille, Ulysse et trois autres héros.

#### ✅ RÉSOLUTION

````python
cheval = ChevalDeTroie(5)

achille = Heros("Achille", 100, 10, 30, "furie")
ulysse = Heros("Ulysse", 90, 12, 25, "soin")
menelas = Heros("Ménélas", 95, 11, 24)
diomede = Heros("Diomède", 95, 12, 26)
ajax = Heros("Ajax", 110, 14, 22, "bouclier")

cheval.ajoute(achille)
cheval.ajoute(ulysse)
cheval.ajoute(menelas)
cheval.ajoute(diomede)
cheval.ajoute(ajax)

print(len(cheval.heros_caches), "héros dans le cheval")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- `ChevalDeTroie(5)` : le 5 arrive dans le paramètre `capacite_max` du constructeur.
- On crée d'abord les héros (des objets `Heros`), puis on les donne un par un à la méthode `ajoute`.

## 3 (c) — la méthode `est_complet`

#### 📝 ÉNONCÉ

> (c) Écrire une méthode `est_complet` indiquant si le cheval contient déjà le nombre maximal de héros.

#### ✅ RÉSOLUTION

On ajoute la méthode `est_complet` à la classe. On en profite pour que `ajoute` refuse un héros quand le cheval est plein (non demandé, mais c'est à ça que sert `est_complet`).

````python
class ChevalDeTroie:

    def __init__(self, capacite_max):
        self.capacite_max = capacite_max
        self.heros_caches = []

    def est_complet(self):                                    # NOUVEAU (3c)
        return len(self.heros_caches) >= self.capacite_max    # NOUVEAU (3c)

    def ajoute(self, heros):
        if self.est_complet():                                # NOUVEAU (3c, non demandé)
            print("Le cheval est plein : " + heros.nom + " ne peut pas entrer")
        else:
            self.heros_caches.append(heros)
            print(heros.nom + " se cache dans le cheval")


cheval = ChevalDeTroie(5)
for h in [achille, ulysse, menelas, diomede, ajax]:
    cheval.ajoute(h)

print("Complet ?", cheval.est_complet())
cheval.ajoute(Heros("Nestor", 70, 9, 18))     # 6e héros : refusé
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **`est_complet`** compare la longueur de la liste à la capacité : elle renvoie directement le booléen `len(...) >= capacite_max`, pas besoin de `if ... return True else return False`.
- Comme on a modifié la classe, on recrée le cheval (l'ancien objet a été fabriqué avec l'ancienne classe) et on y remet les cinq héros de la question (b).
- Le 6e héros (Nestor) est refusé : cela vérifie que `est_complet` et `ajoute` fonctionnent ensemble.

## 4 (c) — la méthode `est_vaincue`

#### 📝 ÉNONCÉ

> (c) Écrire une méthode `est_vaincue` qui renvoie `True` si tous les héros de l'armée sont morts et `False` sinon.

#### ✅ RÉSOLUTION

````python
class Armee:

    def __init__(self, nom):
        self.nom = nom
        self.heros = []

    def ajoute(self, heros):
        self.heros.append(heros)

    def est_vaincue(self):                                    # NOUVEAU (4c)
        for h in self.heros:                                  # NOUVEAU (4c)
            if h.est_en_vie():                                # NOUVEAU (4c)
                return False      # contre-exemple trouvé : l'armée n'est pas vaincue
        return True                                           # NOUVEAU (4c)


grecs = Armee("Grecs")
grecs.ajoute(Heros("Patrocle", 1, 5, 10))
print(grecs.est_vaincue())
grecs.heros[0].points_de_vie = 0
print(grecs.est_vaincue())
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Algorithme « pour tout ».** Pour vérifier que *tous* les éléments vérifient une propriété, on parcourt la liste et on renvoie `False` **dès qu'on trouve un contre-exemple** (un héros vivant). Si on arrive au bout de la boucle sans en trouver, on renvoie `True`.
- On réutilise la méthode `est_en_vie()` de `Heros` : une armée ne regarde pas elle-même les points de vie, elle demande à chaque héros.

⚠️ **Erreur fréquente :** mettre le `return True` **dans** la boucle (dans un `else`) : la fonction s'arrêterait dès le premier héros examiné.

## 4 (d) — créer les deux armées

#### 📝 ÉNONCÉ

> (d) Créer les deux armées suivantes : une armée grecque contenant Achille, Ulysse et plusieurs autres héros ; une armée troyenne contenant Hector, Pâris et plusieurs autres héros.

#### ✅ RÉSOLUTION

````python
def cree_armees():
    grecs = Armee("Grecs")
    grecs.ajoute(Heros("Achille", 100, 10, 30, "furie"))
    grecs.ajoute(Heros("Ulysse", 90, 12, 25, "soin"))
    grecs.ajoute(Heros("Ajax", 110, 14, 22, "bouclier"))
    grecs.ajoute(Heros("Ménélas", 95, 11, 24))
    grecs.ajoute(Heros("Diomède", 95, 12, 26))

    troyens = Armee("Troyens")
    troyens.ajoute(Heros("Hector", 100, 15, 25, "bouclier"))
    troyens.ajoute(Heros("Pâris", 80, 8, 20, "precision"))
    troyens.ajoute(Heros("Énée", 100, 13, 24, "soin"))
    troyens.ajoute(Heros("Sarpédon", 95, 12, 23))
    troyens.ajoute(Heros("Polydamas", 85, 10, 21))
    return grecs, troyens


grecs, troyens = cree_armees()
print(len(grecs.heros), "Grecs et", len(troyens.heros), "Troyens")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

On met la création dans une fonction `cree_armees()` pour pouvoir **recréer des armées neuves** avant chaque bataille (les héros d'une bataille terminée sont morts). Sur la copie, on peut écrire les mêmes lignes sans la fonction.

## 4 (e) — simuler la bataille

#### 📝 ÉNONCÉ

> (e) Écrire un programme permettant de simuler la bataille entre les deux armées. À chaque tour, un héros vivant de l'armée grecque attaque un héros vivant de l'armée troyenne. Les Troyens ripostent ensuite.

#### ✅ RÉSOLUTION

Une fonction `heros_vivants` (non demandée) renvoie la liste des héros encore en vie d'une armée. À chaque tour, on prend le **premier** héros vivant de chaque armée.

````python
def heros_vivants(armee):
    vivants = []
    for h in armee.heros:
        if h.est_en_vie():
            vivants.append(h)
    return vivants


def bataille(grecs, troyens):
    tours = 0
    while not grecs.est_vaincue() and not troyens.est_vaincue():
        tours += 1
        heros_vivants(grecs)[0].joue_tour(heros_vivants(troyens)[0])
        if not troyens.est_vaincue():
            heros_vivants(troyens)[0].joue_tour(heros_vivants(grecs)[0])


seed(7)
grecs, troyens = cree_armees()
bataille(grecs, troyens)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

**Déroulement d'un tour :**
1. un héros grec vivant attaque un héros troyen vivant ;
2. si les Troyens ne sont pas vaincus, un héros troyen vivant riposte contre un héros grec vivant.

La bataille continue **tant qu'aucune armée n'est vaincue** (méthode `est_vaincue` de la question (c)).
- On utilise `joue_tour` (question 2 (d)) : un héros peut donc utiliser sa capacité spéciale pendant la bataille.
- Le compteur `tours` servira à la question (g).

⚠️ **Piège :** il faut vérifier que les Troyens ne sont pas vaincus **avant** qu'ils ripostent. Sinon `heros_vivants(troyens)` est une liste vide et `[0]` provoque une erreur.

## 4 (f) — choisir l'attaquant au hasard

#### 📝 ÉNONCÉ

> (f) Modifier le programme pour choisir aléatoirement le héros qui attaque.

#### ✅ RÉSOLUTION

On remplace « le premier héros vivant » `[0]` par `choice(...)` (module `random`), qui choisit un élément d'une liste au hasard.

````python
def bataille(grecs, troyens):
    tours = 0
    while not grecs.est_vaincue() and not troyens.est_vaincue():
        tours += 1
        choice(heros_vivants(grecs)).joue_tour(choice(heros_vivants(troyens)))       # MODIFIÉ (4f)
        if not troyens.est_vaincue():
            choice(heros_vivants(troyens)).joue_tour(choice(heros_vivants(grecs)))   # MODIFIÉ (4f)


seed(7)
grecs, troyens = cree_armees()
bataille(grecs, troyens)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- `choice(liste)` renvoie un élément de la liste au hasard ; la liste ne doit pas être vide, d'où le même piège qu'en (e).
- On choisit aussi la cible au hasard : sinon tous les Grecs frapperaient toujours le même Troyen.

## 4 (g) — le bilan de la bataille

#### 📝 ÉNONCÉ

> (g) Afficher à la fin de la bataille : le nom de l'armée victorieuse ; le nombre de héros encore en vie ; le nombre de tours qu'a duré la bataille.

#### ✅ RÉSOLUTION

On ajoute le bilan à la fin de la fonction `bataille` de la question (f).

````python
def bataille(grecs, troyens):
    tours = 0
    while not grecs.est_vaincue() and not troyens.est_vaincue():
        tours += 1
        choice(heros_vivants(grecs)).joue_tour(choice(heros_vivants(troyens)))
        if not troyens.est_vaincue():
            choice(heros_vivants(troyens)).joue_tour(choice(heros_vivants(grecs)))

    if grecs.est_vaincue():                                   # NOUVEAU (4g)
        gagnante = troyens                                    # NOUVEAU (4g)
    else:                                                     # NOUVEAU (4g)
        gagnante = grecs                                      # NOUVEAU (4g)
    print("Armée victorieuse :", gagnante.nom)                # NOUVEAU (4g)
    print("Héros encore en vie :", len(heros_vivants(gagnante)))   # NOUVEAU (4g)
    print("Durée de la bataille :", tours, "tours")           # NOUVEAU (4g)


seed(7)
grecs, troyens = cree_armees()
bataille(grecs, troyens)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- L'armée victorieuse est celle qui **n'est pas** vaincue : à la sortie de la boucle, une des deux armées l'est forcément.
- Les héros encore en vie sont ceux de l'armée gagnante (l'autre n'en a plus) : on compte la liste renvoyée par `heros_vivants`.
- Le nombre de tours est le compteur `tours`, augmenté de 1 à chaque passage dans la boucle.
