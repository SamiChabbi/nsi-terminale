# 1. Le principe : une ligne = une photo des variables

Un programme manipule des **variables** : des noms (`x`, `p`, `gauche`…) qui contiennent des valeurs. Ces valeurs changent pendant l'exécution.

Un tableau de trace, c'est :
- **une colonne par variable** (et parfois une colonne en plus pour une condition, ou pour compter quelque chose) ;
- **une ligne par étape** : chaque ligne est une **photo** de toutes les variables à un instant précis.

On avance **ligne par ligne**, jamais colonne par colonne. Sur une ligne, on écrit la valeur de **toutes** les variables : si une variable n'a pas changé, on **recopie** sa valeur de la ligne du dessus.

---
# 2. Des affectations, ligne par ligne

En Python, `x = 2` est une **affectation** : on range la valeur 2 dans la variable `x`.

⚠️ Dans `x = x * y`, Python calcule **d'abord** la partie de droite, avec les valeurs **actuelles** de `x` et `y`, puis range le résultat dans `x`. Le signe `=` ne veut pas dire « est égal », il veut dire « reçoit ».

**Exemple :**

````python
x = 2
y = x + 1
x = x * y
y = y - x
````

Ici, il n'y a pas de boucle : on fait **une ligne par instruction**.

| Instruction | `x` | `y` |
|:---|:---:|:---:|
| `x = 2` | 2 | — |
| `y = x + 1` | 2 | 3 |
| `x = x * y` | 6 | 3 |
| `y = y - x` | 6 | −3 |

- Ligne 2 : `y` reçoit 2 + 1 = 3 ; `x` ne change pas, on recopie 2.
- Ligne 3 : `x` reçoit 2 × 3 = 6 (on utilise l'**ancienne** valeur de `x`, 2).
- Ligne 4 : `y` reçoit 3 − 6 = −3.

Le tiret « — » veut dire que la variable n'existe pas encore.

````python
# Vérification
print("x =", x, "et y =", y)
````

# 3. Une boucle `for`

Une boucle `for` répète des instructions plusieurs fois.

`range(n)` donne les entiers de `0` à `n - 1` : `range(4)` donne `0, 1, 2, 3`. La boucle `for i in range(4):` fait donc **4 tours**, et la variable `i` vaut 0, puis 1, puis 2, puis 3. Les lignes décalées (indentées) sous le `for` sont exécutées à chaque tour.

**Les règles du tableau pour une boucle :**
1. une **première ligne** avec les valeurs **avant** la boucle ;
2. puis **une ligne par tour**, avec les valeurs **à la fin** du tour.

**Exemple :** la fonction `power_bf` calcule $x^n$ en multipliant `n` fois par `x` (c'est la « force brute » de l'exercice 3 du polycopié). On suit `power_bf(2, 4)`, et on ajoute une colonne pour **compter les multiplications**.

````python
def power_bf(x, n):
    p = 1
    for i in range(n):
        p = p * x
    return p
````

Avec `x = 2` et `n = 4` :

| Étape | `i` | `p` | multiplications faites |
|:---|:---:|:---:|:---:|
| avant la boucle | — | 1 | 0 |
| fin du tour 1 | 0 | 2 | 1 |
| fin du tour 2 | 1 | 4 | 2 |
| fin du tour 3 | 2 | 8 | 3 |
| fin du tour 4 | 3 | 16 | 4 |

La boucle est finie après 4 tours : la fonction renvoie `p`, c'est-à-dire **16**, et elle a fait **4 multiplications**.

**Généraliser :** le tableau montre **une multiplication par tour**. Pour `n` quelconque, il y a `n` tours, donc **`n` multiplications**. C'est comme ça qu'on répond à « combien de multiplications ? » : on fait le tableau sur un **petit** exemple, on repère ce qui se répète, puis on généralise.

````python
# Vérification : on affiche i et p à la fin de chaque tour
def power_bf_trace(x, n):
    p = 1
    nb = 0
    print("avant la boucle : p =", p)
    for i in range(n):
        p = p * x
        nb = nb + 1
        print("fin du tour : i =", i, " p =", p, " multiplications =", nb)
    return p


print("renvoie", power_bf_trace(2, 4))
````

# 4. Une boucle `while`

Une boucle `while condition:` répète ses instructions **tant que** la condition est vraie. On ne sait pas à l'avance combien de tours elle fera.

`n // 2` est la **division entière** : on divise et on garde la partie entière (`20 // 2` vaut 10, `5 // 2` vaut 2).

**Les règles :** les mêmes que pour `for`, avec une colonne de plus pour la **condition**, qu'on teste au début de chaque tour. On s'arrête dès qu'elle est **fausse**.

**Exemple :** combien de fois peut-on diviser 20 par 2 avant d'arriver à 1 ?

````python
n = 20
etapes = 0
while n > 1:
    n = n // 2
    etapes = etapes + 1
````

| Étape | `n` | `etapes` | condition `n > 1` (testée ensuite) |
|:---|:---:|:---:|:---:|
| avant la boucle | 20 | 0 | `True` : on entre |
| fin du tour 1 | 10 | 1 | `True` : on continue |
| fin du tour 2 | 5 | 2 | `True` : on continue |
| fin du tour 3 | 2 | 3 | `True` : on continue |
| fin du tour 4 | 1 | 4 | `False` : **on sort** |

À la fin, `etapes` vaut **4**. (Ce nombre de divisions par 2, c'est le **logarithme binaire** du cours sur « diviser pour régner ».)

````python
# Vérification
print("n =", n, "et etapes =", etapes)
````

# 5. Une boucle `while` avec des `if` : la recherche dichotomique

**Ce qu'il faut savoir avant :**
- un tableau (une liste) `t = [2, 5, 7, 9, 12, 15, 20]` contient des valeurs rangées dans des cases numérotées **à partir de 0** : `t[0]` vaut 2, `t[3]` vaut 9, `t[6]` vaut 20 ;
- `len(t)` est le nombre de cases (ici 7) ;
- `if … elif … else` choisit **un seul** des cas : le premier dont la condition est vraie ;
- `return` arrête la fonction **immédiatement** (même au milieu d'une boucle) et renvoie la valeur.

La recherche dichotomique cherche une valeur `x` dans un tableau **trié** : on regarde la case du milieu de la zone de recherche, et on garde seulement la moitié où `x` peut se trouver. La zone de recherche va de l'indice `gauche` à l'indice `droite`.

````python
def recherche(t, x):
    gauche = 0
    droite = len(t) - 1
    while gauche <= droite:
        milieu = (gauche + droite) // 2
        if t[milieu] == x:
            return True
        elif t[milieu] < x:
            gauche = milieu + 1
        else:
            droite = milieu - 1
    return False
````

**Exemple 1 :** `recherche([2, 5, 7, 9, 12, 15, 20], 15)`. On ajoute deux colonnes : la valeur `t[milieu]` regardée, et la **décision** prise par le `if`.

| Étape | `gauche` | `droite` | `milieu` | `t[milieu]` | décision |
|:---|:---:|:---:|:---:|:---:|:---|
| avant la boucle | 0 | 6 | — | — | `0 <= 6` : on entre |
| tour 1 | 4 | 6 | 3 | 9 | 9 < 15 : on va à droite, `gauche = 4` |
| tour 2 | 4 | 6 | 5 | 15 | 15 == 15 : **renvoie `True`** |

**Exemple 2 :** `recherche([2, 5, 7, 9, 12, 15, 20], 8)` (8 n'est pas dans le tableau).

| Étape | `gauche` | `droite` | `milieu` | `t[milieu]` | décision |
|:---|:---:|:---:|:---:|:---:|:---|
| avant la boucle | 0 | 6 | — | — | `0 <= 6` : on entre |
| tour 1 | 0 | 2 | 3 | 9 | 9 > 8 : on va à gauche, `droite = 2` |
| tour 2 | 2 | 2 | 1 | 5 | 5 < 8 : on va à droite, `gauche = 2` |
| tour 3 | 3 | 2 | 2 | 7 | 7 < 8 : on va à droite, `gauche = 3` |
| fin | 3 | 2 | | | `3 <= 2` est faux : on sort, **renvoie `False`** |

Dans l'exemple 2, la boucle s'arrête parce que la **zone de recherche est vide** (`gauche` est passé à droite de `droite`).

````python
# Vérification : on affiche les variables à chaque tour
def recherche_trace(t, x):
    gauche = 0
    droite = len(t) - 1
    print("avant la boucle : gauche =", gauche, " droite =", droite)
    while gauche <= droite:
        milieu = (gauche + droite) // 2
        if t[milieu] == x:
            print("milieu =", milieu, " t[milieu] =", t[milieu], " -> trouvé")
            return True
        elif t[milieu] < x:
            gauche = milieu + 1
        else:
            droite = milieu - 1
        print("fin du tour : gauche =", gauche, " droite =", droite, " milieu =", milieu, " t[milieu] =", t[milieu])
    print("gauche > droite : zone vide")
    return False


print(recherche_trace([2, 5, 7, 9, 12, 15, 20], 15))
print()
print(recherche_trace([2, 5, 7, 9, 12, 15, 20], 8))
````

# 6. Une fonction

Une **fonction** est un morceau de programme auquel on donne un nom. On lui donne des valeurs (les **paramètres**), elle fait son calcul et **renvoie** un résultat avec `return`.

Quand Python rencontre un **appel** de fonction, il met de côté ce qu'il était en train de faire, exécute la fonction avec les valeurs données, puis **revient** là où il en était avec le résultat.

````python
def carre(a):
    return a * a


b = carre(3) + 1
````

| Étape | `a` | `b` | ce qui se passe |
|:---|:---:|:---:|:---|
| appel `carre(3)` | 3 | — | on entre dans la fonction avec `a = 3` |
| `return a * a` | 3 | — | la fonction renvoie 9 |
| retour | | 10 | on revient : `b` reçoit 9 + 1 = 10 |

Dans un tableau de trace, un appel de fonction se note donc en écrivant **la valeur des paramètres**, puis **la valeur renvoyée**.

# 7. Une fonction récursive (un seul appel)

Une fonction **récursive** est une fonction qui **s'appelle elle-même**. Chaque appel attend la réponse de l'appel qu'il a lancé avant de pouvoir finir son propre calcul.

**Exemple :** la fonction `power_dac` calcule $x^n$ beaucoup plus vite que `power_bf` (c'est l'exponentiation rapide, version corrigée, de l'exercice 3 du polycopié). Elle utilise :
- `n % 2` : le **reste** de la division de `n` par 2. Si `n % 2 == 0`, `n` est **pair** ;
- l'idée : si `n` est pair, $x^n = x^{n/2} \times x^{n/2}$ ; si `n` est impair, $x^n = x \times x^{n-1}$ ; et $x^1 = x$.

````python
def power_dac(x, n):
    if n == 1:
        return x
    if n % 2 == 0:
        y = power_dac(x, n // 2)
        return y * y
    else:
        return x * power_dac(x, n - 1)
````

**Les règles du tableau pour une fonction récursive :**
1. **une ligne par appel**, en commençant par l'appel de départ ;
2. pour chaque appel, on écrit les **paramètres**, et quel cas du `if` s'applique ;
3. tant que ce n'est pas le cas de base (`n == 1`), on écrit **ce qu'il faut calculer**, avec l'appel suivant **en attente**, et on passe à la ligne suivante : c'est la **descente** ⬇️ ;
4. au cas de base, on connaît enfin une valeur ;
5. on remplit alors la colonne **« renvoie » de bas en haut** : chaque ligne utilise la valeur de la ligne du dessous. C'est la **remontée** ⬆️.

**Exemple :** `power_dac(2, 10)`, en comptant les multiplications.

| Appel | `n` | cas | ce qu'il faut calculer | multiplications | renvoie (on remplit en remontant ⬆️) |
|:---|:---:|:---|:---|:---:|:---:|
| `power_dac(2, 10)` ⬇️ | 10 | pair | `y = power_dac(2, 5)`, puis `y * y` | 1 | 32 × 32 = **1024** |
| `power_dac(2, 5)` ⬇️ | 5 | impair | `2 * power_dac(2, 4)` | 1 | 2 × 16 = **32** ⬆️ |
| `power_dac(2, 4)` ⬇️ | 4 | pair | `y = power_dac(2, 2)`, puis `y * y` | 1 | 4 × 4 = **16** ⬆️ |
| `power_dac(2, 2)` ⬇️ | 2 | pair | `y = power_dac(2, 1)`, puis `y * y` | 1 | 2 × 2 = **4** ⬆️ |
| `power_dac(2, 1)` | 1 | cas de base | — | 0 | **2** ⬆️ |

- **Valeur renvoyée :** la case en haut à droite, $2^{10} = 1024$.
- **Nombre de multiplications :** la somme de la colonne, **4** (contre 10 pour `power_bf`).
- **Nombre d'appels :** le nombre de lignes, 5.

> ⚠️ On ne peut **pas** calculer la valeur renvoyée pendant la descente : au moment de l'appel `power_dac(2, 10)`, on ne connaît pas encore `power_dac(2, 5)`. L'appel reste **en attente**, et on finit le calcul en **remontant**.

**Pour compter les multiplications avec n = 100**, seules les colonnes `n` et « multiplications » sont utiles :

| `n` | 100 | 50 | 25 | 24 | 12 | 6 | 3 | 2 | 1 |
|:---|:---:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|
| cas | pair | pair | impair | pair | pair | pair | impair | pair | base |
| multiplications | 1 | 1 | 1 | 1 | 1 | 1 | 1 | 1 | 0 |

Total : **8 multiplications** pour calculer $x^{100}$, au lieu de 100.

````python
# Vérification : on affiche chaque appel (décalé selon sa profondeur) et ce qu'il renvoie
def power_dac_trace(x, n, decalage):
    print(decalage + "appel power_dac(" + str(x) + ", " + str(n) + ")")
    if n == 1:
        print(decalage + "cas de base : renvoie " + str(x))
        return x
    if n % 2 == 0:
        y = power_dac_trace(x, n // 2, decalage + "    ")
        resultat = y * y
    else:
        resultat = x * power_dac_trace(x, n - 1, decalage + "    ")
    print(decalage + "power_dac(" + str(x) + ", " + str(n) + ") renvoie " + str(resultat))
    return resultat


power_dac_trace(2, 10, "")
````

# 8. Une fonction récursive avec deux appels : l'arbre des appels

Voici une version **maladroite** de la même fonction (celle de l'exercice 3 du polycopié). Le résultat est juste, mais quand `n` est pair, elle fait **deux appels** :

````python
def power_dac_maladroite(x, n):
    if n == 1:
        return x
    if n % 2 == 0:
        return power_dac_maladroite(x, n // 2) * power_dac_maladroite(x, n // 2)
    else:
        return x * power_dac_maladroite(x, n - 1)
````

Quand un appel en lance **deux**, un tableau ne suffit plus : on dessine un **arbre des appels**.

**Les règles :**
1. on écrit l'appel de départ en haut ;
2. sous chaque appel, on écrit les appels qu'il fait, de gauche à droite ;
3. on s'arrête aux cas de base (ils ne font aucun appel) ;
4. on compte ce qu'on cherche **case par case** (ici : une multiplication par case qui n'est pas un cas de base).

**Exemple :** `power_dac_maladroite(x, 8)`

```
                       power_dac_maladroite(x, 8)                          1 multiplication
                     /                            \
       pdm(x, 4)                                    pdm(x, 4)              1 + 1
       /        \                                   /        \
  pdm(x, 2)    pdm(x, 2)                       pdm(x, 2)    pdm(x, 2)      1 + 1 + 1 + 1
   /    \       /    \                          /    \       /    \
(x,1) (x,1)  (x,1) (x,1)                     (x,1) (x,1)  (x,1) (x,1)      cas de base : 0
```
(`pdm` est une abréviation de `power_dac_maladroite`.)

Total : 1 + 2 + 4 = **7 multiplications**, presque autant que `power_bf` (8). Couper en deux n'a **rien** rapporté !

**Ce que montre l'arbre :** les deux moitiés sont **identiques**. On calcule `pdm(x, 4)` deux fois, `pdm(x, 2)` quatre fois… Dès qu'on voit dans un arbre **deux fois le même appel avec les mêmes arguments**, c'est qu'on refait le même calcul : il suffit de le faire **une seule fois** et de **ranger le résultat dans une variable**. C'est exactement ce que fait `power_dac` (partie 7), dont l'arbre est une simple ligne :

```
power_dac(x, 8) → power_dac(x, 4) → power_dac(x, 2) → power_dac(x, 1)
     1                 1                 1                 0         = 3 multiplications
```

````python
# Vérification : on compte les multiplications des trois versions pour n = 8
compteur = [0]


def multiplier(a, b):
    compteur[0] = compteur[0] + 1
    return a * b


def bf(x, n):
    p = 1
    for i in range(n):
        p = multiplier(p, x)
    return p


def maladroite(x, n):
    if n == 1:
        return x
    if n % 2 == 0:
        return multiplier(maladroite(x, n // 2), maladroite(x, n // 2))
    else:
        return multiplier(x, maladroite(x, n - 1))


def rapide(x, n):
    if n == 1:
        return x
    if n % 2 == 0:
        y = rapide(x, n // 2)
        return multiplier(y, y)
    else:
        return multiplier(x, rapide(x, n - 1))


for nom, fonction in [("force brute", bf), ("maladroite", maladroite), ("rapide", rapide)]:
    compteur[0] = 0
    fonction(2, 8)
    print(nom, ":", compteur[0], "multiplications")
````

# 9. La méthode en résumé

| Ce qu'on suit | Le tableau |
|:---|:---|
| des affectations | une ligne par instruction ; une colonne par variable |
| une boucle `for` | une ligne **avant** la boucle, puis une ligne par tour (valeurs **à la fin** du tour) |
| une boucle `while` | pareil, avec une colonne pour la **condition** ; on s'arrête quand elle est fausse |
| des `if` dans la boucle | une colonne « décision » : quel cas a été choisi |
| une fonction récursive à un appel | une ligne par **appel** ; on descend, puis on remplit « renvoie » en **remontant** |
| une fonction récursive à plusieurs appels | un **arbre des appels** ; on compte case par case |

**Les réflexes :**
- avancer **ligne par ligne**, et recopier les valeurs qui ne changent pas ;
- pour une question « combien d'opérations ? », ajouter une **colonne qui compte**, faire un **petit** exemple, puis **généraliser** ;
- au contrôle, **recopier le tableau** sur la copie même si on ne le finit pas : un tableau commencé juste rapporte déjà des points.
