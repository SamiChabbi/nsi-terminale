# QUESTION 1

#### 📝 ÉNONCÉ

> 1. Écrire une fonction `permute_prochains` qui permute les deux prochains éléments d'une file.

#### ✅ RÉSOLUTION

```python
def permute_prochains(f):
    if f.taille() < 2:
        return
    aux = File()
    x = f.defile()
    y = f.defile()
    aux.enfile(y)
    aux.enfile(x)
    while not f.est_vide():
        aux.enfile(f.defile())
    while not aux.est_vide():
        f.enfile(aux.defile())
```

*Vérification (non demandée) :*

```python
f = File()
f.enfile("a")
f.enfile("b")
f.enfile("c")
f.enfile("d")
permute_prochains(f)
print(f)
assert str(f) == "File(['b', 'a', 'c', 'd'])"

g = File()
g.enfile(1)
permute_prochains(g)             # un seul élément : rien à faire
assert str(g) == "File([1])"
g.enfile(2)
permute_prochains(g)             # deux éléments
assert str(g) == "File([2, 1])"
```

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Les deux prochains éléments** sont les deux premiers (ceux qui vont sortir) : `a b c d` → `b a c d`.
- **Étapes :** on défile les deux premiers et on les range dans `aux` **dans l'ordre inverse** ; on vide ensuite le reste de `f` derrière eux ; enfin on reverse `aux` dans `f`.

| Étape | `f` | `aux` |
|:---|:---|:---|
| départ | `a b c d` | |
| `x = defile()`, `y = defile()` | `c d` | |
| `aux.enfile(y)`, `aux.enfile(x)` | `c d` | `b a` |
| on vide `f` dans `aux` | | `b a c d` |
| on reverse `aux` dans `f` | `b a c d` ✔ | |

- **Cas particulier :** avec moins de 2 éléments, il n'y a rien à permuter (et le deuxième `defile` déclencherait une erreur) : on sort avec `return`. C'est le seul endroit où l'on a besoin de `taille()`.
- **Méthode générale du TD :** on vide `f` dans une file auxiliaire en faisant le travail au passage, puis on reverse (voir l'exercice 2). Les deux boucles s'arrêtent, car chacune enfile dans **l'autre** file.

# QUESTION 2

#### 📝 ÉNONCÉ

> 2. Écrire une fonction `permute_derniers` qui permute les deux derniers éléments d'une file.

#### ✅ RÉSOLUTION

```python
def permute_derniers(f):
    if f.taille() < 2:
        return
    aux = File()
    while f.taille() > 2:
        aux.enfile(f.defile())
    x = f.defile()
    y = f.defile()
    aux.enfile(y)
    aux.enfile(x)
    while not aux.est_vide():
        f.enfile(aux.defile())
```

*Vérification (non demandée) :*

```python
f = File()
f.enfile("a")
f.enfile("b")
f.enfile("c")
f.enfile("d")
permute_derniers(f)
print(f)
assert str(f) == "File(['a', 'b', 'd', 'c'])"

g = File()
permute_derniers(g)              # file vide : rien à faire
assert g.est_vide()
g.enfile(1)
g.enfile(2)
permute_derniers(g)              # deux éléments
assert str(g) == "File([2, 1])"
```

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Idée :** on fait passer dans `aux` tous les éléments **sauf les deux derniers** ; il reste alors exactement les deux derniers dans `f`. On les défile et on les range dans `aux` **dans l'ordre inverse**, puis on reverse `aux` dans `f`.

| Étape | `f` | `aux` |
|:---|:---|:---|
| départ | `a b c d` | |
| tant que `f` a plus de 2 éléments | `c d` | `a b` |
| `x = defile()`, `y = defile()` | | `a b` |
| `aux.enfile(y)`, `aux.enfile(x)` | | `a b d c` |
| on reverse `aux` dans `f` | `a b d c` ✔ | |

- **Pourquoi `while f.taille() > 2` ?** Pour s'arrêter quand il ne reste que les deux derniers, il faut savoir **combien** il en reste : c'est ici que `taille()` (qui fait partie de l'API de l'exercice 1) est utile. Ce n'est pas un compteur : la condition porte sur la file elle-même, qui rétrécit à chaque tour.
- **Vérifier un cas limite :** avec 2 éléments, la première boucle ne fait rien, et on retrouve `permute_prochains` : logique, les deux prochains sont aussi les deux derniers.
- ⚠️ **Erreur fréquente :** **renvoyer** `aux` au lieu de le reverser dans `f`. L'appel `permute_derniers(f)` laisserait alors `f` vide.
