# MISE EN PLACE

> Dans la suite, on utilisera exclusivement uniquement des objets de la classe File définie dans l'exercice 1.
> (Interdit d'utiliser le type list de python.)

On importe la classe `File` de l'exercice 1 : le fichier `file.py` doit être **dans le même dossier** que ce notebook.

La fonction `file_de` (non demandée) sert seulement à fabriquer vite des files pour vérifier les réponses. Une file s'affiche sous la forme `tête < ... < queue` : on **défile à gauche** et on **enfile à droite**.

```python
from file import File


def file_de(*elements):
    f = File()
    for el in elements:
        f.enfile(el)
    return f


print(file_de("a", "b", "c"))
```

# QUESTION 1

#### 📝 ÉNONCÉ

> 1. Écrire une fonction `permute_prochains` qui permute les deux prochains éléments d'une file.

#### ✅ RÉSOLUTION

```python
def permute_prochains(f):
    n = f.taille()
    if n < 2:
        return
    x = f.defile()
    y = f.defile()
    f.enfile(y)
    f.enfile(x)
    for _ in range(n - 2):
        f.enfile(f.defile())
```

*Vérification (non demandée) :*

```python
f = file_de("a", "b", "c", "d")
permute_prochains(f)
print(f)
assert repr(f) == repr(file_de("b", "a", "c", "d"))

g = file_de(1, 2)
permute_prochains(g)
assert repr(g) == repr(file_de(2, 1))
h = file_de(1)
permute_prochains(h)
assert repr(h) == repr(file_de(1))
```

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Les deux prochains éléments** sont les deux premiers (ceux qui vont sortir) : `a b c d` → `b a c d`.
- **Étapes :** on défile les deux premiers, on les ré-enfile **dans l'ordre inverse**, puis on fait passer leur tour aux `n - 2` autres pour ramener la paire devant.

| Étape | File |
|:---|:---|
| départ | `a b c d` |
| `x = defile()`, `y = defile()` | `c d` |
| `enfile(y)`, `enfile(x)` | `c d b a` |
| 2 rotations | `b a c d` ✔ |

- **Cas particulier :** avec moins de 2 éléments, il n'y a rien à permuter (et le deuxième `defile` déclencherait une erreur) : on sort avec `return`.

# QUESTION 2

#### 📝 ÉNONCÉ

> 2. Écrire une fonction `permute_derniers` qui permute les deux derniers éléments d'une file.

#### ✅ RÉSOLUTION

```python
def permute_derniers(f):
    n = f.taille()
    if n < 2:
        return
    for _ in range(n - 2):
        f.enfile(f.defile())
    x = f.defile()
    y = f.defile()
    f.enfile(y)
    f.enfile(x)
```

*Vérification (non demandée) :*

```python
f = file_de("a", "b", "c", "d")
permute_derniers(f)
print(f)
assert repr(f) == repr(file_de("a", "b", "d", "c"))

g = file_de(1, 2)
permute_derniers(g)
assert repr(g) == repr(file_de(2, 1))
h = File()
permute_derniers(h)
assert h.est_vide()
```

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Idée :** on amène d'abord les deux derniers **en tête** (`n - 2` rotations), puis on les défile et on les ré-enfile dans l'ordre inverse. Pas besoin de rotation finale : les autres sont déjà revenus devant.

| Étape | File |
|:---|:---|
| départ | `a b c d` |
| 2 rotations | `c d a b` |
| `x = defile()`, `y = defile()` | `a b` |
| `enfile(y)`, `enfile(x)` | `a b d c` ✔ |

- **Vérifier un cas limite :** avec 2 éléments, 0 rotation, et on retrouve `permute_prochains` : logique, les deux prochains sont aussi les deux derniers.
- ⚠️ **Erreur fréquente :** **renvoyer** une nouvelle file au lieu de modifier celle reçue. L'appel `permute_derniers(f)` ne changerait alors rien à `f`.
