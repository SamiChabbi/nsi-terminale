# Les files — Exercice 3 : permuter des éléments

**Notions travaillées :** modifier une file **en place**, combiner défilements et rotations.

(L'énoncé écrit « que les modifications de la file soi**t** effectuées » : il faut lire « soi**ent** ».)

> **En place** veut dire : la fonction **modifie la file reçue** en argument et ne renvoie rien. On ne fabrique pas une nouvelle file.

## Mise en place

On importe la classe `File` écrite à l'exercice 1 (le fichier `file.py` doit être **dans le même dossier** que ce notebook).

L'énoncé interdit d'utiliser le type `list` : on ne manipule les files **qu'avec leurs méthodes** `enfile`, `defile`, `taille`, `est_vide`. La petite fonction `file_de` sert uniquement à fabriquer rapidement des files pour les tests.

On affiche une file sous la forme `tête < ... < queue` : on **défile à gauche** et on **enfile à droite**.

```python
from file import File   # la classe File de l'exercice 1 (fichier file.py placé à côté du notebook)


def file_de(*elements):
    """Outil pour les tests : construit une file contenant les éléments donnés (le premier en tête)."""
    f = File()
    for el in elements:
        f.enfile(el)
    return f


print(file_de("a", "b", "c"))
```

## Question 1 : `permute_prochains(f)`

On veut échanger les **deux premiers** éléments : `a b c d` → `b a c d`.

**Algorithme**, avec `n = f.taille()` :
1. on défile les deux premiers : `x = a`, `y = b` ; la file vaut `c d` ;
2. on les ré-enfile **dans l'ordre inverse** : `c d b a` ;
3. il faut maintenant ramener `b a` devant : on fait passer leur tour aux `n - 2` autres éléments (`c` et `d`) : `b a c d` ✔.

| Étape | File |
|:---|:---|
| départ | `a b c d` |
| `x = defile()`, `y = defile()` | `c d` |
| `enfile(y)`, `enfile(x)` | `c d b a` |
| 2 rotations | `b a c d` |

⚠️ **Cas particulier :** avec moins de 2 éléments, il n'y a rien à permuter (et le deuxième `defile` déclencherait une erreur). On arrête la fonction avec `return`.

```python
def permute_prochains(f):
    n = f.taille()
    if n < 2:
        return
    x = f.defile()
    y = f.defile()
    f.enfile(y)
    f.enfile(x)
    for _ in range(n - 2):          # les n - 2 autres éléments repassent derrière
        f.enfile(f.defile())


f = file_de("a", "b", "c", "d")
permute_prochains(f)
print(f)
assert repr(f) == repr(file_de("b", "a", "c", "d"))

g = file_de(1, 2)
permute_prochains(g)
assert repr(g) == repr(file_de(2, 1))
h = file_de(1)
permute_prochains(h)
assert repr(h) == repr(file_de(1))
```

## Question 2 : `permute_derniers(f)`

On veut échanger les **deux derniers** éléments : `a b c d` → `a b d c`.

**Idée :** on amène d'abord les deux derniers **en tête**, puis on fait comme à la question 1, mais **dans l'autre ordre** :
1. `n - 2` rotations : les deux derniers arrivent en tête : `c d a b` ;
2. on défile `x = c` puis `y = d`, et on ré-enfile `y` puis `x` : `a b d c` ✔.

Cette fois, plus besoin de rotation finale : les éléments `a b` sont déjà revenus devant.

| Étape | File |
|:---|:---|
| départ | `a b c d` |
| 2 rotations | `c d a b` |
| `x = defile()`, `y = defile()` | `a b` |
| `enfile(y)`, `enfile(x)` | `a b d c` |

💡 **Vérifier sur les cas limites :** avec exactement 2 éléments, on fait 0 rotation et on retrouve le comportement de `permute_prochains` : c'est logique, les deux prochains sont aussi les deux derniers.

```python
def permute_derniers(f):
    n = f.taille()
    if n < 2:
        return
    for _ in range(n - 2):          # les deux derniers arrivent en tête
        f.enfile(f.defile())
    x = f.defile()
    y = f.defile()
    f.enfile(y)
    f.enfile(x)


f = file_de("a", "b", "c", "d")
permute_derniers(f)
print(f)
assert repr(f) == repr(file_de("a", "b", "d", "c"))

g = file_de(1, 2)
permute_derniers(g)
assert repr(g) == repr(file_de(2, 1))
h = File()
permute_derniers(h)
assert h.est_vide()
print("Tous les tests passent ✔")
```

⚠️ **Erreur fréquente :** écrire une fonction qui **renvoie** une nouvelle file au lieu de modifier celle reçue. L'appel `permute_derniers(f)` ne changerait alors rien à `f`. Ici, l'énoncé demande une modification **en place**.
