## Exercice 1 — Factorielle et puissance  ⭐

### Question 1

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `factorielle(n)` qui renvoie $n! = 1 \times 2 \times \cdots \times n$, avec $0! = 1$.

#### ✅ RÉSOLUTION

````python
def factorielle(n):
    if n == 0:
        return 1
    return n * factorielle(n - 1)
````

*Les tests :*

````python
assert factorielle(0) == 1
assert factorielle(1) == 1
assert factorielle(5) == 120
assert factorielle(10) == 3628800
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Cas de base :** `factorielle(0)` vaut `1`.
- **Cas récursif :** $n! = n \times (n-1)!$.
- **Erreur fréquente :** prendre `return 0` comme cas de base : toutes les multiplications donneraient alors `0`.

### Question 2

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `puissance(x, n)` qui renvoie $x^n$ (pour `n` entier positif ou nul), **sans** utiliser `**`.
>
> *Indice : $x^0 = 1$ et $x^n = x \times x^{n-1}$.*

#### ✅ RÉSOLUTION

````python
def puissance(x, n):
    if n == 0:
        return 1
    return x * puissance(x, n - 1)
````

*Les tests :*

````python
assert puissance(2, 0) == 1
assert puissance(2, 10) == 1024
assert puissance(5, 3) == 125
assert puissance(-3, 3) == -27
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Même schéma que la factorielle : on multiplie par `x` à chaque appel, et on s'arrête à `n == 0` où la réponse est `1`.
- Cette version fait `n` multiplications. On peut faire **beaucoup** mieux avec l'idée $x^n = x^{n/2} \times x^{n/2}$ quand `n` est pair : c'est l'**exponentiation rapide**, un des premiers exemples de « diviser pour régner » (pour $n = 100$ : 8 multiplications au lieu de 100).

## Exercice 2 — Maximum d'un tableau  ⭐⭐

### Question 1

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `maximum(t, i)` qui renvoie le plus grand élément de la liste non vide `t` à partir de l'indice `i`. On l'appelle avec `maximum(t, 0)` pour toute la liste.
>
> *Indice : quel est le cas de base, quand il ne reste qu'un seul élément ?*

#### ✅ RÉSOLUTION

````python
def maximum(t, i):
    if i == len(t) - 1:             # il ne reste qu'un élément : le dernier
        return t[i]
    m = maximum(t, i + 1)           # le plus grand de la suite de la liste
    if t[i] > m:
        return t[i]
    return m
````

*Les tests :*

````python
assert maximum([3, 8, 2, 5], 0) == 8
assert maximum([7], 0) == 7
assert maximum([-5, -2, -9], 0) == -2
assert maximum([1, 2, 3, 4], 0) == 4
assert maximum([3, 8, 2, 5], 2) == 5          # à partir de l'indice 2 : [2, 5]
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Cas de base :** quand `i` est l'indice du **dernier** élément, il ne reste que lui : c'est le maximum. (On ne peut pas prendre `i == len(t)` comme pour la somme : le maximum d'une liste vide n'a pas de sens.)
- **Cas récursif :** on fait confiance à l'appel récursif pour donner le maximum `m` de la suite de la liste, puis on le compare à `t[i]`.
- On range le résultat dans une variable `m` pour ne faire l'appel récursif **qu'une seule fois**. Écrire `if t[i] > maximum(t, i + 1): ... else: return maximum(t, i + 1)` le ferait deux fois, et le nombre d'appels doublerait à chaque niveau !

## Exercice 3 — Couper le tableau en deux : somme et maximum  ⭐⭐

On reprend les fonctions `partie_gauche` et `partie_droite` de la question 2 du polycopié (exécuter la cellule ci-dessous) :
- `partie_gauche(t)` renvoie les `n//2` premiers éléments de `t` ;
- `partie_droite(t)` renvoie les `n - n//2` derniers.

Par exemple, avec `t = [3, 1, 4, 2, 5]` : `partie_gauche(t)` vaut `[3, 1]` et `partie_droite(t)` vaut `[4, 2, 5]`.

Dans cet exercice, on **coupe le tableau en deux**, comme dans le tri fusion. Toutes les fonctions suivent le même plan en 4 étapes :
1. **cas de base :** un tableau d'un seul élément ;
2. **couper** le tableau en deux avec `partie_gauche` et `partie_droite` ;
3. **deux appels récursifs**, un sur chaque partie, dont on **range les résultats dans deux variables** `g` et `d` ;
4. **combiner** `g` et `d` pour obtenir le résultat.

````python
def partie_gauche(t):
    n = len(t)
    res = []
    for i in range(n // 2):
        res.append(t[i])
    return res


def partie_droite(t):
    n = len(t)
    res = []
    for i in range(n // 2, n):
        res.append(t[i])
    return res
````

### Question 1

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `somme_dpr(t)` qui renvoie la somme des éléments d'un tableau **non vide** `t`, en suivant le plan en 4 étapes. Ne pas utiliser `sum`.
>
> *Indice : la somme d'un tableau, c'est la somme de sa partie gauche plus la somme de sa partie droite.*

#### ✅ RÉSOLUTION

````python
def somme_dpr(t):
    if len(t) == 1:                     # 1. cas de base : un seul élément
        return t[0]
    g = somme_dpr(partie_gauche(t))     # 2. et 3. on coupe, et on fait un appel
    d = somme_dpr(partie_droite(t))     #          sur chaque partie
    return g + d                        # 4. on combine
````

*Les tests :*

````python
assert somme_dpr([5]) == 5
assert somme_dpr([3, 1, 4, 2]) == 10
assert somme_dpr([3, 1, 4, 2, 5]) == 15
assert somme_dpr([10, -3, 7]) == 14
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Cas de base :** un tableau d'un seul élément a pour somme cet élément.
- **On fait confiance aux appels récursifs :** `g` est la somme de la partie gauche et `d` celle de la partie droite, sans se demander comment elles sont calculées. Il reste juste à les additionner.
- Pourquoi le tableau doit-il être **non vide** ? Avec `[]`, on n'atteint jamais le cas de base (`partie_droite([])` vaut encore `[]`). On pourrait ajouter le cas `len(t) == 0` (somme 0), mais l'énoncé ne le demande pas.
- C'est **exactement** le squelette de `tri_fusion_rec` : seule la dernière ligne change.

### Question 2 — sur papier ✍️

#### 📝 ÉNONCÉ

> **Sur papier**, dessiner l'arbre des appels de `somme_dpr([3, 1, 4, 2])` : sous chaque appel, écrire les deux appels qu'il fait (partie gauche à gauche, partie droite à droite). Puis écrire à côté de chaque appel la valeur qu'il **renvoie**, en remontant depuis le bas.
>
> Combien d'appels sont faits en tout (en comptant le premier) ?

#### ✅ RÉSOLUTION

```
                  somme_dpr([3, 1, 4, 2])   renvoie 4 + 6 = 10
                 /                       \
   somme_dpr([3, 1])  renvoie 4      somme_dpr([4, 2])  renvoie 6
      /          \                      /          \
somme_dpr([3])  somme_dpr([1])    somme_dpr([4])  somme_dpr([2])
  renvoie 3       renvoie 1         renvoie 4       renvoie 2
```
Il y a **7 appels** en tout. La valeur renvoyée par le premier appel est **10**.

````python
# Vérification : on affiche chaque appel et ce qu'il renvoie
def somme_dpr_trace(t, decalage):
    print(decalage + "somme_dpr(" + str(t) + ")")
    if len(t) == 1:
        print(decalage + "  renvoie " + str(t[0]))
        return t[0]
    g = somme_dpr_trace(partie_gauche(t), decalage + "    ")
    d = somme_dpr_trace(partie_droite(t), decalage + "    ")
    print(decalage + "  renvoie " + str(g) + " + " + str(d) + " = " + str(g + d))
    return g + d


somme_dpr_trace([3, 1, 4, 2], "")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **En descendant**, on coupe en deux jusqu'aux tableaux d'un seul élément (les cas de base). **En remontant**, on additionne.
- C'est le même schéma que la question 2.4 (a) du polycopié pour le tri fusion : en haut on coupe, en bas on combine. Pour le tri fusion, on fusionne au lieu d'additionner.

### Question 3

#### 📝 ÉNONCÉ

> Écrire de la même façon une fonction récursive `maximum_dpr(t)` qui renvoie le plus grand élément d'un tableau **non vide** `t`. Ne pas utiliser `max`.

#### ✅ RÉSOLUTION

````python
def maximum_dpr(t):
    if len(t) == 1:
        return t[0]
    g = maximum_dpr(partie_gauche(t))
    d = maximum_dpr(partie_droite(t))
    if g > d:
        return g
    else:
        return d
````

*Les tests :*

````python
assert maximum_dpr([5]) == 5
assert maximum_dpr([3, 8, 2, 5]) == 8
assert maximum_dpr([3, 1, 4, 2, 9]) == 9
assert maximum_dpr([7, 2, 3]) == 7
assert maximum_dpr([-4, -2, -9]) == -2
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Les 3 premières étapes sont **identiques** à `somme_dpr`. Seule la combinaison change : au lieu d'additionner `g` et `d`, on garde le plus grand des deux.
- Le maximum du tableau est forcément soit le maximum de la partie gauche, soit celui de la partie droite.

### Question 4 ✍️

#### 📝 ÉNONCÉ

> Un élève écrit la fonction suivante :
> ```python
> def maximum_dpr_eleve(t):
>     if len(t) == 1:
>         return t[0]
>     if maximum_dpr_eleve(partie_gauche(t)) > maximum_dpr_eleve(partie_droite(t)):
>         return maximum_dpr_eleve(partie_gauche(t))
>     else:
>         return maximum_dpr_eleve(partie_droite(t))
> ```
> Sa fonction renvoie toujours le bon résultat. Expliquer pourquoi c'est quand même une **mauvaise idée**, et comment la corriger.

#### ✅ RÉSOLUTION

La fonction **refait plusieurs fois le même calcul** : le maximum de la partie gauche (ou de la partie droite) est calculé une première fois dans le `if`, puis **une deuxième fois** dans le `return`. Chaque appel fait donc 3 appels récursifs au lieu de 2, et ce gaspillage se répète à tous les niveaux : pour un tableau de 8 éléments, il y a 40 appels au lieu de 15, et l'écart grandit très vite avec la taille.

**Correction :** faire chaque appel **une seule fois** et **ranger son résultat dans une variable** (`g` et `d`), puis utiliser les variables. C'est la fonction `maximum_dpr` de la question 3.

````python
# Vérification : on compte les appels des deux versions
compteur = [0]


def maximum_dpr_compte(t):
    compteur[0] = compteur[0] + 1
    if len(t) == 1:
        return t[0]
    g = maximum_dpr_compte(partie_gauche(t))
    d = maximum_dpr_compte(partie_droite(t))
    if g > d:
        return g
    else:
        return d


def maximum_dpr_eleve(t):
    compteur[0] = compteur[0] + 1
    if len(t) == 1:
        return t[0]
    if maximum_dpr_eleve(partie_gauche(t)) > maximum_dpr_eleve(partie_droite(t)):
        return maximum_dpr_eleve(partie_gauche(t))
    else:
        return maximum_dpr_eleve(partie_droite(t))


for taille in [8, 64, 512]:
    t = []
    for i in range(taille):
        t.append(i)
    compteur[0] = 0
    maximum_dpr_compte(t)
    bon = compteur[0]
    compteur[0] = 0
    maximum_dpr_eleve(t)
    print("tableau de", taille, "éléments :", bon, "appels avec les variables,", compteur[0], "appels sans")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- C'est **le même piège** que la « version maladroite » `power_dac` de l'exercice 3 du polycopié : `power_dac(x, n//2) * power_dac(x, n//2)` calcule deux fois exactement la même chose. La correction est la même : un seul appel, rangé dans une variable.
- Règle à retenir : **un appel récursif dont on a besoin deux fois, on le range dans une variable.**
