# Diviser pour régner — Exercices 4 et 5 : logarithme binaire et ordres de grandeur

> **Définition (cours)**
> Si $n = 2^k$, l'exposant $k$ est le **logarithme binaire** de $n$, noté $\log(n)$.
> C'est aussi le **nombre de chiffres, moins 1**, de l'écriture binaire de $n$.
> En toute rigueur, pour un réel $x > 0$ : $\log(x) = \dfrac{\ln(x)}{\ln(2)}$.

💡 **Intuition :** $\log(n)$ répond à la question « **combien de fois peut-on diviser $n$ par 2 avant d'arriver à 1 ?** ». C'est exactement le nombre d'étapes d'un algorithme qui coupe le problème en deux à chaque fois, comme la recherche dichotomique.

## Exercice 4

Méthode à la main : encadrer $n$ entre deux puissances de 2.

$2^0=1,\ 2^1=2,\ 2^2=4,\ 2^3=8,\ 2^4=16,\ 2^5=32,\ 2^6=64,\ 2^7=128,\ 2^8=256,\ 2^9=512,\ 2^{10}=1024,\ \ldots,\ 2^{16}=65\,536,\ \ldots,\ 2^{24}=16\,777\,216$

- $15$ : $8 < 15 < 16 = 2^4$, donc $\log(15) \approx 4$ (valeur exacte 3,91) ;
- $33$ : $32 = 2^5 < 33$, donc $\log(33) \approx 5$ (5,04) ;
- $128 = 2^7$, donc $\log(128) = 7$ exactement ;
- $1000$ : $1000 \approx 1024 = 2^{10}$, donc $\log(1000) \approx 10$ (9,97). **Retenir : $2^{10} \approx 10^3$** ;
- $60\,000$ : $60\,000 \approx 65\,536 = 2^{16}$, donc $\log(60\,000) \approx 16$ (15,87) ;
- $\log(n) = 24$ : $n = 2^{24} = 16\,777\,216$, soit environ **16 millions**.

| $n$ | 15 | 33 | 128 | 1000 | 60 000 | **16 777 216** |
|:---|:---|:---|:---|:---|:---|:---|
| $\log(n)$ | **≈ 4** | **≈ 5** | **7** | **≈ 10** | **≈ 16** | 24 |

On vérifie avec Python (`math.log2`), puis avec le **nombre de chiffres en binaire** (`bin(n)` donne l'écriture binaire, précédée de `"0b"`) :

```python
from math import log2

for n in [15, 33, 128, 1000, 60000, 2**24]:
    ecriture = bin(n)[2:]
    print(f"n = {n:>9} | log2(n) = {log2(n):6.2f} | binaire : {ecriture:>25} ({len(ecriture)} chiffres)")
```

⚠️ **Remarque sur la définition du cours :** pour un $n$ qui n'est pas une puissance de 2, le polycopié propose de prendre le **nombre de chiffres** de l'écriture binaire. Cela donne une valeur **arrondie au-dessus** : 4 pour 15, 6 pour 33, 10 pour 1000, 16 pour 60 000. Selon qu'on arrondit au plus proche (avec $\ln(x)/\ln(2)$) ou au-dessus (avec le nombre de chiffres), on peut trouver 5 ou 6 pour $n = 33$. Les deux réponses sont acceptables : ce qui compte ici, c'est **l'ordre de grandeur**.

## Exercice 5

```python
puissances = [1, 2, 3, 4, 6, 9]
print(f"{'n':>10} | {'log(n)':>7} | {'n log(n)':>10} | {'n²':>8}")
for p in puissances:
    n = 10**p
    print(f"{'10^' + str(p):>10} | {log2(n):7.1f} | {n * log2(n):10.1e} | {float(n * n):8.0e}")
```

| $n$ | 10 | 100 | 1000 | $10^4$ | $10^6$ | $10^9$ |
|:---|:---|:---|:---|:---|:---|:---|
| $\log(n)$ | ≈ 3,3 | ≈ 6,6 | ≈ 10 | ≈ 13 | ≈ 20 | ≈ 30 |
| $n$ | 10 | 100 | 1000 | $10^4$ | $10^6$ | $10^9$ |
| $n\log(n)$ | ≈ 33 | ≈ 660 | ≈ $10^4$ | ≈ $1{,}3\times10^5$ | ≈ $2\times10^7$ | ≈ $3\times10^{10}$ |
| $n^2$ | 100 | $10^4$ | $10^6$ | $10^8$ | $10^{12}$ | $10^{18}$ |

💡 **Astuce de calcul :** comme $2^{10} \approx 10^3$, on a $\log(10^3) \approx 10$, $\log(10^6) \approx 20$ et $\log(10^9) \approx 30$.

### Comparaison des ordres de grandeur

Imaginons un ordinateur qui fait $10^9$ opérations par seconde, et $n = 10^9$ données :

```python
n = 10**9
vitesse = 10**9          # opérations par seconde
for nom, ops in [("log(n)", log2(n)), ("n", n), ("n log(n)", n * log2(n)), ("n²", n * n)]:
    secondes = ops / vitesse
    if secondes < 1:
        texte = f"{secondes * 1e6:.2f} microsecondes"
    elif secondes < 3600:
        texte = f"{secondes:.0f} seconde(s)"
    else:
        texte = f"{secondes / (3600 * 24 * 365):.0f} ans"
    print(f"{nom:>9} : {texte}")
```

**Conclusion :**
- $\log(n)$ ne grandit presque pas : on passe de $n = 1000$ à $n = 10^9$ (un million de fois plus) et $\log(n)$ ne fait que passer de 10 à 30 ;
- $n\log(n)$ reste proche de $n$ (au plus 30 fois plus en pratique) ;
- $n^2$ explose : pour $n = 10^9$, il faut **30 secondes** avec un algorithme en $n\log(n)$, mais **plus de 30 ans** avec un algorithme en $n^2$.

C'est tout l'intérêt de « diviser pour régner » : passer d'un algorithme en $n$ à un algorithme en $\log(n)$ (recherche dichotomique, exponentiation rapide), ou d'un algorithme en $n^2$ à un algorithme en $n\log(n)$ (tri fusion, plus courte distance).

> **Propositions du cours (à compléter)**
> - Le coût moyen de la **recherche dichotomique** dans un tableau de taille $n$ est **de l'ordre de $\log(n)$** comparaisons.
> - Le coût moyen du **tri fusion** d'un tableau de taille $n$ est **de l'ordre de $n\log(n)$** comparaisons.
>
> Vérification de $c_k = 2^k \times k$ : on a $c_0 = 0$ (un tableau de taille 1 est déjà trié) et, si $c_{k-1} = 2^{k-1}(k-1)$, alors $c_k = 2c_{k-1} + 2^k = 2^k(k-1) + 2^k = 2^k \times k$.
