> **Définition (cours)**
> Si un entier n est une puissance de 2, c'est-à-dire si $n = 2^k$ où $k \in \mathbb{N}$, alors l'exposant k est appelé logarithme (binaire) de n, noté $\log(n)$.
> C'est aussi le nombre de chiffres, moins 1, de l'écriture binaire de n.
> Si n n'est pas une puissance de 2, on peut ainsi définir son logarithme comme étant le nombre chiffres dans son écriture binaire.
> *Remarque :* en toute rigueur, $\log(x) = \dfrac{\ln(x)}{\ln(2)}$ pour un réel $x > 0$.

# EXERCICE 4

#### 📝 ÉNONCÉ

> Utiliser la définition pour compléter le tableau suivant avec des valeurs approchées entières.
>
> | n | 15 | 33 | 128 | 1000 | 60 000 | |
> |:---|:---|:---|:---|:---|:---|:---|
> | log(n) | | | | | | 24 |

#### ✅ RÉSOLUTION

| $n$ | 15 | 33 | 128 | 1000 | 60 000 | **16 777 216** |
|:---|:---|:---|:---|:---|:---|:---|
| $\log(n)$ | **≈ 4** | **≈ 5** | **7** | **≈ 10** | **≈ 16** | 24 |

```python
from math import log2

for n in [15, 33, 128, 1000, 60000, 2**24]:
    ecriture = bin(n)[2:]
    print("n =", n, "  log2(n) =", round(log2(n), 2), "  binaire :", ecriture, "  chiffres :", len(ecriture))
```

#### 💡 EXPLICATION PÉDAGOGIQUE

**Intuition :** $\log(n)$ répond à la question « **combien de fois peut-on diviser $n$ par 2 avant d'arriver à 1 ?** ». C'est exactement le nombre d'étapes d'un algorithme qui coupe le problème en deux à chaque fois, comme la recherche dichotomique.

**Méthode à la main :** encadrer $n$ entre deux puissances de 2.
$2^0=1,\ 2^1=2,\ 2^2=4,\ 2^3=8,\ 2^4=16,\ 2^5=32,\ 2^6=64,\ 2^7=128,\ 2^8=256,\ 2^9=512,\ 2^{10}=1024,\ \ldots,\ 2^{16}=65\,536,\ \ldots,\ 2^{24}=16\,777\,216$

- $15$ : $8 < 15 < 16 = 2^4$, donc $\log(15) \approx 4$ (valeur exacte 3,91) ;
- $33$ : $32 = 2^5 < 33$, donc $\log(33) \approx 5$ (5,04) ;
- $128 = 2^7$, donc $\log(128) = 7$ exactement ;
- $1000$ : $1000 \approx 1024 = 2^{10}$, donc $\log(1000) \approx 10$ (9,97). **Retenir : $2^{10} \approx 10^3$** ;
- $60\,000$ : $60\,000 \approx 65\,536 = 2^{16}$, donc $\log(60\,000) \approx 16$ (15,87) ;
- $\log(n) = 24$ : $n = 2^{24} = 16\,777\,216$, soit environ **16 millions**.

⚠️ **Remarque sur la définition du cours :** pour un $n$ qui n'est pas une puissance de 2, prendre le **nombre de chiffres** de l'écriture binaire donne une valeur **arrondie au-dessus** : 4 pour 15, 6 pour 33, 10 pour 1000, 16 pour 60 000. On peut donc trouver 5 ou 6 pour $n = 33$. Les deux réponses sont acceptables : ce qui compte ici, c'est **l'ordre de grandeur**.

La vérification Python utilise `math.log2` et `bin(n)` (écriture binaire, précédée de `"0b"`).

# EXERCICE 5

#### 📝 ÉNONCÉ

> Compléter le tableau suivant et comparer les ordres de grandeur des différents modes de croissance.
>
> | n | 10 | 100 | 1000 | 10⁴ | 10⁶ | 10⁹ |
> |:---|:---|:---|:---|:---|:---|:---|
> | log(n) | | | | | | |
> | n | | | | | | |
> | n log(n) | | | | | | |
> | n² | | | | | | |

#### ✅ RÉSOLUTION

| $n$ | 10 | 100 | 1000 | $10^4$ | $10^6$ | $10^9$ |
|:---|:---|:---|:---|:---|:---|:---|
| $\log(n)$ | ≈ 3,3 | ≈ 6,6 | ≈ 10 | ≈ 13 | ≈ 20 | ≈ 30 |
| $n$ | 10 | 100 | 1000 | $10^4$ | $10^6$ | $10^9$ |
| $n\log(n)$ | ≈ 33 | ≈ 660 | ≈ $10^4$ | ≈ $1{,}3\times10^5$ | ≈ $2\times10^7$ | ≈ $3\times10^{10}$ |
| $n^2$ | 100 | $10^4$ | $10^6$ | $10^8$ | $10^{12}$ | $10^{18}$ |

**Comparaison :** $\log(n)$ ne grandit presque pas ; $n\log(n)$ reste proche de $n$ (au plus 30 fois plus ici) ; $n^2$ explose.

```python
puissances = [1, 2, 3, 4, 6, 9]
print(f"{'n':>10} | {'log(n)':>7} | {'n log(n)':>10} | {'n²':>8}")
for p in puissances:
    n = 10**p
    print(f"{'10^' + str(p):>10} | {log2(n):7.1f} | {n * log2(n):10.1e} | {float(n * n):8.0e}")
```

*Illustration (non demandée) : temps de calcul pour $n = 10^9$ données, avec un ordinateur qui fait $10^9$ opérations par seconde.*

```python
n = 10**9
vitesse = 10**9          # opérations par seconde
for nom, ops in [("log(n)", log2(n)), ("n", n), ("n log(n)", n * log2(n)), ("n²", n * n)]:
    secondes = ops / vitesse
    if secondes < 1:
        texte = f"{secondes * 1e6:.2f} microsecondes"
    elif secondes < 3600:
        texte = f"{secondes:.0f} seconde(s)"
    else:
        texte = f"{secondes / (3600 * 24 * 365):.0f} ans"
    print(f"{nom:>9} : {texte}")
```

#### 💡 EXPLICATION PÉDAGOGIQUE

**Astuce de calcul :** comme $2^{10} \approx 10^3$, on a $\log(10^3) \approx 10$, $\log(10^6) \approx 20$ et $\log(10^9) \approx 30$.

**Ce qu'il faut retenir :**
- $\log(n)$ ne grandit presque pas : on passe de $n = 1000$ à $n = 10^9$ (un million de fois plus) et $\log(n)$ ne fait que passer de 10 à 30 ;
- pour $n = 10^9$, il faut **30 secondes** avec un algorithme en $n\log(n)$, mais **plus de 30 ans** avec un algorithme en $n^2$.

C'est tout l'intérêt de « diviser pour régner » : passer d'un algorithme en $n$ à un algorithme en $\log(n)$ (recherche dichotomique, exponentiation rapide), ou d'un algorithme en $n^2$ à un algorithme en $n\log(n)$ (tri fusion, plus courte distance).

# Les Propositions du cours (à compléter)

#### 📝 ÉNONCÉ

> **Proposition** — Le coût moyen, en nombre de comparaisons, de la recherche dichotomique dans un tableau de taille n est ……
> **Proposition** — Le coût moyen, en nombre de comparaisons, du tri fusion d'un tableau de taille n est ……
> *Élément de preuve :* on suppose que $n = 2^k$ et on note $c_k$ le coût du tri d'un tableau de taille $2^k$ : $c_k = 2c_{k-1} + 2^k$ (coût des appels récursifs + coût de la fusion), et on peut vérifier que l'on a bien $c_k = 2^k \times k = n\log(n)$.

#### ✅ RÉSOLUTION

- Recherche dichotomique : **de l'ordre de $\log(n)$** comparaisons.
- Tri fusion : **de l'ordre de $n\log(n)$** comparaisons.

#### 💡 EXPLICATION PÉDAGOGIQUE

Vérification de $c_k = 2^k \times k$ (par récurrence) : on a $c_0 = 0$ (un tableau de taille 1 est déjà trié) et, si $c_{k-1} = 2^{k-1}(k-1)$, alors $c_k = 2c_{k-1} + 2^k = 2^k(k-1) + 2^k = 2^k \times k$.
